# Box Plots

The **box plot** (box-and-whiskers plot) is a standardized way of displaying a distribution from five statistics: the **minimum**, the **first quartile**, the **median**, the **third quartile**, and the **maximum**. One glance reveals the center, the spread, the skew, and any outliers. This notebook constructs one step by step for the newspaper advertising data, explains the 1.5·IQR whisker rule, and shows how to read the full-dataset plot with its outliers. It is based on the thirteenth video of Lesson 3 (Describing Distributions).

## 1. Introduction

Quantiles gave us numbers; the box plot turns them into a picture. It compresses a whole dataset into five values arranged so that the *shape* of the distribution — where the bulk sits, which way it leans, how far the stragglers reach — is visible at a glance. Because the recipe is standardized, box plots are also the tool of choice for comparing several groups side by side.

After completing this notebook you will be able to:

- Compute the **five-number summary** (min, $Q_1$, median, $Q_3$, max) and the **interquartile range** $\mathrm{IQR} = Q_3 - Q_1$.
- Construct a box plot: box, median line, and whiskers with the **1.5·IQR rule**.
- Identify **outliers** as points beyond the whiskers.
- Read skewness and dispersion off a box plot, and know what a box plot *cannot* show.

## 2. Theory

### The ingredients

For the 12-point newspaper sample, divide the sorted data into four equal blocks and take the three quartiles (midpoints between straddling values, as in the last notebook):

$$
Q_1 = \frac{18.3 + 18.4}{2} = 18.35 \qquad
Q_2 = \frac{25.9 + 29.7}{2} = 27.8 \qquad
Q_3 = \frac{51.2 + 54.7}{2} = 52.95
$$

The **interquartile range** is where the middle 50% of the data lies:

$$
\mathrm{IQR} = Q_3 - Q_1 = 34.6
$$

Finally note the extremes: minimum $8.7$ and maximum $75$.

### Building the plot

1. **The box** spans the quartiles: from $Q_1 = 18.35$ to $Q_3 = 52.95$.
2. **The median line** is drawn across the box at $Q_2 = 27.8$.
3. **The whiskers** extend from the ends of the box out to the *fences*:
   $$Q_1 - 1.5 \cdot \mathrm{IQR} \qquad \text{and} \qquad Q_3 + 1.5 \cdot \mathrm{IQR}$$
   — but a whisker must **never go beyond the actual min or max** of the data. If the fence lies outside the data range (as happens here: $18.35 - 51.9 = -33.55$ is far below the minimum), simply cut the whisker at the min/max value.

### Reading a box plot

- **Skew:** compare the two halves of the box. Here $Q_3 - Q_2 = 25.15$ is much bigger than $Q_2 - Q_1 = 9.45$ — the data is right-skewed.
- **Outliers:** any point *outside* the whiskers is flagged as an outlier — too far from the center. In the 12-point sample both whiskers end at the min and max, so there are no outliers.
- **Dispersion:** the lengths of the box and whiskers show how spread out the data is (box = middle 50%, whiskers = the tails' reach).

### The full newspaper dataset

Using all 200 rows of the advertising data: $Q_1 = 12.75$, $Q_2 = 25.75$, $Q_3 = 45.1$, so $\mathrm{IQR} = 32.35$; minimum $0.3$ and maximum $114$.

- The **bottom** whisker runs from $Q_1$ down to the smallest value, 0.3 (the fence $12.75 - 48.525 < 0.3$ is below the data — cut).
- The **top** whisker ends at the fence $Q_3 + 1.5 \cdot \mathrm{IQR} = 45.1 + 48.525 = 93.625$ — because this time there *are* data points beyond it.
- The two points above 93.625 (newspaper budgets of about 100.9 and 114) fall outside the whisker: they are marked individually as **outliers**.

## 3. Worked Examples

### Example 1: the five-number summary of the sample

$$
(\min,\ Q_1,\ Q_2,\ Q_3,\ \max) = (8.7,\ 18.35,\ 27.8,\ 52.95,\ 75)
$$

Fences: $18.35 - 1.5(34.6) = -33.55$ and $52.95 + 1.5(34.6) = 104.85$. All 12 points lie inside $[-33.55, 104.85]$ → whiskers cut at 8.7 and 75, no outliers.

### Example 2: reading the skew

Box halves: median sits $9.45$ above $Q_1$ but $25.15$ below $Q_3$ — the right half of the middle 50% is stretched. Together with the long upper whisker ($75 - 52.95 = 22.05$ vs. $18.35 - 8.7 = 9.65$ below), the plot says: **right-skewed**, exactly what the skewness notebook would compute numerically.

### Example 3: when is a point an outlier?

Full dataset: a budget of `$100.9k` lies above the upper fence $93.625$ → outlier. A budget of `$93k` would not (inside the fence). The 1.5·IQR rule is a *convention* — a robust, quartile-based screen for "suspiciously far from the bulk", not a law of nature.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)

newspaper = np.array([8.7, 14.2, 18.3, 18.4, 23.2, 25.9,
                      29.7, 35.2, 51.2, 54.7, 65.9, 75.0])
```

### The five-number summary and the fences

```python
def five_number_summary(data):
    q1, q2, q3 = np.quantile(data, [0.25, 0.5, 0.75], method="midpoint")
    return data.min(), q1, q2, q3, data.max()

mn, q1, q2, q3, mx = five_number_summary(newspaper)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

print(f"five-number summary: min = {mn}, Q1 = {q1}, median = {q2:.1f}, Q3 = {q3}, max = {mx}")
print(f"IQR = Q3 - Q1 = {iqr:.1f}")
print(f"fences: [{lower_fence:.2f}, {upper_fence:.2f}]")
print(f"whiskers cut at the data range -> [{max(mn, lower_fence)}, {min(mx, upper_fence)}]")
outliers = newspaper[(newspaper < lower_fence) | (newspaper > upper_fence)]
print(f"outliers: {outliers if outliers.size else 'none'}")
```

### The box plot with matplotlib

`plt.boxplot` implements exactly this recipe (`whis=1.5` is the default). One convention note: matplotlib ends each whisker at the most extreme *data point* inside the fence, which for this sample coincides with the min and max.

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.boxplot(newspaper, vert=False, widths=0.5, patch_artist=True,
           boxprops=dict(facecolor="tab:orange", alpha=0.6),
           medianprops=dict(color="tab:red", linewidth=2))
for value, name in [(q1, "Q1\n18.35"), (q2, "median\n27.8"), (q3, "Q3\n52.95")]:
    ax.annotate(name, xy=(value, 1.32), ha="center", fontsize=10)
ax.plot(newspaper, np.full_like(newspaper, 0.6), "o", color="tab:blue",
        alpha=0.6, label="the 12 data points")
ax.set_title("Box plot of the 12-point newspaper sample (no outliers)")
ax.set_xlabel("newspaper ad budget ($1000s)")
ax.set_yticks([])
ax.grid(True, alpha=0.3, axis="x")
ax.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

### The full dataset, drawn from its summary statistics

We don't need the 200 raw values — matplotlib's `bxp` draws a box plot straight from precomputed statistics. Using the video's numbers, the top whisker stops at the fence 93.625 and the two budgets beyond it appear as outlier points.

```python
full_stats = [{
    "label": "newspaper (all 200 rows)",
    "whislo": 0.3,               # bottom whisker: cut at the data minimum
    "q1": 12.75,
    "med": 25.75,
    "q3": 45.1,
    "whishi": 93.625,            # top whisker: the fence Q3 + 1.5 * IQR
    "fliers": [100.9, 114.0],    # the two points beyond the fence
}]

iqr_full = 45.1 - 12.75
print(f"IQR = {iqr_full:.2f}, upper fence = 45.1 + 1.5 * {iqr_full:.2f} = "
      f"{45.1 + 1.5 * iqr_full:.3f}")

fig, ax = plt.subplots(figsize=(9, 3.5))
ax.bxp(full_stats, vert=False, patch_artist=True, widths=0.5,
       boxprops=dict(facecolor="tab:blue", alpha=0.5),
       medianprops=dict(color="tab:red", linewidth=2))
ax.annotate("2 outliers", xy=(107, 1), xytext=(95, 1.35),
            arrowprops=dict(arrowstyle="->"), fontsize=11)
ax.set_title("Full newspaper dataset: whisker stops at the fence, outliers marked")
ax.set_xlabel("newspaper ad budget ($1000s)")
ax.set_yticks([])
ax.grid(True, alpha=0.3, axis="x")
plt.tight_layout()
plt.show()
```

### Box plots reveal shape: three known distributions side by side

Symmetric data gives a centered median and equal whiskers; skewed data pushes the median toward one end of the box and stretches the opposite whisker (plus a trail of flagged points in the long tail).

```python
rng = np.random.default_rng(42)

samples = [rng.normal(50, 10, size=2000),
           rng.uniform(20, 80, size=2000),
           rng.exponential(15, size=2000)]
labels = ["normal\n(symmetric)", "uniform\n(no tails)", "exponential\n(right-skewed)"]

fig, ax = plt.subplots(figsize=(9, 5))
ax.boxplot(samples, tick_labels=labels, patch_artist=True,
           boxprops=dict(facecolor="tab:blue", alpha=0.5),
           medianprops=dict(color="tab:red", linewidth=2))
ax.set_title("Reading shape from a box plot")
ax.set_ylabel("value")
ax.grid(True, alpha=0.3, axis="y")
plt.tight_layout()
plt.show()

print("Normal: centered median, symmetric whiskers, few mild outliers.")
print("Uniform: centered median, NO outliers ever (no tails to flag).")
print("Exponential: median low in the box, long upper whisker, many flagged points.")
```

## 5. Visualizations

### Anatomy of the box plot, fully annotated

Every part of the construction labeled on the newspaper sample: box, median, IQR, fences, and the cut whiskers.

```python
fig, ax = plt.subplots(figsize=(11, 5))

# the box
ax.fill_between([q1, q3], 0.8, 1.2, color="tab:orange", alpha=0.5)
ax.plot([q1, q1], [0.8, 1.2], color="black")
ax.plot([q3, q3], [0.8, 1.2], color="black")
ax.plot([q2, q2], [0.8, 1.2], color="tab:red", linewidth=3)

# whiskers (cut at min / max)
ax.plot([mn, q1], [1, 1], color="black")
ax.plot([q3, mx], [1, 1], color="black")
ax.plot([mn, mn], [0.93, 1.07], color="black", linewidth=2)
ax.plot([mx, mx], [0.93, 1.07], color="black", linewidth=2)

# the fences that were cut
for fence, label in [(lower_fence, "lower fence\nQ1 - 1.5·IQR = -33.55"),
                     (upper_fence, "upper fence\nQ3 + 1.5·IQR = 104.85")]:
    ax.axvline(fence, color="gray", linestyle=":", linewidth=1.5)
    ax.annotate(label, xy=(fence, 0.62), ha="center", fontsize=9, color="gray")

ax.annotate("min 8.7\n(whisker cut here)", xy=(mn, 1.07), xytext=(mn - 4, 1.3),
            ha="center", fontsize=9, arrowprops=dict(arrowstyle="->"))
ax.annotate("max 75\n(whisker cut here)", xy=(mx, 1.07), xytext=(mx + 4, 1.3),
            ha="center", fontsize=9, arrowprops=dict(arrowstyle="->"))
ax.annotate("Q1 = 18.35", xy=(q1, 0.8), xytext=(q1, 0.68), ha="center", fontsize=10)
ax.annotate("median = 27.8", xy=(q2, 1.2), xytext=(q2, 1.42), ha="center",
            fontsize=10, color="tab:red")
ax.annotate("Q3 = 52.95", xy=(q3, 0.8), xytext=(q3, 0.68), ha="center", fontsize=10)
ax.annotate("IQR = 34.6 (middle 50% of the data)", xy=((q1 + q3) / 2, 0.87),
            ha="center", fontsize=10)

ax.set_xlim(-45, 118)
ax.set_ylim(0.45, 1.6)
ax.set_yticks([])
ax.set_xlabel("newspaper ad budget ($1000s)")
ax.set_title("Anatomy of a box plot: box, median, whiskers, and the fences that got cut")
ax.grid(True, alpha=0.3, axis="x")
plt.tight_layout()
plt.show()
```

### What a box plot cannot show

A warning to close: the five-number summary can hide structure. A bimodal dataset (two clusters) gets a box plot that looks like any ordinary wide, single-peaked one: nothing in the box or whiskers reveals the empty gap in the middle — always pair box plots with a histogram when the shape matters.

```python
rng = np.random.default_rng(42)

bimodal = np.concatenate([rng.normal(30, 4, size=1000), rng.normal(70, 4, size=1000)])
unimodal = rng.normal(50, 16.5, size=2000)

fig, axes = plt.subplots(2, 2, figsize=(11, 5.5), sharex="col",
                         gridspec_kw={"width_ratios": [2, 1]})
for row, (data, name) in enumerate([(bimodal, "bimodal"), (unimodal, "unimodal")]):
    axes[row, 0].hist(data, bins=60, color="tab:blue", alpha=0.7)
    axes[row, 0].set_title(f"{name}: histogram shows the truth")
    axes[row, 0].set_ylabel("count")
    axes[row, 1].boxplot(data, vert=False, widths=0.5)
    axes[row, 1].set_title(f"{name}: box plot (no sign of two clusters)")
    axes[row, 1].set_yticks([])
    for ax in axes[row]:
        ax.grid(True, alpha=0.3)
axes[1, 0].set_xlabel("value")
axes[1, 1].set_xlabel("value")
plt.tight_layout()
plt.show()
```

## 7. Solutions

```python
# Exercise 1: fences of the small dataset
small = np.array([3, 5, 7, 8, 12, 13, 14, 21])
q1s, q3s = np.quantile(small, [0.25, 0.75], method="midpoint")
iqr_s = q3s - q1s
lo, hi = q1s - 1.5 * iqr_s, q3s + 1.5 * iqr_s
print(f"Exercise 1: IQR = {iqr_s}, fences = [{lo}, {hi}]")
print(f"All points inside -> whiskers cut at min {small.min()} and max {small.max()};")
print("no outliers (21 < 24.75).")
```

```python
# Exercise 2: replace 21 with 35
small2 = np.array([3, 5, 7, 8, 12, 13, 14, 35])
q1s, q3s = np.quantile(small2, [0.25, 0.75], method="midpoint")
lo, hi = q1s - 1.5 * (q3s - q1s), q3s + 1.5 * (q3s - q1s)
print(f"Exercise 2: quartiles unchanged (Q1 = {q1s}, Q3 = {q3s}) -- order-based!")
print(f"fences = [{lo}, {hi}]; 35 > {hi} -> 35 IS an outlier now.")
```

```python
# Exercise 3: reading skew from summary values
print("Exercise 3: right-skewed. Evidence:")
print("  1) within the box: Q3 - median = 16 vs median - Q1 = 4 -- the upper")
print("     half of the middle 50% is stretched 4x;")
print("  2) whiskers: upper reaches 30 above Q3, lower only 10 below Q1.")
```

```python
# Exercise 4: where does matplotlib end the whisker?
rng = np.random.default_rng(0)
x = rng.exponential(15, size=200)

q1e, q3e = np.quantile(x, [0.25, 0.75])
fence = q3e + 1.5 * (q3e - q1e)
inside = x[x <= fence]
print(f"Exercise 4: upper fence = {fence:.2f}")
print(f"largest data point inside the fence = {inside.max():.2f}")
print(f"outliers beyond the fence: {np.sort(x[x > fence]).round(1)}")
print("matplotlib draws the whisker cap at the largest INSIDE point, slightly")
print("short of the fence -- the video's drawing at the fence itself is the")
print("other common convention. Same box, same outliers either way.")
```

```python
# Exercise 5: how much does 1.5*IQR flag on a NORMAL distribution?
rng = np.random.default_rng(0)
z = rng.standard_normal(2_000_000)

q1n, q3n = np.quantile(z, [0.25, 0.75])
iqr_n = q3n - q1n
flagged = np.mean((z < q1n - 1.5 * iqr_n) | (z > q3n + 1.5 * iqr_n))
print(f"Exercise 5: fences at +-{q3n + 1.5 * iqr_n:.3f} sigma (theory 2.698)")
print(f"fraction flagged: {flagged:.5f}  (theory ~0.7%)")
print("Even perfectly clean normal data flags ~0.7% of points: an 'outlier' is a")
print("point far from the bulk, NOT necessarily an error -- it deserves a look,")
print("not automatic deletion.")
```

```python
# Exercise 6: three channels side by side
tv = np.array([7.8, 66.1, 44.5, 100.4, 151.5, 180.8,
               199.8, 214.7, 218.4, 228.3, 240.1, 293.6])
radio = np.array([2.6, 5.7, 8.6, 10.8, 16.9, 24.6,
                  26.7, 33.5, 36.9, 39.6, 41.3, 45.9])

fig, ax = plt.subplots(figsize=(9, 5))
ax.boxplot([newspaper, tv, radio], tick_labels=["newspaper", "TV", "radio"],
           patch_artist=True,
           boxprops=dict(facecolor="tab:blue", alpha=0.5),
           medianprops=dict(color="tab:red", linewidth=2))
ax.set_title("Ad budgets by channel (12-point samples)")
ax.set_ylabel("budget ($1000s)")
ax.grid(True, alpha=0.3, axis="y")
plt.tight_layout()
plt.show()

print("Exercise 6: TV budgets are by far the largest (median ~190 vs ~28 and ~26).")
print("Radio is the most symmetric (median near the box center, similar whiskers);")
print("newspaper leans right. No points fall beyond any fence -> no outliers.")
```

## 8. Conclusion

- A box plot displays the **five-number summary**: min, $Q_1$, median, $Q_3$, max — for the newspaper sample $(8.7, 18.35, 27.8, 52.95, 75)$ with $\mathrm{IQR} = 34.6$.
- Construction: **box** from $Q_1$ to $Q_3$, **median line** inside, **whiskers** out to $Q_1 - 1.5\,\mathrm{IQR}$ and $Q_3 + 1.5\,\mathrm{IQR}$ — but never beyond the actual min/max (cut them there).
- Points beyond the whiskers are flagged as **outliers**: in the full 200-row dataset the upper whisker stops at the fence $93.625$ and the two budgets above it (≈ 100.9 and 114) are drawn individually.
- Reading the plot: an off-center median and unequal whiskers reveal **skew** ($Q_3 - Q_2 = 25.15 \gg Q_2 - Q_1 = 9.45$ here → right-skewed); box and whisker lengths show **dispersion**; the quartile basis makes everything **outlier-robust**.
- Limits: the five numbers can hide multimodality — pair box plots with histograms when shape matters, and remember flagged outliers are "worth a look," not automatically errors.

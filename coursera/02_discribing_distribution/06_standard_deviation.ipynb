# The Standard Deviation

The variance measures spread well — but in the *wrong units*: if heights are in meters, their variance is in meters *squared*, which measures area. The fix is simply to take the square root, giving the **standard deviation** $\mathrm{std}(X) = \sqrt{\mathrm{Var}(X)}$ — a spread measure in the same units as the data. For the normal distribution the standard deviation earns its keep through the famous **68–95–99.7 rule**, and even has a visual signature: $\sigma$ is the distance from the center to where the bell changes concavity. It is based on the sixth video of Lesson 3 (Describing Distributions).

## 1. Introduction

You now have two numbers describing a distribution: the mean (center) and the variance (spread). But try to *report* a variance: "the average height is 1.70 meters, with a variance of 0.01 square meters." Square meters?! Spread stated as an area is hard to interpret and impossible to compare directly with the values themselves.

Why this matters:

- A spread measure in the **same units** as the data can be laid on the same axis: "1.70 m ± 0.10 m" is instantly meaningful.
- The standard deviation is the $\sigma$ in the normal distribution $N(\mu, \sigma^2)$ you met last week — the parameter that controls the width of the bell.
- "Within one/two/three standard deviations of the mean" is the everyday vocabulary of statistics, from lab reports to opinion polls.

After completing this notebook you will be able to:

- Explain the units problem of the variance and how the square root fixes it.
- Compute $\mathrm{std}(X) = \sqrt{\mathrm{Var}(X)}$ for discrete and continuous distributions.
- Locate $\sigma$ on a normal curve visually (the concavity change) and state the **68–95–99.7 rule** with its finer strip-by-strip breakdown.
- Use the scaling rule $\mathrm{std}(aX + b) = |a|\,\mathrm{std}(X)$.

## 2. Theory

### The units problem

Recall both forms of the variance:

$$
\mathrm{Var}(X) = E\big[(X - \mu)^2\big] = E[X^2] - E[X]^2
$$

Track the units. Say $X$ is measured in meters (heights, lengths):

- $E[X]$ is in **meters** — the average height, same units as the variable. Good.
- $E[X^2]$ is in **meters²**, and so is $E[X]^2$ — so $\mathrm{Var}(X)$ is in **meters²**. That is a unit of *area*, not length. Not so useful for describing heights!

### The fix: take the square root

$$
\mathrm{std}(X) = \sqrt{\mathrm{Var}(X)}
$$

The **standard deviation** is measured in meters again — the same units as the distribution itself. It is the standard way to report spread. (The common symbol is $\sigma$, with the variance written $\sigma^2$.)

One consequence worth noting: since $\mathrm{Var}(aX+b) = a^2\,\mathrm{Var}(X)$, taking square roots gives

$$
\mathrm{std}(aX + b) = |a|\,\mathrm{std}(X)
$$

— the standard deviation scales *linearly* with $|a|$, which is exactly what a same-units measure should do (double all the values in a dataset, and its spread in those units doubles).

### The standard deviation on the normal distribution

The normal distribution's PDF, from last week:

$$
f_X(x) = \frac{1}{\sqrt{2\pi}\,\sigma} \, e^{-\frac12 \frac{(x-\mu)^2}{\sigma^2}}
$$

with $\mu$ the center of the bell and $\sigma$ the spread — and $\sigma$ is precisely the standard deviation.

**Visual cue:** $\sigma$ is where the bell **changes concavity**. Walking away from the peak, the curve first bends downward (concave down) and then switches to bending upward (concave up); the switch happens exactly at $\mu - \sigma$ and $\mu + \sigma$.

### The 68–95–99.7 rule

For a normal distribution, the area under the curve within $k$ standard deviations of the mean is always the same, whatever $\mu$ and $\sigma$ are:

| interval | area (approx.) |
|---|---|
| $\mu \pm 1\sigma$ | **68%** |
| $\mu \pm 2\sigma$ | **95%** |
| $\mu \pm 3\sigma$ | **99.7%** |

The finer, strip-by-strip breakdown (per side):

- within $\pm 1\sigma$: **68.2%** in the central band;
- between $1\sigma$ and $2\sigma$: **13.5%** in each strip;
- between $2\sigma$ and $3\sigma$: **2.35%** in each strip;
- beyond $3\sigma$: **0.15%** in each tail — the tails extend infinitely, but carry almost nothing.

In statistics it is very common to describe values as being "within one, two, or three standard deviations of the mean" — for normal data that immediately translates into how usual or unusual a value is.

## 3. Worked Examples

### Example 1: the two coin games, in dollars

From the last notebook: the ±\$1 game has variance 1 (dollars²) and the ±\$100 game has variance 10,000 (dollars²). Their standard deviations:

$$
\mathrm{std}(X_1) = \sqrt{1} = \$1, \qquad \mathrm{std}(X_2) = \sqrt{10{,}000} = \$100
$$

Now the spread measure matches the stakes exactly — \$1 of typical swing versus \$100 — instead of the hard-to-read 1 vs 10,000.

### Example 2: the fair die

$\mathrm{Var}(X) = \tfrac{35}{12} \approx 2.917$, so

$$
\mathrm{std}(X) = \sqrt{35/12} \approx 1.71 \text{ pips}
$$

— a typical roll lands about 1.7 away from the mean of 3.5. And for the game $Y = 2X - 5$: $\mathrm{std}(Y) = |2| \cdot 1.71 \approx \$3.42$ (not $4\times$ — that factor belongs to the variance).

### Example 3: heights with the 68–95–99.7 rule

Say adult heights are approximately normal with $\mu = 1.70$ m and $\sigma = 0.10$ m. Then:

- about **68%** of people are between $1.60$ and $1.80$ m ($\mu \pm 1\sigma$);
- about **95%** are between $1.50$ and $1.90$ m ($\mu \pm 2\sigma$);
- about **99.7%** are between $1.40$ and $2.00$ m ($\mu \pm 3\sigma$).

A 2.00 m person sits at $+3\sigma$ — taller than roughly 99.85% of this population (all but the 0.15% right tail).

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)
```

### From variance to standard deviation

```python
def variance(values, pmf):
    mean = np.sum(values * pmf)
    return np.sum((values - mean)**2 * pmf)

half = np.array([0.5, 0.5])
sixth = np.full(6, 1 / 6)
faces = np.arange(1, 7)

for name, values, pmf in [("win/lose $1  ", np.array([-1, 1]), half),
                          ("win/lose $100", np.array([-100, 100]), half),
                          ("fair die     ", faces, sixth),
                          ("game 2X - 5  ", 2 * faces - 5, sixth)]:
    var = variance(values, pmf)
    print(f"{name}:  Var = {var:>10.4f}   std = {np.sqrt(var):>8.4f}")
```

### The scaling rule: std is linear in $|a|$, variance is quadratic

```python
rng = np.random.default_rng(42)
x = rng.integers(1, 7, size=100_000)

print(f"std(X)      = {np.std(x):.4f}   (theory sqrt(35/12) = {np.sqrt(35/12):.4f})")
print(f"std(2X)     = {np.std(2*x):.4f}   (theory 2 * 1.7078  = {2*np.sqrt(35/12):.4f})")
print(f"std(2X - 5) = {np.std(2*x - 5):.4f}   (the -5 changes nothing)")
print(f"std(-2X)    = {np.std(-2*x):.4f}   (|a| = 2: sign is irrelevant)")
```

### Verifying the 68–95–99.7 rule numerically

Integrate the normal PDF over $\mu \pm k\sigma$ with `np.trapezoid` — the exact values are 68.27%, 95.45%, 99.73%, whatever $\mu$ and $\sigma$ you pick.

```python
mu, sigma = 1.70, 0.10          # heights in meters

x = np.linspace(mu - 6 * sigma, mu + 6 * sigma, 20001)
pdf = np.exp(-0.5 * ((x - mu) / sigma)**2) / (sigma * np.sqrt(2 * np.pi))

for k in [1, 2, 3]:
    inside = (x >= mu - k * sigma) & (x <= mu + k * sigma)
    area = np.trapezoid(pdf[inside], x[inside])
    print(f"P(mu - {k}s < X < mu + {k}s) = {area:.4f}   "
          f"({mu - k*sigma:.2f} m to {mu + k*sigma:.2f} m)")
```

### The rule in action on simulated heights

Draw a large sample of normal heights and simply *count* how many land within each band.

```python
rng = np.random.default_rng(42)

mu, sigma = 1.70, 0.10
heights = rng.normal(mu, sigma, size=1_000_000)

for k in [1, 2, 3]:
    fraction = np.mean(np.abs(heights - mu) <= k * sigma)
    print(f"within {k} std of the mean: {fraction:.4f}")
print(f"\nsample mean = {heights.mean():.4f} m, sample std = {heights.std():.4f} m")
```

## 5. Visualizations

### The 68–95–99.7 rule, strip by strip

The video's final slide: the central band holds 68.2%, each 1–2$\sigma$ strip 13.5%, each 2–3$\sigma$ strip 2.35%, and each infinite tail just 0.15%.

```python
mu, sigma = 0, 1
x = np.linspace(-4, 4, 1000)
pdf = np.exp(-0.5 * x**2) / np.sqrt(2 * np.pi)

plt.figure(figsize=(11, 5.5))
plt.plot(x, pdf, color="black", linewidth=1.5)

bands = [(-1, 1, "tab:blue", "68.2%"),
         (-2, -1, "tab:cyan", "13.5%"), (1, 2, "tab:cyan", "13.5%"),
         (-3, -2, "tab:orange", "2.35%"), (2, 3, "tab:orange", "2.35%"),
         (-4, -3, "tab:red", "0.15%"), (3, 4, "tab:red", "0.15%")]
for lo, hi, color, label in bands:
    mask = (x >= lo) & (x <= hi)
    plt.fill_between(x[mask], pdf[mask], color=color, alpha=0.6)
    plt.annotate(label, xy=((lo + hi) / 2, 0.015), ha="center", fontsize=10,
                 rotation=90 if abs(lo) >= 2.5 else 0)

for k in range(-3, 4):
    plt.axvline(k, color="gray", linewidth=0.7, alpha=0.5)
ticks = range(-3, 4)
plt.xticks(list(ticks), [r"$\mu$" if k == 0 else rf"$\mu{k:+d}\sigma$" for k in ticks])
plt.title("The 68-95-99.7 rule: area under the normal curve, band by band")
plt.xlabel("value")
plt.ylabel("density")
plt.grid(True, alpha=0.2)
plt.show()
```

### Where is $\sigma$ on the bell? The concavity change

The curvature of the bell flips sign exactly at $\mu \pm \sigma$: concave *down* inside (curving over the peak), concave *up* outside (flattening into the tails).

```python
mu, sigma = 0, 1
x = np.linspace(-4, 4, 1000)
pdf = np.exp(-0.5 * ((x - mu) / sigma)**2) / (sigma * np.sqrt(2 * np.pi))

concave_down = np.abs(x - mu) < sigma    # second derivative < 0 here

plt.figure(figsize=(10, 5))
plt.plot(x[concave_down], pdf[concave_down], color="tab:red", linewidth=3,
         label="concave down (bending over the peak)")
plt.plot(x[~concave_down & (x < mu)], pdf[~concave_down & (x < mu)],
         color="tab:blue", linewidth=3, label="concave up (flattening tails)")
plt.plot(x[~concave_down & (x > mu)], pdf[~concave_down & (x > mu)],
         color="tab:blue", linewidth=3)
for k, lbl in [(-1, r"$\mu - \sigma$"), (1, r"$\mu + \sigma$")]:
    plt.axvline(k, color="black", linestyle=":", linewidth=1.5)
    plt.annotate(f"concavity changes\nat {lbl}", xy=(k, 0.242),
                 xytext=(k * 2.4, 0.30), ha="center",
                 arrowprops=dict(arrowstyle="->"), fontsize=10)
plt.title("A visual cue for the standard deviation: the inflection points of the bell")
plt.xlabel("value")
plt.ylabel("density")
plt.grid(True, alpha=0.3)
plt.legend(loc="upper left")
plt.show()
```

### Same mean, different $\sigma$ — same rule for all

Three bells centered at the same $\mu$: wider $\sigma$ stretches the curve, but $\mu \pm 2\sigma$ always captures 95% of each.

```python
x = np.linspace(-10, 10, 1000)

plt.figure(figsize=(10, 5))
for sigma, color in [(0.7, "tab:blue"), (1.5, "tab:orange"), (3.0, "tab:green")]:
    pdf = np.exp(-0.5 * (x / sigma)**2) / (sigma * np.sqrt(2 * np.pi))
    plt.plot(x, pdf, color=color, linewidth=2, label=f"$\\sigma$ = {sigma}")
    plt.fill_between(x, pdf, where=np.abs(x) <= 2 * sigma, color=color, alpha=0.15)
plt.title("Normal curves with different standard deviations (shaded: $\\mu \\pm 2\\sigma$, 95% each)")
plt.xlabel("value")
plt.ylabel("density")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
# Exercise 5: the rule is normal-specific -- uniform on [0, 1]
rng = np.random.default_rng(0)

sigma_uniform = np.sqrt(1 / 12)
u = rng.random(size=1_000_000)
within_1 = np.mean(np.abs(u - 0.5) <= 1 * sigma_uniform)
within_2 = np.mean(np.abs(u - 0.5) <= 2 * sigma_uniform)
print(f"Exercise 5: uniform sigma = sqrt(1/12) = {sigma_uniform:.4f}")
print(f"  within 1 sigma: {within_1:.4f}  (exact 2/sqrt(12) = {2/np.sqrt(12):.4f}, vs normal 0.682)")
print(f"  within 2 sigma: {within_2:.4f}  (2 sigma exceeds the interval edges -> everything, vs 0.954)")
print("Same sigma concept, different shape -> different areas. 68-95-99.7 is a")
print("property of the BELL, not of standard deviations in general.")
```

```python
# Exercise 6: the 50% interval of the standard normal
x = np.linspace(-6, 6, 200001)
pdf = np.exp(-0.5 * x**2) / np.sqrt(2 * np.pi)

zs = np.linspace(0.5, 0.9, 4001)
best_z = None
for z in zs:
    mask = np.abs(x) <= z
    if np.trapezoid(pdf[mask], x[mask]) >= 0.5:
        best_z = z
        break

mask = np.abs(x) <= best_z
print(f"Exercise 6: z = {best_z:.3f} gives area {np.trapezoid(pdf[mask], x[mask]):.4f}")
print("About 0.674 sigma on each side of the mean captures exactly half of a")
print("normal distribution -- the historical 'probable error'.")
```

## 8. Conclusion

- The variance has a units problem: for data in meters it is measured in **meters²** — an area, not a length.
- The **standard deviation** fixes it: $\mathrm{std}(X) = \sqrt{\mathrm{Var}(X)}$, a spread measure in the *same units* as the distribution (the ±\$1 and ±\$100 games get std \$1 and \$100; the die gets $\sqrt{35/12} \approx 1.71$).
- Consequence of the variance scaling rule: $\mathrm{std}(aX+b) = |a|\,\mathrm{std}(X)$ — linear, as a same-units measure should be.
- On the normal distribution the standard deviation is the parameter $\sigma$, visible as the distance from $\mu$ to the bell's **inflection points** (where concavity changes).
- The **68–95–99.7 rule**: about 68% of a normal distribution lies within $\mu \pm \sigma$, 95% within $\mu \pm 2\sigma$, 99.7% within $\mu \pm 3\sigma$ — in finer strips: 68.2% center, 13.5% per 1–2$\sigma$ strip, 2.35% per 2–3$\sigma$ strip, 0.15% per tail.
- "Within $k$ standard deviations of the mean" is standard statistical language for how typical or extreme a value is — for normal data, the rule converts it into probabilities instantly (but beware: other shapes give other areas).

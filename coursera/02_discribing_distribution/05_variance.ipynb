# Variance

The expected value tells you a lot about a distribution — but not the whole story. Two distributions can share the same mean while one is very narrow and the other very wide. The measure that captures this difference in **spread** is the **variance**. This notebook builds it up step by step (and shows why the naive "average deviation" fails), derives the famous shortcut formula $\mathrm{Var}(X) = E[X^2] - (E[X])^2$, and proves the scaling property $\mathrm{Var}(aX + b) = a^2\,\mathrm{Var}(X)$. It is based on the fifth video of Lesson 3 (Describing Distributions).

## 1. Introduction

Consider two coin-flip games. Game 1: heads wins **`$1`**, tails loses **`$1`**. Game 2: heads wins **`$100`**, tails loses **`$100`**. Both have an expected value of exactly **0** — play either over and over and on average you break even, so the fair price of both is zero. Yet the games are clearly very different: in the second, you are winning or losing a hundred dollars at a time.

The expected value cannot see this difference. To quantify the **spread of possible outcomes** you need a new measure — the variance.

Why this matters:

- Risk lives in the spread, not the mean: two investments with equal average return can carry wildly different risk.
- Variance is the second pillar of describing distributions (after the center) and feeds directly into the standard deviation, the normal distribution's $\sigma$, and much of statistics.

After completing this notebook you will be able to:

- Use the shortcut formula $\mathrm{Var}(X) = E[X^2] - (E[X])^2$ and reproduce its derivation.
- Apply the scaling rule $\mathrm{Var}(aX + b) = a^2\,\mathrm{Var}(X)$ — and explain why $b$ has no effect.

## 2. Theory

### Attempt 1: the deviation — and why its average is useless

The idea of spread: how far are the points from the expected value? The **deviation** of an outcome $x$ is

$$
x - E[X]
$$

For game 1 the deviations are $1 - 0 = 1$ and $-1 - 0 = -1$. Their average: $\frac{-1 + 1}{2} = 0$. No variation at all?!

The average deviation is **always zero**, for every random variable — the mean sits exactly where positive and negative deviations cancel (it is the balancing point!). Game 2's deviations, $+100$ and $-100$, also average to zero.

### Attempt 2: the absolute deviation

The problem was signs cancelling, so an intuitive fix is the **absolute deviation** $\lvert x - E[X] \rvert$ — now every deviation is positive and averaging makes sense. This works, but the absolute-value function has messy mathematical properties (it has a corner — no derivative at 0), so this approach is rarely used.

### The fix that stuck: square the deviations

The preferred method: **square** each deviation. Squares are always positive, like absolute values, but without the mathematical issues. The average of the squared deviations is the **variance**:

$$
\mathrm{Var}(X) = E\big[(X - E[X])^2\big]
$$

Reading the formula as four steps:

1. find the mean $E[X]$;
2. find each value's deviation from that mean;
3. **square** the deviations;
4. average the squared deviations.

Like every expectation, this is a **weighted average**: with equally likely outcomes you divide by the count; in general you weight each squared deviation by its probability $p(x)$:

$$
\mathrm{Var}(X) = \sum_x (x - E[X])^2 \, p(x)
$$

For the two games: game 1 has variance $\tfrac12(-1)^2 + \tfrac12(1)^2 = 1$; game 2 has variance $\tfrac12(-100)^2 + \tfrac12(100)^2 = 10{,}000$. A big number — but remember, it measures the average *squared* deviation, not the average deviation.

### Variance ignores the center

Compare two more games. Game 1: win `$2` / lose `$2` ($E = 0$). Game 2: win `$3` / lose `$1` ($E = \tfrac12(-1) + \tfrac12(3) = 1$). Different means — but in both games the two outcomes sit **2 away from their own mean**, so both variances equal $\tfrac12(2)^2 + \tfrac12(2)^2 = 4$. Variance measures spread *around the mean*, wherever that mean is.

### The shortcut formula

The definition is intuitive, but calculations are often easier with the equivalent form

$$
\mathrm{Var}(X) = E[X^2] - (E[X])^2
$$

Derivation — expand the square and use the linearity facts from the previous notebooks ($E[X+Y] = E[X]+E[Y]$; constants slide out of $E$; $E[X]$ is itself just a constant):

$$
\begin{aligned}
\mathrm{Var}(X) &= E\big[(X - E[X])^2\big] \\
  &= E\big[X^2 - 2\,E[X]\,X + E[X]^2\big] \\
  &= E[X^2] - E\big[2\,E[X]\,X\big] + E\big[E[X]^2\big] && \text{(expectations add)} \\
  &= E[X^2] - 2\,E[X]\,E[X] + E[X]^2 && \text{(constants move out; } E[\text{const}] = \text{const)} \\
  &= E[X^2] - 2\,E[X]^2 + E[X]^2 \\
  &= E[X^2] - (E[X])^2
\end{aligned}
$$

In words: variance is *the mean of the square minus the square of the mean* — exactly the gap we noticed at the end of the last notebook.

### Scaling: $\mathrm{Var}(aX + b) = a^2 \, \mathrm{Var}(X)$

Return to the dice game with net winnings $Y = 2X - 5$ (roll $X$, win double, pay `$5`): outcomes $-3, -1, 1, 3, 5, 7$, each with probability $\tfrac16$. We know $E[X] = 3.5$ and $E[Y] = 2$.

Look at what the transformation does to the *deviations*: an outcome that was $0.5$ away from the mean of $X$ is now $1$ away from the mean of $Y$; the largest deviation grows from $2.5$ to $5$. **Every deviation doubles** — and since variance averages *squared* deviations, the variance becomes $2^2 = 4$ times larger:

$$
\mathrm{Var}(2X - 5) = 4\,\mathrm{Var}(X)
$$

The $-5$ has no effect at all: adding a constant just slides the whole distribution (and its mean) sideways — it changes where the distribution is centered, not how spread out it is. In general:

$$
\mathrm{Var}(aX + b) = a^2\,\mathrm{Var}(X)
$$

Contrast with the mean, where $E[aX + b] = a\,E[X] + b$ — the mean feels both $a$ and $b$; the variance feels only $a$, and squared.

## 3. Worked Examples

### Example 1: the `$1` game and the `$100` game

Both games: $E[X] = \tfrac12(-w) + \tfrac12(w) = 0$ where $w$ is the stake.

$$
\mathrm{Var}(X_1) = \tfrac12(-1-0)^2 + \tfrac12(1-0)^2 = 1
\qquad
\mathrm{Var}(X_2) = \tfrac12(-100-0)^2 + \tfrac12(100-0)^2 = 10{,}000
$$

Same mean, spread differing by a factor of 10,000 (in squared dollars).

### Example 2: same variance, different centers

Game 1 (win `$2` / lose `$2`): $E[X_1] = 0$,

$$
\mathrm{Var}(X_1) = \tfrac12(-2 - 0)^2 + \tfrac12(2 - 0)^2 = 4
$$

Game 2 (win `$3` / lose `$1`): $E[X_2] = 1$,

$$
\mathrm{Var}(X_2) = \tfrac12(-1 - 1)^2 + \tfrac12(3 - 1)^2 = \tfrac12(4) + \tfrac12(4) = 4
$$

Different expected outcomes, identical spread.

### Example 3: variance of a fair die, both ways

$E[X] = 3.5$ and $E[X^2] = \tfrac{91}{6}$ (from the last notebooks). By the shortcut:

$$
\mathrm{Var}(X) = E[X^2] - (E[X])^2 = \tfrac{91}{6} - 3.5^2 = 15.1\overline{6} - 12.25 = \tfrac{35}{12} \approx 2.92
$$

(The definition gives the same: average of $(1-3.5)^2, \dots, (6-3.5)^2$.)

### Example 4: the dice game $Y = 2X - 5$

$$
\mathrm{Var}(Y) = \mathrm{Var}(2X - 5) = 2^2 \, \mathrm{Var}(X) = 4 \cdot \tfrac{35}{12} = \tfrac{35}{3} \approx 11.67
$$

No need to recompute deviations for the six net winnings — though doing so gives exactly the same number.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)
```

### A variance function, step by step

The four steps of the definition, as code — for any discrete distribution given by values and probabilities.

```python
def variance(values, pmf):
    mean = np.sum(values * pmf)                # step 1: the mean
    deviations = values - mean                 # step 2: deviations
    squared_deviations = deviations**2         # step 3: square them
    return np.sum(squared_deviations * pmf)    # step 4: weighted average

half = np.array([0.5, 0.5])

game_1dollar = np.array([-1, 1])
game_100dollar = np.array([-100, 100])
print(f"Var of the $1 game:   {variance(game_1dollar, half)}")
print(f"Var of the $100 game: {variance(game_100dollar, half)}")

game_pm2 = np.array([-2, 2])
game_win3_lose1 = np.array([-1, 3])
print(f"\nVar of win $2 / lose $2 (mean 0): {variance(game_pm2, half)}")
print(f"Var of win $3 / lose $1 (mean 1): {variance(game_win3_lose1, half)}")
```

### The average (unsquared) deviation really is always zero

```python
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)
mean = np.sum(faces * pmf)

avg_deviation = np.sum((faces - mean) * pmf)
avg_abs_deviation = np.sum(np.abs(faces - mean) * pmf)
avg_sq_deviation = np.sum((faces - mean)**2 * pmf)

print(f"Die roll, mean = {mean}")
print(f"Average deviation:          {avg_deviation:.10f}   <- always exactly 0")
print(f"Average absolute deviation: {avg_abs_deviation:.4f}   <- works, but rarely used")
print(f"Average squared deviation:  {avg_sq_deviation:.4f}   <- the variance (35/12)")
```

### The shortcut formula agrees with the definition

```python
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)

mean = np.sum(faces * pmf)
by_definition = np.sum((faces - mean)**2 * pmf)
by_shortcut = np.sum(faces**2 * pmf) - mean**2

print(f"Var(X) by definition E[(X - E[X])^2]:  {by_definition:.4f}")
print(f"Var(X) by shortcut  E[X^2] - (E[X])^2: {by_shortcut:.4f}")
print(f"Exact value: 35/12 = {35/12:.4f}")
```

### $\mathrm{Var}(aX + b) = a^2\,\mathrm{Var}(X)$ — simulation check

Scaling by 2 quadruples the variance; the $-5$ changes nothing. `np.var` computes exactly the average squared deviation of a sample.

```python
rng = np.random.default_rng(42)
x = rng.integers(1, 7, size=100_000)

print(f"Var(X)      = {np.var(x):.4f}    (theory 35/12  = {35/12:.4f})")
print(f"Var(2X)     = {np.var(2*x):.4f}   (theory 4*35/12 = {4*35/12:.4f})")
print(f"Var(2X - 5) = {np.var(2*x - 5):.4f}   (same: the -5 has no effect)")
print(f"Var(X + 100)= {np.var(x + 100):.4f}    (shifting changes nothing)")
```

## 5. Visualizations

### Same mean, different spread

The two motivating games on a shared (symlog) axis: both balance at `$0`, but the `$100` game's outcomes sit vastly farther from the pivot.

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 5.5), sharex=True)

for ax, w, name in [(axes[0], 1, "game 1: win/lose $1"),
                    (axes[1], 100, "game 2: win/lose $100")]:
    ax.axhline(0, color="black", linewidth=2)
    ax.plot([-w, w], [0.15, 0.15], "s", markersize=14, color="tab:blue")
    ax.plot(0, -0.18, "^", markersize=16, color="black")
    ax.set_title(f"{name}:  E = 0,  Var = {w**2:,}")
    ax.set_ylim(-0.5, 0.5)
    ax.set_yticks([])
    ax.grid(True, alpha=0.3, axis="x")

axes[1].set_xscale("symlog", linthresh=2)
axes[1].set_xlabel("outcome ($, symlog scale)")
plt.suptitle("Two games with the same expected value and very different variance")
plt.tight_layout()
plt.show()
```

### Why squaring: deviations, absolute deviations, squared deviations

For the fair die: raw deviations cancel to zero; absolute and squared deviations don't. The squared version (right) is what the variance averages.

```python
faces = np.arange(1, 7)
mean = 3.5
deviations = faces - mean

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, values, title in [
        (axes[0], deviations, "deviations  (average = 0)"),
        (axes[1], np.abs(deviations), "absolute deviations"),
        (axes[2], deviations**2, "squared deviations  (average = Var = 2.92)")]:
    colors = ["tab:red" if v < 0 else "tab:blue" for v in values]
    ax.bar(faces, values, color=colors, edgecolor="black", alpha=0.85)
    ax.axhline(values.mean(), color="black", linestyle="--", linewidth=1.5,
               label=f"average = {values.mean():.2f}")
    ax.set_title(title)
    ax.set_xlabel("die face")
    ax.grid(True, alpha=0.3)
    ax.legend()
axes[0].set_ylabel("value")
plt.suptitle("Three attempts at measuring spread (fair die, mean 3.5)")
plt.tight_layout()
plt.show()
```

### The dice game: deviations double, variance quadruples

Top: the die roll $X$, mean 3.5. Bottom: net winnings $Y = 2X - 5$, mean 2. Every outcome's distance to its mean doubles (e.g. $0.5 \to 1$, and the largest $2.5 \to 5$), so the average *squared* deviation grows by $2^2 = 4$.

```python
faces = np.arange(1, 7)
wins = 2 * faces - 5

fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
for ax, values, mean, name, color in [
        (axes[0], faces, 3.5, "die roll X:  Var = 35/12 = 2.92", "tab:red"),
        (axes[1], wins, 2.0, "net winnings Y = 2X - 5:  Var = 4 · 35/12 = 11.67", "tab:green")]:
    ax.axhline(0, color="black", linewidth=1.5)
    ax.vlines(values, 0, 1/6, color=color, linewidth=6)
    ax.axvline(mean, color="black", linestyle="--", linewidth=1.5)
    for v in values:
        ax.annotate("", xy=(v, 0.20), xytext=(mean, 0.20),
                    arrowprops=dict(arrowstyle="->", color="gray", lw=1))
    ax.set_title(f"{name}   (mean {mean}, dashed)")
    ax.set_ylim(0, 0.30)
    ax.set_yticks([0, 1/6])
    ax.set_yticklabels(["0", "1/6"])
    ax.grid(True, alpha=0.3, axis="x")
axes[1].set_xlabel("outcome ($)")
plt.suptitle("Var(aX + b) = a²·Var(X): scaling stretches every deviation, shifting moves none")
plt.tight_layout()
plt.show()
```

## 7. Solutions

```python
# Exercise 1: fair 0/1 coin
values = np.array([0, 1])
pmf = np.array([0.5, 0.5])
e_x = np.sum(values * pmf)
e_x2 = np.sum(values**2 * pmf)
print(f"Exercise 1: E[X] = {e_x}, E[X^2] = {e_x2}")
print(f"Var(X) = E[X^2] - (E[X])^2 = {e_x2} - {e_x**2} = {e_x2 - e_x**2}")
```

```python
# Exercise 2: scaling rules, no summation
var_x = 35 / 12
print(f"Exercise 2: Var(3X + 100) = 9 * 35/12 = {9 * var_x:.4f}  (the +100 is irrelevant)")
print(f"Var(-X) = (-1)^2 * Var(X) = {1 * var_x:.4f}  (flipping the sign changes no distances)")
```

```python
# Exercise 3: risky game vs sure thing
values = np.array([10, 110])
pmf = np.array([0.9, 0.1])
mean = np.sum(values * pmf)
var = np.sum((values - mean)**2 * pmf)
print(f"Exercise 3: risky game: E = {mean}, Var = {var:.0f}")
print("A sure $20 also has E = 20 but Var = 0. Same average, no spread --")
print("the variance is exactly what separates the two offers (risk).")
```

```python
# Exercise 4: Bernoulli variance p(1-p)
print("Exercise 4:")
print(f"{'p':>5} {'Var (computed)':>15} {'p(1-p)':>8}")
for p in np.arange(0.0, 1.01, 0.125):
    values = np.array([0, 1])
    pmf = np.array([1 - p, p])
    mean = np.sum(values * pmf)
    var = np.sum((values - mean)**2 * pmf)
    print(f"{p:>5.3f} {var:>15.5f} {p*(1-p):>8.5f}")
print()
print("Maximum at p = 0.5: a fair coin is the MOST unpredictable; at p = 0 or 1")
print("the outcome is certain and the variance collapses to 0.")
```

```python
# Exercise 5: proof and numerical check of Var(aX + b) = a^2 Var(X)
# Proof: E[Y] = a E[X] + b, so
#   Var(Y) = E[(aX + b - (a E[X] + b))^2]   <- the b's cancel inside
#          = E[(a(X - E[X]))^2]
#          = a^2 E[(X - E[X])^2] = a^2 Var(X).
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)
a, b = -3, 7

y = a * faces + b
mean_y = np.sum(y * pmf)
var_y = np.sum((y - mean_y)**2 * pmf)
print(f"Exercise 5: Var(-3X + 7) = {var_y:.4f}, a^2 Var(X) = {9 * 35/12:.4f}")
```

```python
# Exercise 6: variance of the uniform bus wait on [0, 60]
x = np.linspace(0, 60, 4001)
f = np.full_like(x, 1 / 60)

mean = np.trapezoid(x * f, x)
e_x2 = np.trapezoid(x**2 * f, x)
var = e_x2 - mean**2
print(f"Exercise 6: E[X] = {mean:.2f}, Var = {var:.2f}, formula (60-0)^2/12 = {60**2/12:.2f}")
print()
print("Die comparison: (6-1)^2/12 = 2.0833 but Var(die) = 35/12 = 2.9167.")
print("The die is DISCRETE: its mass sits at 6 separated points rather than")
print("smeared continuously, which adds extra spread; the exact discrete-uniform")
print("formula is (n^2 - 1)/12 = (36 - 1)/12 = 35/12.")
```

## 8. Conclusion

- The expected value alone cannot distinguish a narrow distribution from a wide one — two games can both be "worth `$0`" while one risks `$1` and the other `$100`.
- The average deviation is **always zero** (the mean is the balancing point), so spread needs positive quantities: absolute deviations work but are mathematically awkward; **squared deviations** are the standard.
- **Variance** is the average squared deviation, weighted by probabilities: $\mathrm{Var}(X) = E[(X - E[X])^2] = \sum_x (x - E[X])^2 p(x)$.
- The shortcut $\mathrm{Var}(X) = E[X^2] - (E[X])^2$ follows from linearity of expectation and is usually easier to compute.
- Variance ignores where the distribution is centered: win `$2`/lose `$2` and win `$3`/lose `$1` both have variance 4.
- **Scaling rule:** $\mathrm{Var}(aX + b) = a^2\,\mathrm{Var}(X)$ — multiplying stretches all deviations by $a$ (variance by $a^2$); adding $b$ merely slides the distribution and changes nothing about its spread.
- For the die: $\mathrm{Var}(X) = \tfrac{35}{12} \approx 2.92$, and the game $2X-5$ has variance $4 \cdot \tfrac{35}{12} \approx 11.67$.

One wrinkle remains: variance is measured in *squared* units (squared dollars!). The fix — taking its square root — is the **standard deviation**, coming up next.

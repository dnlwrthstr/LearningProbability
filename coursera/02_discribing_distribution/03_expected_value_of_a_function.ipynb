# Expected Value of a Function of a Random Variable

You know how to find $E(X)$. But what if you need the expected value of the *square* of a random variable, or its cube, or any other function of it? This notebook shows the simple recipe: keep the probabilities, replace the values by $g(x)$. Along the way we price two dice games and discover one of the most useful facts in all of probability — **expectation is linear**: $E(aX + b) = a\,E(X) + b$. It is based on the third video of Lesson 3 (Describing Distributions).

## 1. Introduction

Many quantities you care about are not a random variable itself but a *function* of it: the square of a dice roll (a game payoff), a temperature converted from Celsius to Fahrenheit, a profit computed as revenue times a rate minus a fixed cost. If you know the distribution of $X$, how do you get the expected value of $g(X)$ without re-deriving the whole distribution of $g(X)$?

This matters because:

- Pricing games and decisions usually means taking expectations of *transformed* outcomes (payoffs, costs, utilities).
- The most important descriptor after the mean — the **variance**, coming in the next videos — is itself the expected value of a function of $X$: namely $(X - \mu)^2$.
- The linearity rule lets you move expectations through linear transformations instantly, with no summation at all.

After completing this notebook you will be able to:

- Compute $E[g(X)]$ for a discrete random variable using $\sum_x g(x)\,p(x)$.
- Price a game whose payoff is a function of a die roll.
- State and use the **linearity of expectation**: $E(aX + b) = a\,E(X) + b$, including the special cases $E(aX) = a\,E(X)$ and $E(b) = b$.
- Explain why, in general, $E[g(X)] \ne g(E[X])$ — unless $g$ is linear.

## 2. Theory

### Recap: the expected value

Suppose $X$ takes the values $x_1, x_2, x_3, x_4$ with probabilities $p(x_1), p(x_2), p(x_3), p(x_4)$. The expected value multiplies each outcome by its probability and sums:

$$
E[X] = x_1\,p(x_1) + x_2\,p(x_2) + x_3\,p(x_3) + x_4\,p(x_4)
$$

### The expected value of a function: same recipe

If you are interested in some function of $X$ — say $X^2$, $X^3$, or any $g(X)$ — the result is very similar. You do the same summation, but multiply $g(x)$ by the probability of $x$ each time:

$$
E[g(X)] = g(x_1)\,p(x_1) + g(x_2)\,p(x_2) + g(x_3)\,p(x_3) + g(x_4)\,p(x_4)
$$

That is the whole trick: **replace each $x$ with $g(x)$ and keep the probabilities the same.** In general, for a discrete random variable,

$$
E[g(X)] = \sum_x g(x)\, p(x)
$$

(For a continuous random variable the same idea reads $E[g(X)] = \int g(x) f(x)\,dx$ — the weights become the PDF, exactly as with the plain mean.)

Why does this work? The outcome $g(x)$ happens exactly when the outcome $x$ happens — so it inherits its probability. No new distribution needs to be computed.

### Linearity of expectation

For the special case of a **linear** function $g(x) = ax + b$, something remarkable happens. Whatever the distribution of $X$:

$$
E[aX + b] = a\,E[X] + b
$$

The expectation is a **linear operator** — it passes through multiplication by a constant and addition of a constant. Two immediate takeaways:

- $E[aX] = a\,E[X]$ — scaling the variable scales the mean.
- $E[b] = b$ — the expected value of a constant is the constant itself.

### Caution: this only works for linear functions

It is tempting to compute $E[g(X)]$ as $g(E[X])$ — plug the mean into the function. **For non-linear $g$ this is wrong.** For a fair die, $E[X^2] = \frac{91}{6} \approx 15.17$, but $(E[X])^2 = 3.5^2 = 12.25$. The weighted average of the squares is not the square of the weighted average. Linearity is precisely the property that makes $g(x) = ax+b$ the exception.

## 3. Worked Examples

### Example 1: the square game

You roll a fair die and your friend pays you the **square** of the amount you roll. What is a fair price to pay for this game?

Same playbook as before: plot the distribution of the payoffs $1, 4, 9, 16, 25, 36$ (each with probability $\tfrac{1}{6}$) and find its balance point:

$$
E[X^2] = \tfrac{1}{6}(1) + \tfrac{1}{6}(4) + \tfrac{1}{6}(9) + \tfrac{1}{6}(16) + \tfrac{1}{6}(25) + \tfrac{1}{6}(36)
       = \frac{91}{6} \approx 15.17
$$

You have summed the squares of the rolls times the probability of each roll — that *is* the expected value of $X^2$. The fair price is about **`$15.17`**. Nothing harder than before: you just transformed the problem into one you already know.

### Example 2: the double game — and a shortcut

Your friend changes the game: they pay you **twice** the number you roll, but you must pay **`$5`** up front. Your net win is $2X - 5$:

| roll $x$ | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|
| double $2x$ | 2 | 4 | 6 | 8 | 10 | 12 |
| net win $2x - 5$ | $-3$ | $-1$ | $1$ | $3$ | $5$ | $7$ |

(A negative win means you lost money.) Averaging the six equally likely net wins:

$$
E[2X - 5] = \frac{(2\cdot 1) + (2\cdot 2) + \cdots + (2\cdot 6)}{6} + (-5)
$$

Group the six copies of $-5$: they contribute $\frac{6 \cdot (-5)}{6} = -5$. Factor the common 2 out of the remaining sum:

$$
E[2X - 5] = 2 \cdot \underbrace{\frac{1+2+3+4+5+6}{6}}_{E[X] \,=\, 3.5} + (-5) = 2 \cdot 3.5 - 5 = 2
$$

Notice anything cool? The expected value of $2X - 5$ is $2 \cdot E[X] - 5$ — the constants moved *outside* the expectation. On average you win `$2` per play, so this game is worth playing (any entry price below `$7` total would be).

> **Note (erratum):** the course video's slide lists the net wins as $-3, -2, -1, 0, 1, 2$ and concludes $E = -0.5$. Those numbers actually belong to the game "win the roll, pay `$4`" ($X - 4$, with $E[X-4] = 3.5 - 4 = -0.5$) — for $2X-5$ the wins step by 2, as in the table above. Either version demonstrates the same rule: $E[aX+b] = a\,E[X]+b$.

### Example 3: reading the general rule

The derivation above never used anything specific to dice. For any distribution, the constants $a$ and $b$ can be pulled through the sum $\sum_x (ax + b)\,p(x) = a\sum_x x\,p(x) + b\sum_x p(x) = a\,E[X] + b$ — using the fact that all probabilities sum to 1. That last step is also why $E[b] = b$.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)
```

### $E[g(X)]$ for a die: keep the probabilities, transform the values

```python
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)

def expected_value(values, pmf, g=lambda x: x):
    return np.sum(g(values) * pmf)

print(f"E[X]      = {expected_value(faces, pmf):.4f}")
print(f"E[X^2]    = {expected_value(faces, pmf, lambda x: x**2):.4f}   (91/6 = {91/6:.4f})")
print(f"E[X^3]    = {expected_value(faces, pmf, lambda x: x**3):.4f}")
print(f"E[2X - 5] = {expected_value(faces, pmf, lambda x: 2*x - 5):.4f}")
```

### Simulation: the square game really is worth `$15.17`

Play the game many times and watch the average payoff converge to $91/6$.

```python
rng = np.random.default_rng(42)

n_games = 100_000
rolls = rng.integers(1, 7, size=n_games)
payoffs = rolls**2

print(f"Theoretical fair price E[X^2]:  {91/6:.4f}")
print(f"Average payoff over {n_games:,} games: {payoffs.mean():.4f}")
```

### Linearity check — and where the naive shortcut fails

$E[2X-5]$ matches $2\,E[X]-5$ exactly. But $E[X^2]$ does **not** match $(E[X])^2$: plugging the mean into a non-linear function gives the wrong answer.

```python
rng = np.random.default_rng(42)
rolls = rng.integers(1, 7, size=100_000)

mean_x = rolls.mean()

print("Linear function g(x) = 2x - 5:")
print(f"  E[2X - 5] (simulated)  = {(2*rolls - 5).mean():.4f}")
print(f"  2 E[X] - 5             = {2*mean_x - 5:.4f}   <- matches: linearity")
print()
print("Non-linear function g(x) = x^2:")
print(f"  E[X^2] (simulated)     = {(rolls**2).mean():.4f}")
print(f"  (E[X])^2               = {mean_x**2:.4f}   <- does NOT match")
```

### $E$ of a constant is the constant

A "random" variable that always equals 7 has expected value 7 — the balance point of a single spike.

```python
constant = np.full(100_000, 7)
print(f"E[7] = {constant.mean()}")
print(f"E[0 * X + 7] via linearity: 0 * E[X] + 7 = {0 * 3.5 + 7}")
```

## 5. Visualizations

### Transforming the values moves the bars, not the probabilities

Left: the PMF of a die roll $X$ — six bars of height $\tfrac16$ at $1,\dots,6$, balancing at 3.5. Right: the PMF of $X^2$ — the *same six bars*, slid to the positions $1, 4, 9, \dots, 36$. The balance point moves to $91/6 \approx 15.17$, far from $3.5^2 = 12.25$.

```python
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5), sharey=True)

for ax, values, name in [(axes[0], faces, "X (the roll)"),
                         (axes[1], faces**2, "X$^2$ (the payoff)")]:
    mean = np.sum(values * pmf)
    ax.bar(values, pmf, color="tab:blue", edgecolor="black", alpha=0.85, width=0.8)
    ax.plot(mean, -0.012, "^", markersize=15, color="black", clip_on=False)
    ax.axvline(mean, color="tab:red", linestyle="--", linewidth=1.5,
               label=f"E = {mean:.2f}")
    ax.set_title(f"PMF of {name}")
    ax.set_xlabel("value")
    ax.grid(True, alpha=0.3)
    ax.legend()

axes[1].axvline(3.5**2, color="gray", linestyle=":", linewidth=1.5)
axes[1].annotate("$(E[X])^2 = 12.25$\n(not the mean!)", xy=(12.25, 0.12),
                 xytext=(17, 0.135), arrowprops=dict(arrowstyle="->"), fontsize=10)

axes[0].set_ylabel("probability")
plt.suptitle("Same probabilities, transformed values: the balance point of $X^2$ is $E[X^2]$")
plt.tight_layout()
plt.show()
```

### The double game on a seesaw

The six equally likely net wins $2x - 5$ balance at $E[2X-5] = 2$ — exactly $2 \cdot 3.5 - 5$.

```python
faces = np.arange(1, 7)
net_wins = 2 * faces - 5
mean_win = net_wins.mean()          # equal probabilities -> plain average

plt.figure(figsize=(9, 4))
plt.axhline(0, color="black", linewidth=2)
plt.plot(net_wins, np.full(6, 0.25), "s", markersize=14, color="tab:red",
         label="net win of each roll (prob. 1/6 each)")
plt.plot(mean_win, -0.28, "^", markersize=18, color="black")
plt.annotate(f"balance: E[2X-5] = {mean_win:.0f} = 2·E[X] - 5",
             xy=(mean_win, -0.28), xytext=(mean_win + 1.2, -0.9),
             arrowprops=dict(arrowstyle="->"), fontsize=11)
plt.title("Net wins of the double game balance at 2·3.5 - 5 = 2")
plt.xlabel("net win ($)")
plt.yticks([])
plt.xticks(net_wins)
plt.ylim(-1.4, 1)
plt.grid(True, alpha=0.3, axis="x")
plt.legend(loc="upper left")
plt.show()
```

### Linearity in action: shift and scale the whole distribution

Applying $g(x) = ax + b$ slides and stretches the PMF — and the mean follows the *same* rule, $a\,E[X] + b$, without any new computation.

```python
rng = np.random.default_rng(42)
rolls = rng.integers(1, 7, size=100_000)

transforms = [("X", rolls, 1, 0),
              ("2X", 2 * rolls, 2, 0),
              ("2X - 5", 2 * rolls - 5, 2, -5)]

plt.figure(figsize=(10, 5))
colors = ["tab:blue", "tab:orange", "tab:green"]
for (label, data, a, b), color in zip(transforms, colors):
    values, counts = np.unique(data, return_counts=True)
    plt.bar(values, counts / len(data), width=0.5, alpha=0.6, color=color,
            edgecolor="black", label=f"{label}:  mean = {data.mean():.2f}"
                                     f"  (rule: {a}·3.5 + {b} = {a*3.5 + b})")
    plt.axvline(data.mean(), color=color, linestyle="--", linewidth=1.5)

plt.title("E[aX + b] = a·E[X] + b: the mean follows the transformation")
plt.xlabel("value")
plt.ylabel("relative frequency")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 7. Solutions

```python
# Exercise 1: fair coin, X in {0, 1}
values = np.array([0, 1])
pmf = np.array([0.5, 0.5])
e_x = np.sum(values * pmf)
e_x2 = np.sum(values**2 * pmf)
print(f"Exercise 1: E[X] = {e_x}, E[X^2] = {e_x2}")
print("They coincide because 0^2 = 0 and 1^2 = 1: squaring changes nothing")
print("when the only values are 0 and 1.")
```

```python
# Exercise 2: linearity, no summation needed
print(f"Exercise 2: E[3X + 2] = 3 * 3.5 + 2 = {3 * 3.5 + 2}")
```

```python
# Exercise 3: Celsius to Fahrenheit
e_f = 9 / 5 * 22 + 32
print(f"Exercise 3: E[F] = 9/5 * 22 + 32 = {e_f} degrees Fahrenheit")
print("F is a LINEAR function of C, so only E[C] matters -- linearity holds")
print("for every distribution, whatever its shape.")
```

```python
# Exercise 4: the cube game
rng = np.random.default_rng(0)

faces = np.arange(1, 7)
fair_price = np.sum(faces**3 / 6)
simulated = (rng.integers(1, 7, size=100_000)**3).mean()
print(f"Exercise 4: E[X^3] = {fair_price:.2f}  (= 441/6)")
print(f"Simulated: {simulated:.2f}")
print(f"Wrong answer (E[X])^3 = 3.5^3 = {3.5**3:.2f} -- underestimates the fair price.")
```

```python
# Exercise 5: expected deviation from the mean, and expected squared deviation
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)
mu = np.sum(faces * pmf)

e_deviation = np.sum((faces - mu) * pmf)
e_sq_deviation = np.sum((faces - mu)**2 * pmf)
print(f"Exercise 5: E[X - mu] = {e_deviation:.4f}")
print("One line: E[X - mu] = E[X] - mu = 3.5 - 3.5 = 0 (linearity with a=1, b=-mu).")
print(f"E[(X - mu)^2] = {e_sq_deviation:.4f}  <- the variance, coming up next!")
```

```python
# Exercise 6: the variance shortcut identity
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)
mu = np.sum(faces * pmf)

lhs = np.sum((faces - mu)**2 * pmf)
rhs = np.sum(faces**2 * pmf) - mu**2
print(f"Exercise 6: E[(X-mu)^2] = {lhs:.4f},  E[X^2] - (E[X])^2 = {rhs:.4f}")
print("Proof by linearity: E[(X-mu)^2] = E[X^2 - 2*mu*X + mu^2]")
print("  = E[X^2] - 2*mu*E[X] + mu^2   (constants move out, E of constant is itself)")
print("  = E[X^2] - 2*mu^2 + mu^2 = E[X^2] - mu^2.")
```

## 8. Conclusion

- To find the expected value of a *function* of a random variable, keep the probabilities and transform the values: $E[g(X)] = \sum_x g(x)\,p(x)$ (with an integral and the PDF in the continuous case).
- The square game prices at $E[X^2] = \tfrac{91}{6} \approx 15.17$ — computed with the distribution of $X$, no new distribution needed.
- **Expectation is linear:** $E[aX + b] = a\,E[X] + b$ for *any* distribution of $X$. In particular $E[aX] = a\,E[X]$ and $E[b] = b$.
- Linearity is special to linear functions: in general $E[g(X)] \ne g(E[X])$ — e.g. $E[X^2] = 15.17 \ne 12.25 = (E[X])^2$.
- The gap between those two numbers, $E[X^2] - (E[X])^2$, is exactly the **variance** — the subject of the next video.

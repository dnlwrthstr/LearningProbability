# Standardizing a Distribution

Distributions are nicest to work with when their mean is 0 and their standard deviation is 1. This notebook formalizes the two-step recipe that gets *any* distribution there: **center** it by subtracting the mean, then **scale** it by dividing by the standard deviation. The combined transformation $Z = \frac{X - \mu}{\sigma}$ is called **standardizing**, and both steps are one-line consequences of the linearity and scaling rules from the earlier notebooks. It is based on the eighth video of Lesson 3 (Describing Distributions).

## 1. Introduction

You have seen this move before — standardizing appeared when we studied the normal distribution in Lesson 2. Now, with expectations, variance, and their transformation rules in hand, we can *prove* why it works for any distribution, not just the normal.

Why this matters:

- Standardized variables are **comparable**: "2 standard deviations above the mean" means the same thing for exam scores, heights, or stock returns.
- Many tools expect standardized inputs: z-scores and normal tables in statistics, feature scaling in machine learning.
- The two proofs are perfect practice for the rules you just learned: $E[aX+b] = aE[X]+b$ and $\mathrm{Var}(cX) = c^2\mathrm{Var}(X)$.

After completing this notebook you will be able to:

- **Center** a distribution ($X - \mu$) and prove the new mean is 0.
- **Scale** a distribution ($X/\sigma$) and prove the new standard deviation is 1.
- **Standardize** any distribution with $Z = \frac{X-\mu}{\sigma}$ and interpret $z$-values.

## 2. Theory

### Step 1: centering — make the mean zero

Take a distribution with mean $\mu$. It is always nice when a distribution has mean 0, so consider the new variable

$$
X - \mu
$$

where $\mu$ is just a fixed number. Why does it have expectation zero? Expectation is **linear**, so with $Y = -\mu$:

$$
E[X - \mu] = E[X] - E[\mu] = E[X] - \mu = \mu - \mu = 0
$$

(using that the expectation of the constant $\mu$ is $\mu$ itself). Subtracting the mean slides the whole distribution so it balances at 0. This is called **centering**. Note the spread is untouched: the standard deviation is still $\sigma$.

### Step 2: scaling — make the standard deviation one

Now suppose the mean is already 0 but the standard deviation is $\sigma$. Just as we wanted mean 0, we would like standard deviation 1. The fix: divide by $\sigma$, giving $X / \sigma$.

The math, using the shortcut formula for a constant $c$:

$$
\mathrm{Var}(cX) = E[(cX)^2] - \big(E[cX]\big)^2 = c^2 E[X^2] - c^2 \big(E[X]\big)^2 = c^2\,\mathrm{Var}(X)
$$

— a constant comes out of the variance **as a square**. With $c = \tfrac{1}{\sigma}$:

$$
\mathrm{Var}\left(\frac{X}{\sigma}\right) = \frac{1}{\sigma^2}\,\mathrm{Var}(X)
\qquad\Longrightarrow\qquad
\mathrm{std}\left(\frac{X}{\sigma}\right) = \frac{1}{\sigma}\,\mathrm{std}(X) = \frac{\sigma}{\sigma} = 1
$$

(taking square roots turns the $\tfrac{1}{\sigma^2}$ into $\tfrac{1}{\sigma}$ and the variance into the standard deviation). This is called **scaling**.

### The whole recipe: standardizing

Putting the two steps together for a distribution $X$ with mean $\mu$ and standard deviation $\sigma$:

| step | variable | mean | std |
|---|---|---|---|
| start | $X$ | $\mu$ | $\sigma$ |
| **center** (subtract $\mu$) | $X - \mu$ | $0$ | $\sigma$ |
| **scale** (divide by $\sigma$) | $Z = \dfrac{X - \mu}{\sigma}$ | $0$ | $1$ |

Any time you are asked to *standardize* a distribution: **subtract the mean, divide by the standard deviation.** The result is a nice distribution with mean 0 and standard deviation 1.

The standardized value $z = \frac{x - \mu}{\sigma}$ of an outcome is its **z-score**: how many standard deviations it sits above (positive) or below (negative) the mean — a universal, unit-free coordinate for "how unusual is this value?".

## 3. Worked Examples

### Example 1: standardizing a die roll

$X$ = fair die: $\mu = 3.5$, $\sigma = \sqrt{35/12} \approx 1.708$.

- Centering: $X - 3.5$ takes the values $-2.5, -1.5, -0.5, 0.5, 1.5, 2.5$ — mean 0, spread unchanged.
- Scaling: $Z = \frac{X - 3.5}{1.708}$ takes values $\approx -1.46, -0.88, -0.29, 0.29, 0.88, 1.46$ — mean 0, std 1.

A roll of 6 has $z = \frac{6 - 3.5}{1.708} \approx 1.46$: about one and a half standard deviations above average.

### Example 2: exam scores across different exams

Exam A has $\mu = 70$, $\sigma = 8$; exam B has $\mu = 50$, $\sigma = 15$. You scored 82 on A, your friend scored 74 on B. Who did relatively better?

$$
z_{\text{you}} = \frac{82 - 70}{8} = 1.5
\qquad
z_{\text{friend}} = \frac{74 - 50}{15} = 1.6
$$

Your friend's score is 1.6 standard deviations above their exam's mean — slightly more exceptional than your 1.5, even though the raw comparison (82 vs 74) suggests otherwise. Standardizing makes different scales comparable.

### Example 3: standardizing a normal variable

For $X \sim N(\mu, \sigma^2)$, the standardized $Z = \frac{X-\mu}{\sigma} \sim N(0, 1)$ — the **standard normal** from Lesson 2. Standardizing is exactly why one table (or one function) for $N(0,1)$ suffices for every normal distribution. And by the 68–95–99.7 rule, $|z| > 2$ happens only about 5% of the time, $|z| > 3$ about 0.3%.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)
```

### The two steps on a die roll

```python
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)

mu = np.sum(faces * pmf)
sigma = np.sqrt(np.sum((faces - mu)**2 * pmf))

centered = faces - mu
standardized = (faces - mu) / sigma

def describe(values, pmf, label):
    mean = np.sum(values * pmf)
    std = np.sqrt(np.sum((values - mean)**2 * pmf))
    print(f"{label:<22} mean = {mean:>7.4f}   std = {std:.4f}")
    print(f"{'':<22} values: {np.round(values, 3)}")

describe(faces, pmf, "X (die roll)")
describe(centered, pmf, "X - mu (centered)")
describe(standardized, pmf, "Z = (X-mu)/sigma")
```

### Standardizing works on any distribution

Center-and-scale a skewed exponential sample and a uniform sample: whatever the shape, the result has mean 0 and std 1 — note that the *shape itself does not change*, only its position and width.

```python
rng = np.random.default_rng(42)

samples = {
    "normal(10, 2)": rng.normal(10, 2, size=100_000),
    "exponential(5)": rng.exponential(5, size=100_000),
    "uniform[0, 60]": rng.uniform(0, 60, size=100_000),
}

for name, x in samples.items():
    z = (x - x.mean()) / x.std()
    print(f"{name:<15} raw: mean = {x.mean():>7.3f}, std = {x.std():>7.3f}"
          f"   standardized: mean = {z.mean():>6.3f}, std = {z.std():.3f}")
```

### z-scores: comparing across different scales

```python
# You: 82 points on exam A (mu 70, sigma 8)
# Friend: 74 points on exam B (mu 50, sigma 15)
z_you = (82 - 70) / 8
z_friend = (74 - 50) / 15

print(f"Your z-score:     {z_you:.2f}")
print(f"Friend's z-score: {z_friend:.2f}")
print("The friend did relatively better: their score is more standard deviations")
print("above their exam's mean, even though the raw score is lower.")
```

## 5. Visualizations

### Centering slides, scaling squeezes

The three stages for a normal distribution $N(10, 2^2)$: the original bell, the centered bell at 0 (same width), and the standardized bell (width 1).

```python
rng = np.random.default_rng(42)

x = rng.normal(10, 2, size=100_000)
stages = [(x, "X ~ N(10, 4)", "tab:blue"),
          (x - 10, "X - mu  (centered: mean 0, std 2)", "tab:orange"),
          ((x - 10) / 2, "Z = (X-mu)/sigma  (standardized: mean 0, std 1)", "tab:green")]

plt.figure(figsize=(11, 5))
for data, label, color in stages:
    plt.hist(data, bins=100, density=True, alpha=0.5, color=color, label=label)
plt.axvline(0, color="black", linewidth=1)
plt.title("Standardizing = center (slide to 0), then scale (squeeze to width 1)")
plt.xlabel("value")
plt.ylabel("density")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

### The shape survives standardization

Standardize a skewed distribution: it does *not* become a bell — it becomes the same skewed shape with mean 0 and std 1. Standardizing changes coordinates, not character.

```python
rng = np.random.default_rng(42)

x = rng.exponential(5, size=100_000)
z = (x - x.mean()) / x.std()

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
axes[0].hist(x, bins=100, density=True, color="tab:blue", alpha=0.7)
axes[0].axvline(x.mean(), color="tab:red", linestyle="--", label=f"mean = {x.mean():.1f}")
axes[0].set_title("skewed waiting times (exponential)")
axes[0].set_xlabel("minutes")

axes[1].hist(z, bins=100, density=True, color="tab:green", alpha=0.7)
axes[1].axvline(0, color="tab:red", linestyle="--", label="mean = 0")
axes[1].set_title("the same data standardized: still skewed!")
axes[1].set_xlabel("z (standard deviations from the mean)")

for ax in axes:
    ax.set_ylabel("density")
    ax.grid(True, alpha=0.3)
    ax.legend()
plt.tight_layout()
plt.show()
```

### Different normals, one standard normal

Standardize three very different normal distributions — all three collapse onto the same $N(0,1)$ curve.

```python
rng = np.random.default_rng(42)

params = [(70, 8, "exam scores N(70, 8²)"),
          (1.70, 0.10, "heights N(1.7, 0.1²)"),
          (15, np.sqrt(5), "response times N(15, 5)")]

grid = np.linspace(-4, 4, 400)
standard_pdf = np.exp(-0.5 * grid**2) / np.sqrt(2 * np.pi)

plt.figure(figsize=(10, 5))
for (mu, sigma, label), color in zip(params, ["tab:blue", "tab:orange", "tab:green"]):
    z = (rng.normal(mu, sigma, size=100_000) - mu) / sigma
    plt.hist(z, bins=80, density=True, alpha=0.35, color=color, label=f"standardized {label}")
plt.plot(grid, standard_pdf, color="black", linewidth=2, label="standard normal N(0, 1)")
plt.title("After standardizing, every normal distribution is the same distribution")
plt.xlabel("z")
plt.ylabel("density")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 7. Solutions

```python
# Exercise 1: z-scores for mu = 20, sigma = 4
for value in [28, 20, 14]:
    z = (value - 20) / 4
    print(f"x = {value}:  z = {z:+.1f}")
print("28 is 2 standard deviations above the mean (quite high); 20 is exactly")
print("average (z = 0); 14 is 1.5 standard deviations below the mean.")
```

```python
# Exercise 2: the +-$100 game (mu = 0, sigma = 100)
values = np.array([-100.0, 100.0])
pmf = np.array([0.5, 0.5])
mu = np.sum(values * pmf)                       # already 0: centering changes nothing
sigma = np.sqrt(np.sum((values - mu)**2 * pmf))
z = (values - mu) / sigma
print(f"Exercise 2: the game is already centered (mu = {mu:.0f}), std stays {sigma:.0f}")
print(f"After standardizing the outcomes become {z}: mean 0, std 1.")
```

```python
# Exercise 3: Z as a linear transformation aX + b
# E[Z] = a*mu + b = mu/sigma - mu/sigma = 0
# Var(Z) = a^2 Var(X) = (1/sigma^2) * sigma^2 = 1
faces = np.arange(1, 7)
pmf = np.full(6, 1 / 6)
mu = np.sum(faces * pmf)
sigma = np.sqrt(np.sum((faces - mu)**2 * pmf))

a, b = 1 / sigma, -mu / sigma
z = a * faces + b
mean_z = np.sum(z * pmf)
var_z = np.sum((z - mean_z)**2 * pmf)
print(f"Exercise 3: E[Z] = {mean_z:.10f}, Var(Z) = {var_z:.10f}")
print("By the rules: E[Z] = mu/sigma - mu/sigma = 0, Var(Z) = sigma^2/sigma^2 = 1.")
```

```python
# Exercise 4: un-standardizing to generate any normal
rng = np.random.default_rng(0)

z = rng.standard_normal(100_000)
x = 7 * z + 42                      # sigma Z + mu
print(f"Exercise 4: E[sigma Z + mu] = sigma*0 + mu = 42, std = |sigma|*1 = 7")
print(f"Simulated: mean = {x.mean():.3f}, std = {x.std():.3f}")
```

```python
# Exercise 5: standardizing with ESTIMATED parameters
rng = np.random.default_rng(0)

n, trials = 10, 20_000
samples = rng.normal(50, 10, size=(trials, n))
z = (samples - samples.mean(axis=1, keepdims=True)) / samples.std(axis=1, keepdims=True)

print(f"Exercise 5: pooled standardized values: mean = {z.mean():.4f}, std = {z.std():.4f}")
print("The pooled mean is 0 and std is 1 by construction, but each small sample's")
print("z-scores used NOISY estimates of mu and sigma -- individual values are not")
print("exactly N(0,1) (e.g. a z-score can never exceed sqrt(n-1) = 3 for n = 10).")
```

```python
# Exercise 6: normal intuition vs an exact discrete distribution
faces = np.arange(1, 7)
mu, sigma = 3.5, np.sqrt(35 / 12)
z = (faces - mu) / sigma
extreme = np.abs(z) >= 1.4

print(f"Exercise 6: z-scores of the faces: {np.round(z, 3)}")
print(f"|z| >= 1.4 happens for faces {faces[extreme]} -> P = {extreme.sum()}/6 = {extreme.sum()/6:.3f}")
print("A normal variable exceeds |z| = 1.4 about 16% of the time; the die does it")
print("33% of the time. The 68-95-99.7-style intuition is about the BELL shape --")
print("a flat distribution puts far more mass at its extremes.")
```

## 8. Conclusion

- **Centering:** $X - \mu$ has mean 0, because $E[X - \mu] = E[X] - \mu = 0$ (linearity; the expectation of a constant is the constant). The spread is unchanged.
- **Scaling:** dividing by $\sigma$ makes the standard deviation 1, because constants leave the variance *as squares*: $\mathrm{Var}(cX) = c^2\mathrm{Var}(X)$, so $\mathrm{std}(X/\sigma) = \sigma/\sigma = 1$.
- **Standardizing = center + scale:** $Z = \frac{X - \mu}{\sigma}$ has mean 0 and standard deviation 1, for *any* starting distribution — though the shape (skew, discreteness) is preserved, only position and width change.
- z-scores are the universal currency of "how unusual": they make values from different scales directly comparable, and for normal data they plug straight into the 68–95–99.7 rule.
- The recipe reverses: $X = \sigma Z + \mu$ turns standard samples into samples with any desired mean and spread.

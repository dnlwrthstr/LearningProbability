# The Mean and the Expected Value

This notebook introduces the **mean** of a distribution and its formal name in probability: the **expected value** $E(X)$. The key intuition is physical — the mean is the *balancing point* of a distribution. We compute it as a weighted average for discrete variables, see the continuous analogue, use it to price a betting game, and clear up a common misconception (the mean is *not* where the data splits in half — that is the median). It is based on the first video of Lesson 3 (Describing Distributions).

## 1. Introduction

You already know how to *build* probability distributions. This lesson is about *describing* them: single numbers that summarize where a distribution sits and how it spreads. Mean, median, and mode describe the **center**; variance describes the **spread**; later come skewness, kurtosis, and distributions of several variables (joint, marginal, conditional, covariance).

We start with the most familiar of them all: the mean. Why does it matter?

- It is the single best-known summary of a dataset ("the average").
- In probability it becomes the **expected value** — the long-run average outcome of a random experiment.
- It answers practical questions directly, e.g. *what is a fair price to pay for a bet?*

After completing this notebook you will be able to:

- Compute the mean of a dataset and recognize it as a **weighted average** where the weights are probabilities.
- Read and write the expected-value notation $E(X)$ and the formula $E(X) = \sum_x x \, p(x)$.
- Interpret the mean visually as the **balancing point** of a distribution (discrete or continuous).
- Use expected value to find the fair price of a game of chance.
- Explain why the mean and the median are generally *different* points.

## 2. Theory

### Intuition: the balancing point

Take a sample of 10 kids of different ages: **three** kids of age 0, **two** of age 1, **four** of age 2, and **one** of age 3. Replace each kid with an equal ball and stack the balls over their ages on a seesaw. By trial and error, the plank balances at one specific point — at age **1.3**. That tipping point *is* the mean.

### From the familiar average to a weighted average

The usual recipe — sum all ages, divide by the count:

$$
\text{mean} = \frac{3 \cdot 0 + 2 \cdot 1 + 4 \cdot 2 + 1 \cdot 3}{10} = \frac{13}{10} = 1.3
$$

Now distribute the 10 into every term:

$$
\text{mean} = \frac{3}{10} \cdot 0 + \frac{2}{10} \cdot 1 + \frac{4}{10} \cdot 2 + \frac{1}{10} \cdot 3 = 1.3
$$

The fractions are exactly the **probabilities** of each age: $P(\text{age}=0) = 3/10$, $P(\text{age}=1) = 2/10$, and so on. The mean is a **weighted average of the values, weighted by their probabilities**.

### Expected value: the formal name

In probability the mean of a distribution is called the **expected value**. If $X$ is the random variable (here: the age of a kid), it is written

$$
E(X)
$$

and read "E of X". *Mean* and *expected value* refer to the same concept. In this example, the expected age of a kid is $E(X) = 1.3$ years.

### The general formula (discrete)

For a discrete random variable $X$ with probability mass function $p(x)$:

$$
E(X) = \sum_x x \, p(x)
$$

— sum over all possible values, each weighted by its probability. Two consequences:

- If **every value has the same probability**, the expected value sits exactly in the middle of the values.
- As soon as one value gets **more weight**, the balancing point shifts toward it. (And since all probabilities must sum to 1, raising one probability forces the others down.)

### The continuous case

For a continuous random variable the sum becomes an integral and the PMF becomes the probability density function $f(x)$:

$$
E(X) = \int_{-\infty}^{\infty} x \, f(x) \, dx
$$

Don't worry if you have not studied integrals — the idea is the same weighted average: an integral just sums the contributions of infinitely many, infinitesimally narrow bars. Visually, nothing changes: instead of balancing stacked balls you balance a continuous shape.

A useful special case: for a **uniform distribution** on the interval $[a, b]$ the shape is a flat rectangle, so it balances in the very middle:

$$
E(X) = \frac{a + b}{2}
$$

### Common misconception: the mean is not the halfway split

It is tempting to think the mean splits the data 50/50. It does not — the point with half the probability on each side is the **median** (next video). The mean is the *balance* point, and balance depends on **distance as well as mass**: a small mass far away can balance a big mass close by.

Extreme picture: an elephant sits right next to the pivot and a mouse sits kilometers away on the other side — the seesaw can still balance, because the mouse is so far out. In physics this is **torque** (force × lever arm). That is why, for a skewed distribution, the mean is pulled toward the long tail even though less than half of the probability lies there.

## 3. Worked Examples

### Example 1: the kids' ages

Ages: three 0s, two 1s, four 2s, one 3.

$$
E(X) = \frac{3}{10}\cdot 0 + \frac{2}{10}\cdot 1 + \frac{4}{10}\cdot 2 + \frac{1}{10}\cdot 3
     = 0 + 0.2 + 0.8 + 0.3 = 1.3
$$

### Example 2: what is a fair price for a coin game?

You flip a fair coin: heads pays **`$10`**, tails pays **`$0`**.

- Would you pay `$6` per play? What about `$4`?
- Long-term thinking: half the time you win `$10`, half the time `$0`.

$$
E(X) = 0.5 \cdot 10 + 0.5 \cdot 0 = 5
$$

The expected payoff is **`$5`**, so `$5` is the *maximum* you should be willing to pay: at that price you break even on average. Paying `$4` is a good deal; paying `$6` loses you `$1` per play in the long run.

### Example 3: three coins, `$1` per head

Flip 3 fair coins and win `$1` for each head. The number of heads $X$ has PMF

| $x$ (heads) | 0 | 1 | 2 | 3 |
|---|---|---|---|---|
| $p(x)$ | $\tfrac{1}{8}$ | $\tfrac{3}{8}$ | $\tfrac{3}{8}$ | $\tfrac{1}{8}$ |

$$
E(X) = \tfrac{1}{8}\cdot 0 + \tfrac{3}{8}\cdot 1 + \tfrac{3}{8}\cdot 2 + \tfrac{1}{8}\cdot 3 = \tfrac{12}{8} = 1.5
$$

Sanity check without any formula: you flip 3 coins and about half come up heads — half of 3 is **1.5**. The histogram of the 8 equally likely outcomes, arranged by number of heads, also balances exactly at 1.5. So the fair price for this game is `$1.50`.

### Example 4: waiting for the bus (continuous, uniform)

A bus arrives once an hour and you show up at a random time, so your waiting time is uniform on $[0, 60]$ minutes. A flat shape balances in its middle:

$$
E(X) = \frac{0 + 60}{2} = 30 \text{ minutes.}
$$

Early on, with only a few recorded waits, the sample average bounces around (e.g. 27.83 after a handful of days), but as you collect more data it settles near 30.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)
```

### The kids' ages: plain average vs. weighted average

Both computations give the same number — the second makes the probabilities visible.

```python
# The sample: three kids of age 0, two of age 1, four of age 2, one of age 3
ages = np.array([0, 0, 0, 1, 1, 2, 2, 2, 2, 3])

# Way 1: sum everything, divide by the count
plain_mean = ages.sum() / len(ages)

# Way 2: weighted average -- values weighted by their probabilities
values = np.array([0, 1, 2, 3])
counts = np.array([3, 2, 4, 1])
probabilities = counts / counts.sum()          # [0.3, 0.2, 0.4, 0.1]
expected_value = np.sum(values * probabilities)

print(f"Plain average:          {plain_mean}")
print(f"Weighted average E(X):  {expected_value}")
print(f"Probabilities used:     {probabilities}")
```

### The coin game: theory vs. long-run simulation

Expected value is a statement about the *long term*. Simulate many plays and watch the average payoff converge to `$5`.

```python
rng = np.random.default_rng(42)

n_plays = 100_000
# Each play: heads (probability 0.5) pays $10, tails pays $0
payoffs = rng.choice([10, 0], size=n_plays)

theoretical = 0.5 * 10 + 0.5 * 0
print(f"Theoretical expected payoff: ${theoretical:.2f}")
print(f"Average payoff over {n_plays:,} plays: ${payoffs.mean():.4f}")

for price in [4, 5, 6]:
    net = payoffs.mean() - price
    verdict = "good deal" if net > 0 else ("break even" if abs(net) < 0.05 else "losing money")
    print(f"  Pay ${price} per play -> average net ${net:+.2f} per play ({verdict})")
```

### Three coins: PMF formula and simulation agree

```python
rng = np.random.default_rng(42)

# Exact: weighted average over the PMF of the number of heads in 3 tosses
heads_values = np.array([0, 1, 2, 3])
pmf = np.array([1, 3, 3, 1]) / 8
expected_heads = np.sum(heads_values * pmf)
print(f"E(X) from the PMF: {expected_heads}")

# Simulation: flip 3 coins many times, count heads
n_games = 100_000
heads_per_game = rng.integers(0, 2, size=(n_games, 3)).sum(axis=1)
print(f"Average heads over {n_games:,} games: {heads_per_game.mean():.4f}")
```

### Continuous distributions: the same weighted average, with an integral

For a continuous variable, $E(X) = \int x f(x)\,dx$. Numerically that is just `np.trapezoid(x * f, x)` — summing thin bars. We check two cases: the uniform bus wait (balance in the middle) and a right-skewed wait time where the mean is pulled toward the tail.

```python
# Uniform waiting time on [0, 60] minutes
x = np.linspace(0, 60, 2001)
f_uniform = np.full_like(x, 1 / 60)                 # flat density
mean_uniform = np.trapezoid(x * f_uniform, x)
print(f"Uniform on [0, 60]:  E(X) = {mean_uniform:.2f}  (formula (a+b)/2 = 30)")

# Right-skewed waiting time (exponential-shaped density, like phone hold times)
x = np.linspace(0, 30, 4001)
scale = 5.0
f_skew = (1 / scale) * np.exp(-x / scale)            # exponential PDF, theoretical mean 5
mean_skew = np.trapezoid(x * f_skew, x)

# The median: the point with 50% of the probability on each side
cdf = np.cumsum(f_skew) * (x[1] - x[0])
median_skew = x[np.searchsorted(cdf, 0.5)]

print(f"Skewed wait times:   E(X) = {mean_skew:.2f}, median = {median_skew:.2f}")
print("(theoretical mean is 5; the tiny gap comes from cutting the tail at x = 30)")
print("The mean sits to the RIGHT of the median: the long tail pulls the balance point.")
```

## 5. Visualizations

### The kids' ages balance at 1.3

Each ball is one kid, stacked over its age — the seesaw tips at the mean.

```python
values = np.array([0, 1, 2, 3])
counts = np.array([3, 2, 4, 1])
mean_age = np.sum(values * counts) / counts.sum()

plt.figure(figsize=(8, 5))
for age, count in zip(values, counts):
    for level in range(count):
        plt.plot(age, level + 0.5, "o", markersize=22, color="tab:blue",
                 markeredgecolor="black", zorder=3)
plt.axhline(0, color="black", linewidth=2)
plt.plot(mean_age, -0.35, "^", markersize=18, color="black")   # the pivot
plt.annotate(f"balancing point\n(mean = {mean_age})", xy=(mean_age, -0.35),
             xytext=(mean_age + 0.7, 1.5),
             arrowprops=dict(arrowstyle="->"), fontsize=11)
plt.title("Kids' ages: the distribution balances at the mean")
plt.xlabel("Age")
plt.ylabel("Number of kids")
plt.xticks(values)
plt.ylim(-1, 5)
plt.grid(True, alpha=0.3)
plt.show()
```

### Equal weights balance in the middle — extra weight shifts the pivot

Left: five equally likely values balance at the center (3). Right: value 1 gets extra probability (the others must shrink so everything still sums to 1) and the balance point moves toward it.

```python
values = np.array([1, 2, 3, 4, 5])
pmf_equal = np.full(5, 1 / 5)
pmf_shifted = np.array([0.45, 0.15, 0.15, 0.15, 0.10])   # more weight on 1

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), sharey=True)
for ax, pmf, label in [(axes[0], pmf_equal, "equal weights"),
                       (axes[1], pmf_shifted, "extra weight on x = 1")]:
    mean = np.sum(values * pmf)
    ax.bar(values, pmf, width=0.6, color="tab:blue", edgecolor="black", alpha=0.8)
    ax.plot(mean, -0.03, "^", markersize=16, color="black", clip_on=False)
    ax.set_title(f"{label}:  E(X) = {mean:.1f}")
    ax.set_xlabel("x")
    ax.set_xticks(values)
    ax.grid(True, alpha=0.3)
axes[0].set_ylabel("probability")
plt.suptitle("The expected value is the balancing point of the PMF")
plt.tight_layout()
plt.show()
```

### The running average converges to the expected value

One more way to see what "expected" means: play the `$10`-or-nothing coin game over and over and track the average payoff so far.

```python
rng = np.random.default_rng(7)

n_plays = 2000
payoffs = rng.choice([10, 0], size=n_plays)
running_average = np.cumsum(payoffs) / np.arange(1, n_plays + 1)

plt.figure(figsize=(8, 5))
plt.plot(running_average, color="tab:blue", label="average payoff so far")
plt.axhline(5, color="tab:red", linestyle="--", label="E(X) = $5")
plt.title("Coin game: the long-run average approaches the expected value")
plt.xlabel("Number of plays")
plt.ylabel("Average payoff ($)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

### Mean vs. median on a skewed distribution

The median splits the area 50/50. The mean is the balance point — pulled toward the long tail, exactly like the far-away mouse balancing the elephant.

```python
x = np.linspace(0, 30, 4001)
scale = 5.0
f = (1 / scale) * np.exp(-x / scale)

mean = np.trapezoid(x * f, x)
cdf = np.cumsum(f) * (x[1] - x[0])
median = x[np.searchsorted(cdf, 0.5)]

plt.figure(figsize=(9, 5))
plt.plot(x, f, color="tab:blue", linewidth=2, label="PDF of the waiting time")
plt.fill_between(x, f, where=x <= median, color="orange", alpha=0.5,
                 label="50% of probability (below median)")
plt.fill_between(x, f, where=x > median, color="gold", alpha=0.5,
                 label="50% of probability (above median)")
plt.axvline(median, color="black", linestyle=":", linewidth=2,
            label=f"median = {median:.2f} (50/50 split)")
plt.axvline(mean, color="tab:red", linestyle="--", linewidth=2,
            label=f"mean = {mean:.2f} (balance point)")
plt.title("On a skewed distribution the mean and the median differ")
plt.xlabel("Waiting time (minutes)")
plt.ylabel("Density")
plt.xlim(0, 25)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 8. Conclusion

- The **mean** of a distribution is its **balancing point** — where the seesaw of stacked probabilities tips.
- In probability it is called the **expected value**, written $E(X)$; *mean* and *expected value* are the same concept.
- It is a **weighted average**: $E(X) = \sum_x x\,p(x)$ for discrete variables, $E(X) = \int x f(x)\,dx$ for continuous ones — the weights are the PMF or the PDF.
- Expected value answers real questions: the fair price of a game is its expected payoff (`$5` for the `$10`-or-nothing coin flip; `$1.50` for the three-coin game).
- A uniform distribution on $[a, b]$ balances in the middle: $E(X) = \frac{a+b}{2}$.
- The mean is **not** the 50/50 split of the probability — that is the **median**. On skewed distributions the mean is pulled toward the long tail, because balance depends on distance as well as mass (torque: a far-away mouse can balance a nearby elephant).

# The Sample Mean

You cannot measure the whole world's heights — but you can measure a sample and use its average as an **estimate**. The population mean is called $\mu$; a sample's average is the **sample mean** $\bar{x}$, and each sample you draw gives its own estimate $\bar{x}_1, \bar{x}_2, \dots$. On Statistopia ($N = 10$, $\mu = 160$ cm) we compare three estimates and read off the two morals: samples must be **random**, and **bigger samples estimate better**. (Means and proportions transfer to samples cleanly; the *variance* will need a twist — next video.) It is based on the second video of Lesson 5 (Sampling and Point Estimation).

## 1. Introduction

Want the average height of the entire world population? Take a small sample, average it, and use that as your estimate. The same works for proportions. (For the variance, something curious fails — the sample variance comes out *close* but not right; that story is next.)

This notebook makes the mean-estimation story precise on a population small enough to see whole.

After completing this notebook you will be able to:

- Distinguish the **population mean** $\mu$ from the **sample mean** $\bar{x}$, and compute both.
- Label repeated estimates $\bar{x}_1, \bar{x}_2, \dots$ and understand that each sample gives a different one.
- Explain, with Statistopia's numbers, why non-random samples mislead and why larger $n$ beats smaller $n$.
- Verify by simulation that the sample mean is *centered* on $\mu$ and that its spread shrinks with $n$.

## 2. Theory

### The population mean $\mu$

Statistopia (population version, $N = 10$) has heights, in cm:

$$
151,\ 153,\ 155,\ 157,\ 159,\ 161,\ 163,\ 165,\ 167,\ 169
$$

The **population mean** averages everyone:

$$
\mu = \frac{1}{N}\sum_{i=1}^{N} x_i = \frac{1600}{10} = 160 \text{ cm}
$$

### The sample mean $\bar{x}$

Suppose you can only measure $n = 6$ randomly chosen people — say the ones at $153, 155, 159, 163, 165, 169$. Their average is the **sample mean**:

$$
\bar{x}_1 = \frac{153 + 155 + 159 + 163 + 165 + 169}{6} = 160.67 \text{ cm}
$$

The subscript labels the sample: $\bar{x}_1$ is the first estimate; another sample yields $\bar{x}_2$, and so on. In general, for a sample $x_1, \dots, x_n$:

$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i
$$

### Moral 1: randomness matters

A second sample of the same size $n = 6$ — but consisting of the **six shortest people** — gives

$$
\bar{x}_2 = \frac{151 + 153 + 155 + 157 + 159 + 161}{6} = 156 \text{ cm}
$$

$\bar{x}_1$ is the better estimate, and the reason is the sampling: the second set was not random, so it systematically underestimates (the previous notebook's lesson, now with numbers).

### Moral 2: size matters

Now take a properly random but tiny sample, $n = 2$ — say the people at $155$ and $161$:

$$
\bar{x}_3 = \frac{155 + 161}{2} = 158 \text{ cm}
$$

$\bar{x}_1$ is a better estimate than $\bar{x}_3$ simply because it used 6 people instead of 2. Small samples are at the mercy of who happens to be drawn; larger samples average the luck away. **The bigger the sample size, the better the estimate** — a phenomenon explored in detail in the coming lessons (it has a name: the law of large numbers).

## 3. Worked Examples

### Example 1: the three estimates side by side

| estimate | sample | $n$ | value | error vs $\mu = 160$ |
|---|---|---|---|---|
| $\bar{x}_1$ | random | 6 | 160.67 | $+0.67$ |
| $\bar{x}_2$ | six shortest | 6 | 156.00 | $-4.00$ |
| $\bar{x}_3$ | random | 2 | 158.00 | $-2.00$ |

Same population, three very different answers — quality of the estimate depends on *how* and *how much* you sample.

### Example 2: a sample proportion

The same logic estimates proportions. Suppose 4 of the 10 islanders are taller than 162 cm (true proportion $p = 0.4$). In the random sample $\{153, 155, 159, 163, 165, 169\}$, three of six qualify: sample proportion $\hat{p} = \tfrac{3}{6} = 0.5$ — an estimate of $p$, with the same random-and-large caveats.

### Example 3: averaging away the luck

With $n = 2$, drawing the two extremes $\{151, 169\}$ gives $\bar{x} = 160$ (lucky!), but drawing $\{151, 153\}$ gives $152$ (unlucky). With $n = 6$, no single unlucky person can drag the average that far — one 151 among six people moves the mean by at most a couple of centimeters. That damping is exactly why larger samples are more reliable.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)

population = np.array([151, 153, 155, 157, 159, 161, 163, 165, 167, 169])
mu = population.mean()
print(f"Statistopia heights: {population}")
print(f"population mean mu = {mu} cm")
```

### The video's three estimates

```python
sample_1 = np.array([153, 155, 159, 163, 165, 169])     # random, n = 6
sample_2 = np.sort(population)[:6]                       # the six shortest
sample_3 = np.array([155, 161])                          # random, n = 2

for name, s in [("x_bar_1 (random, n=6)   ", sample_1),
                ("x_bar_2 (shortest, n=6) ", sample_2),
                ("x_bar_3 (random, n=2)   ", sample_3)]:
    print(f"{name}: mean = {s.mean():>7.2f}   error = {s.mean() - mu:+.2f}")
```

### The sample mean is centered on $\mu$ — if you sample randomly

Draw thousands of random samples (with replacement, as good i.i.d. citizens): the *average of the sample means* sits on $\mu$, for any $n$. The biased shortest-6 rule never recovers.

```python
rng = np.random.default_rng(0)

trials = 100_000
for n in [2, 6]:
    means = population[rng.integers(0, 10, size=(trials, n))].mean(axis=1)
    print(f"n = {n}: average of {trials:,} sample means = {means.mean():.3f}"
          f"   spread (std) = {means.std():.3f}")
print(f"biased shortest-6 rule: always {np.sort(population)[:6].mean():.1f}"
      f" -- off by {np.sort(population)[:6].mean() - mu:+.1f} every single time")
```

### Size buys accuracy: the error shrinks like $1/\sqrt{n}$

(A preview of the standard error — the spread of $\bar{x}$ divided into the population's own spread.)

```python
rng = np.random.default_rng(1)

sigma = population.std()
print(f"population std sigma = {sigma:.3f}\n")
print(f"{'n':>4} {'std of x_bar (sim)':>19} {'sigma/sqrt(n)':>14}")
for n in [1, 2, 6, 10, 30, 100]:
    means = population[rng.integers(0, 10, size=(50_000, n))].mean(axis=1)
    print(f"{n:>4} {means.std():>19.3f} {sigma / np.sqrt(n):>14.3f}")
```

### Sample proportions work the same way

```python
rng = np.random.default_rng(2)

tall = population > 162
p_true = tall.mean()
print(f"true proportion taller than 162 cm: p = {p_true}")

sample_1_tall = (sample_1 > 162).mean()
print(f"sample 1 estimate: p_hat = {sample_1_tall:.2f}")

props = tall[rng.integers(0, 10, size=(100_000, 6))].mean(axis=1)
print(f"average of 100,000 random n=6 sample proportions: {props.mean():.4f}")
```

## 5. Visualizations

### The number line: $\mu$ and the three estimates

The video's picture — everyone on the line, with the population mean and the three sample means marked.

```python
fig, ax = plt.subplots(figsize=(11, 4))
ax.axhline(0, color="black", linewidth=1)
ax.plot(population, np.zeros(10), "o", markersize=14, color="tab:blue",
        label="the 10 islanders", zorder=3)

for value, name, color, y in [(mu, "$\\mu$ = 160", "tab:red", 0.55),
                              (sample_1.mean(), "$\\bar{x}_1$ = 160.67", "tab:green", 0.36),
                              (sample_2.mean(), "$\\bar{x}_2$ = 156", "tab:orange", 0.36),
                              (sample_3.mean(), "$\\bar{x}_3$ = 158", "tab:purple", 0.18)]:
    ax.axvline(value, color=color, linestyle="--", linewidth=2, ymax=0.75)
    ax.annotate(name, xy=(value, y), ha="center", fontsize=12, color=color)

ax.set_xlabel("height (cm)")
ax.set_yticks([])
ax.set_ylim(-0.3, 0.75)
ax.set_xticks(population)
ax.grid(True, alpha=0.3, axis="x")
ax.legend(loc="lower right")
ax.set_title("Three sample means, one population mean")
plt.tight_layout()
plt.show()
```

### The distribution of $\bar{x}$: n = 2 vs n = 6

Every random sample gives a different $\bar{x}$; drawing many shows the estimate's own distribution. Both center on $\mu$ — the $n = 6$ one is visibly tighter.

```python
rng = np.random.default_rng(3)

fig, ax = plt.subplots(figsize=(10, 5))
for n, color in [(2, "tab:purple"), (6, "tab:green")]:
    means = population[rng.integers(0, 10, size=(60_000, n))].mean(axis=1)
    ax.hist(means, bins=40, density=True, alpha=0.55, color=color,
            label=f"n = {n} (std {means.std():.2f})")
ax.axvline(mu, color="tab:red", linestyle="--", linewidth=2, label="$\\mu$ = 160")
ax.set_title("Sample means fluctuate around $\\mu$ -- less so for bigger n")
ax.set_xlabel("sample mean (cm)")
ax.set_ylabel("density")
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
```

### Watching an estimate improve as the sample grows

One long sampling run: the running sample mean homes in on $\mu$ as more people are measured.

```python
rng = np.random.default_rng(8)

draws = population[rng.integers(0, 10, size=500)]
running = np.cumsum(draws) / np.arange(1, 501)

plt.figure(figsize=(10, 4.5))
plt.plot(running, color="tab:blue", label="running sample mean")
plt.axhline(mu, color="tab:red", linestyle="--", linewidth=2, label="$\\mu$ = 160")
plt.title("More data, better estimate: the sample mean converges to the population mean")
plt.xlabel("number of people measured (n)")
plt.ylabel("estimate (cm)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 7. Solutions

```python
# Exercise 1: two small sample means
for s in [np.array([151, 157, 167]), np.array([159, 161])]:
    print(f"sample {s}: mean = {s.mean():.2f}, error = {s.mean() - mu:+.2f}")
```

```python
# Exercise 2: the two formulas
print("Exercise 2:")
print("  mu    = (1/N) * sum of ALL population values")
print("  x_bar = (1/n) * sum of the SAMPLED values")
print("The arithmetic is identical (add and divide by the count); what differs is")
print("the input: mu uses the whole population (a fixed number), x_bar uses one")
print("random sample (so it is itself random and varies sample to sample).")
```

```python
# Exercise 3: all 45 samples of size 2
from itertools import combinations
all_means = [np.mean(pair) for pair in combinations(population, 2)]
print(f"Exercise 3: {len(all_means)} samples; average of all sample means = "
      f"{np.mean(all_means):.1f} = mu")
print("Averaged over every possible sample, x_bar hits mu exactly: the sample")
print("mean is an UNBIASED estimator -- individual samples err, but with no")
print("systematic direction.")
```

```python
# Exercise 4: worst cases shrink with n
from itertools import combinations
for n in [2, 6]:
    means_n = [np.mean(c) for c in combinations(population, n)]
    worst = max(means_n, key=lambda m: abs(m - mu))
    print(f"n = {n}: most extreme sample mean = {worst:.1f} "
          f"(distance {abs(worst - mu):.1f} from mu)")
print("With n = 2 you can draw the two shortest (152) or two tallest (168) --")
print("8 cm off either way; with n = 6 even the six shortest average 156, only")
print("4 cm off -- extremes get diluted as n grows.")
```

```python
# Exercise 5: P(|x_bar - mu| <= 2) vs n
rng = np.random.default_rng(4)

ns = [2, 4, 6, 8, 10, 20, 50]
probs = []
for n in ns:
    means = population[rng.integers(0, 10, size=(40_000, n))].mean(axis=1)
    probs.append(np.mean(np.abs(means - mu) <= 2))

plt.figure(figsize=(8, 4.5))
plt.plot(ns, probs, "o-", color="tab:blue")
plt.axhline(0.95, color="tab:red", linestyle="--", label="95%")
plt.title("Probability the estimate lands within 2 cm of the truth")
plt.xlabel("sample size n")
plt.ylabel("P(|x_bar - mu| <= 2)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

for n, p in zip(ns, probs):
    print(f"n = {n:>2}: {p:.3f}")
```

```python
# Exercise 6: mean squared error of the two rules
rng = np.random.default_rng(5)

means_random = population[rng.integers(0, 10, size=(200_000, 6))].mean(axis=1)
mse_random = np.mean((means_random - mu)**2)
mse_biased = (156.0 - mu)**2                    # deterministic: always 156

print(f"Exercise 6: MSE(random n=6)     = {mse_random:.2f}")
print(f"            MSE(shortest-6 rule) = {mse_biased:.2f}")
print("The biased rule is perfectly consistent -- consistently WRONG. Its zero")
print("variance cannot compensate for a 4 cm bias (bias^2 = 16 > ~5.5). Random")
print("sampling wins: small unbiased wobble beats confident systematic error.")
```

## 8. Conclusion

- The **population mean** $\mu = \frac1N \sum x_i$ is the target; the **sample mean** $\bar{x} = \frac1n \sum x_i$ is the estimate, and each sample yields its own $\bar{x}_1, \bar{x}_2, \dots$ (Statistopia: $\mu = 160$; $\bar{x}_1 = 160.67$, $\bar{x}_2 = 156$, $\bar{x}_3 = 158$).
- **Random sampling matters:** the shortest-six sample is off by $-4$ cm *every time* — bias never averages away.
- **Sample size matters:** $\bar{x}$ is centered on $\mu$ and its spread shrinks like $\sigma/\sqrt{n}$ — bigger samples give better estimates (much more on this in the coming lessons).
- The same recipe estimates **proportions** ($\hat p$ from a sample), and averaged over all possible samples $\bar{x}$ equals $\mu$ exactly — the sample mean is *unbiased*.
- Estimating the **variance** from a sample, however, fails in a subtle way — the sample variance comes out close but not right. Fixing it is the next notebook's story.

# The Margin of Error: Critical Values, Standard Errors, and a Flipped Inequality

The last two notebooks promised margins of error; this one computes them. The recipe has three parts: standardize (any normal becomes the **z-distribution**), look up a **critical value** ($z_{0.975} = 1.96$ leaves 2.5% in each tail), and scale by the **standard error** $\sigma/\sqrt{n}$ — giving $\text{margin} = z_{1-\alpha/2}\cdot\sigma/\sqrt{n}$. Then one clean piece of algebra flips "$\bar{X}$ is probably near $\mu$" into "$\mu$ is probably in $\bar{x} \pm \text{margin}$" — the confidence interval. And when the population isn't normal at all, the central limit theorem slips in to save the day. It is based on the third video of Lesson 7 (Confidence Intervals).

## 1. Introduction

You know a confidence interval's two ingredients — a sample mean and a margin of error added to each side — and you know what the margin *responds to* (sample size, confidence level). This notebook derives what it *is*, using nothing but the shape of the normal distribution and a little algebra on inequalities.

After completing this notebook you will be able to:

- Convert any normal distribution to the standard normal (z-distribution) and read positions as **z-scores**.
- Define and look up **critical values** $z_{\alpha/2}$ and $z_{1-\alpha/2}$, and connect them to the significance level $\alpha$.
- Compute the **standard error** $\sigma/\sqrt{n}$ and assemble the margin of error $z_{1-\alpha/2}\cdot\sigma/\sqrt{n}$.
- Run the inequality flip that turns a probability statement about $\bar{X}$ into an interval that bounds $\mu$.
- Explain why the central limit theorem lets the whole construction survive non-normal populations.

## 2. Theory

### The setup

Heights follow $X \sim N(\mu, \sigma^2)$ with $\mu$ unknown (and $\sigma$, for now, magically known). You take one sample of size $n$ and compute $\bar{x}$. It's probably close to $\mu$ — unlucky draws of many short or many tall people push it low or high — and the goal is to *quantify* "close." From the last notebook:

$$
\bar{X} \sim N\!\left(\mu, \frac{\sigma^2}{n}\right)
$$

### z-scores and the z-distribution

For any normal distribution, roughly **68%** of the curve lies within one standard deviation of the mean and roughly **95%** within two. Distances measured in standard deviations are **z-scores**: the point two standard deviations above the mean has $z = 2$; one below has $z = -1$.

The name comes from the **standard normal** (or z-) distribution — mean 0, variance 1 — reached from any normal by standardizing:

$$
Z = \frac{X - \mu}{\sigma}
$$

On the z-distribution, a point's z-score *is* its value: at 2 you are two standard deviations above the mean of 0. This is why the whole topic is usually discussed on the standard normal — the math is simplest there.

### Critical values

To capture *exactly* 95%, two standard deviations is a bit generous; the precise cutoffs are $\pm 1.96$. Such cutoffs are called **critical values**: points inside of which an exact percentage of the distribution is contained. The notation indexes them by the **area to the left**:

- $z_{0.025} = -1.96$ — the point with 2.5% of the curve to its left;
- $z_{0.975} = +1.96$ — the point with 97.5% to its left.

Why 0.025 and 0.975? Symmetry: to keep 95% in the middle you exclude $\alpha = 0.05$, half per tail. In general the two critical values are

$$
z_{\alpha/2} \quad\text{and}\quad z_{1-\alpha/2}
$$

For $\alpha = 0.1$: $z_{0.05} \approx -1.645$ and $z_{0.95} \approx +1.645$, bracketing 90%. Nobody computes these by hand — you use a lookup table or a software library (below, we compute them with the inverse error function).

### From critical value to margin of error

On a non-standardized normal, the critical value counts standard deviations, so multiply by the standard deviation. For $\bar{X}$ the standard deviation is

$$
\sigma_{\bar{X}} = \frac{\sigma}{\sqrt{n}} \qquad \text{(the \textbf{standard error})}
$$

and therefore 95% of all sample means lie within $1.96\,\sigma/\sqrt{n}$ of $\mu$:

$$
\text{margin of error} \;=\; z_{1-\alpha/2}\cdot\frac{\sigma}{\sqrt{n}}
$$

### The flip: bounding $\mu$ instead of $\bar{X}$

What you have so far is a bound on the *sample mean*:

$$
P\!\left(\mu - z_{1-\alpha/2}\tfrac{\sigma}{\sqrt{n}} \;\le\; \bar{X} \;\le\; \mu + z_{1-\alpha/2}\tfrac{\sigma}{\sqrt{n}}\right) = 0.95
$$

But the goal was to bound $\mu$. Rearrange the event inside — the algebra of inequalities, applied to all three parts at once:

1. subtract $\mu$: $\;-z\frac{\sigma}{\sqrt{n}} \le \bar{X} - \mu \le z\frac{\sigma}{\sqrt{n}}$
2. subtract $\bar{X}$: $\;-\bar{X} - z\frac{\sigma}{\sqrt{n}} \le -\mu \le -\bar{X} + z\frac{\sigma}{\sqrt{n}}$
3. multiply by $-1$ (inequalities flip direction!): $\;\bar{X} + z\frac{\sigma}{\sqrt{n}} \ge \mu \ge \bar{X} - z\frac{\sigma}{\sqrt{n}}$
4. read it left-to-right:

$$
P\!\left(\bar{X} - z_{1-\alpha/2}\tfrac{\sigma}{\sqrt{n}} \;\le\; \mu \;\le\; \bar{X} + z_{1-\alpha/2}\tfrac{\sigma}{\sqrt{n}}\right) = 0.95
$$

The same event, re-centered: the confidence interval is the sample mean plus and minus the margin of error, and it bounds $\mu$ with probability $1 - \alpha$.

### What if the population isn't normal?

Everything above used only the distribution of the *sample mean*, never the population's. And the **central limit theorem** (Lesson 5) says that for large enough $n$, $\bar{X}$ is approximately $N(\mu, \sigma^2/n)$ *whatever* the population looks like — even if you don't know how it behaves. So nothing changes: the same margins of error work, so long as the sample is large enough.

## 3. Worked Examples

### Example 1: a complete margin-of-error computation

Statistopia heights, $\sigma = 10$ cm; you measure $n = 49$ people and get $\bar{x} = 172.0$ cm. At 95% ($\alpha = 0.05$):

$$
\text{standard error} = \frac{10}{\sqrt{49}} = 1.43 \text{ cm}, \qquad
\text{margin} = 1.96 \times 1.43 = 2.8 \text{ cm}
$$

$$
\text{confidence interval} = 172.0 \pm 2.8 = [169.2,\; 174.8]
$$

### Example 2: reading the critical-value notation

$z_{0.975}$ asks: *which z-score has 97.5% of the curve to its left?* Answer: $+1.96$. Its mirror $z_{0.025} = -1.96$ has only 2.5% to its left. By symmetry $z_{\alpha/2} = -z_{1-\alpha/2}$ always, which is why one number suffices and the interval is symmetric. Common values:

| confidence | $\alpha$ | $z_{1-\alpha/2}$ |
|---|---|---|
| 90% | 0.10 | 1.645 |
| 95% | 0.05 | 1.960 |
| 99% | 0.01 | 2.576 |

### Example 3: the flip, with numbers

With $\bar{x} = 172.0$ and margin 2.8: "$\bar{X}$ within 2.8 of $\mu$" is the event $\mu - 2.8 \le 172.0 \le \mu + 2.8$. Subtract, negate, flip: $172.0 - 2.8 \le \mu \le 172.0 + 2.8$. Nothing probabilistic happened — the two descriptions are the *same event*, which is why the 95% carries over untouched.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project). The critical-value lookup is the inverse error function: $z_p = \sqrt{2}\,\mathrm{erfinv}(2p - 1)$.

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(9)

def z_score(p):
    """Critical value z_p: the point with probability p to its left."""
    return float(sp.sqrt(2) * sp.erfinv(2 * sp.Float(p) - 1))

def margin_of_error(sigma, n, alpha=0.05):
    return z_score(1 - alpha / 2) * sigma / np.sqrt(n)

def confidence_interval(x_bar, sigma, n, alpha=0.05):
    m = margin_of_error(sigma, n, alpha)
    return x_bar - m, x_bar + m
```

### The lookup table, computed

The critical values the video quotes — and a check of the 68/95 rule of thumb against the exact cutoffs.

```python
print("area-to-left notation:")
for p in (0.025, 0.975, 0.05, 0.95):
    print(f"  z_{p} = {z_score(p):+.3f}")

print("\nconfidence level -> z_(1 - alpha/2):")
for alpha in (0.10, 0.05, 0.01):
    print(f"  {1 - alpha:.0%}: alpha = {alpha:.2f} -> z = {z_score(1 - alpha / 2):.3f}")

within = lambda z: float(sp.erf(z / sp.sqrt(2)))
print(f"\nrule of thumb check: within 1 sd = {within(1):.1%} (~68%), "
      f"within 2 sd = {within(2):.2%} (~95%), within 1.96 sd = {within(1.96):.2%}")
```

### Example 1 end to end — and the flip verified by simulation

Build the interval from Example 1, then check that the two descriptions of the event — "$\bar{X}$ lands within the margin of $\mu$" and "the interval $\bar{X} \pm$ margin captures $\mu$" — are literally the same event, sample by sample.

```python
MU_TRUE, SIGMA, N = 170.0, 10.0, 49

sample = rng.normal(MU_TRUE, SIGMA, size=N)
x_bar = sample.mean()
lower, upper = confidence_interval(x_bar, SIGMA, N)
print(f"n = {N}: x-bar = {x_bar:.1f}, standard error = {SIGMA / np.sqrt(N):.2f}, "
      f"margin = {margin_of_error(SIGMA, N):.2f}")
print(f"95% confidence interval: [{lower:.1f}, {upper:.1f}]")

# the flip: same event, two phrasings
m = margin_of_error(SIGMA, N)
xbars = rng.normal(MU_TRUE, SIGMA / np.sqrt(N), size=100_000)
event_xbar_near_mu = np.abs(xbars - MU_TRUE) <= m       # bound on x-bar
event_interval_has_mu = (xbars - m <= MU_TRUE) & (MU_TRUE <= xbars + m)  # bound on mu
print(f"\nP(x-bar within margin of mu)  = {event_xbar_near_mu.mean():.3f}")
print(f"P(interval captures mu)       = {event_interval_has_mu.mean():.3f}")
print(f"identical outcome in every sample: {(event_xbar_near_mu == event_interval_has_mu).all()}")
```

## 5. Visualizations

### The z-distribution and its critical values

95% of the standard normal lies between $z_{0.025} = -1.96$ and $z_{0.975} = +1.96$; tightening to $\pm 1.645$ keeps only 90%.

```python
z_axis = np.linspace(-4, 4, 600)
z_density = np.exp(-0.5 * z_axis**2) / np.sqrt(2 * np.pi)

plt.figure(figsize=(10, 4.8))
plt.plot(z_axis, z_density, color="black", linewidth=1.5)
z95 = z_score(0.975)
inside = np.abs(z_axis) <= z95
plt.fill_between(z_axis, z_density, where=inside, color="tab:green", alpha=0.35,
                 label=r"95% between $z_{0.025}$ and $z_{0.975}$")
plt.fill_between(z_axis, z_density, where=~inside, color="tab:red", alpha=0.35,
                 label=r"$\alpha/2$ = 2.5% per tail")
for z, name in [(-z95, r"$z_{0.025} = -1.96$"), (z95, r"$z_{0.975} = +1.96$")]:
    plt.axvline(z, color="tab:red", linestyle="--", linewidth=1.2)
    plt.annotate(name, xy=(z, 0.30), ha="center", fontsize=10)
z90 = z_score(0.95)
for z in (-z90, z90):
    plt.axvline(z, color="tab:orange", linestyle=":", linewidth=1.2)
plt.annotate(r"$\pm z_{0.95} = \pm 1.645$ (90%)", xy=(0, 0.06),
             ha="center", fontsize=10, color="tab:orange")
plt.xlabel("z")
plt.ylabel("density")
plt.title("Critical values are cutoffs: subscript = area to the left")
plt.grid(True, alpha=0.3)
plt.legend(loc="upper left", fontsize=9)
plt.show()
```

### Standardizing: every normal is the z-distribution in disguise

Subtract the mean, divide by the standard deviation. The height curve and the z-curve carry identical information — z-scores just relabel the axis.

```python
fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.2))

heights = np.linspace(130, 210, 500)
h_density = np.exp(-0.5 * ((heights - 170) / 10) ** 2) / (10 * np.sqrt(2 * np.pi))
axes[0].plot(heights, h_density, color="tab:blue", linewidth=2)
axes[0].set_title(r"$X \sim N(170, 10^2)$: heights in cm", fontsize=10)
axes[0].set_xlabel("height (cm)")
axes[0].set_ylim(0, 0.048)
for k in (-2, -1, 0, 1, 2):
    axes[0].axvline(170 + 10 * k, color="gray", linestyle=":", linewidth=1)
    axes[0].annotate(f"z = {k:+d}" if k else r"$\mu$", xy=(170 + 10 * k, 0.0435),
                     ha="center", fontsize=8)

axes[1].plot(z_axis, z_density, color="tab:purple", linewidth=2)
axes[1].set_title(r"$Z = (X - \mu)/\sigma \sim N(0, 1)$: the z-distribution", fontsize=10)
axes[1].set_xlabel("z")
for k in (-2, -1, 0, 1, 2):
    axes[1].axvline(k, color="gray", linestyle=":", linewidth=1)
for ax in axes:
    ax.set_ylabel("density")
    ax.grid(True, alpha=0.3)
plt.suptitle("Standardization: same curve, axis measured in standard deviations")
plt.tight_layout()
plt.show()
```

### The central limit theorem saves the day

A wildly non-normal population — exponential waiting times, all skew — and yet the sample means of $n = 40$ line up on the promised normal curve, and the usual 95% recipe still covers at 95%.

```python
pop_mean = 5.0                               # exponential with mean 5: skewed hard right
n = 40
sample_means = rng.exponential(pop_mean, size=(20_000, n)).mean(axis=1)

fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.4))

x_pop = np.linspace(0, 25, 400)
axes[0].plot(x_pop, np.exp(-x_pop / pop_mean) / pop_mean, color="tab:red", linewidth=2)
axes[0].set_title("The population: exponential (nothing normal here)", fontsize=10)
axes[0].set_xlabel("waiting time")

axes[1].hist(sample_means, bins=60, density=True, color="tab:green", alpha=0.6,
             label=r"observed $\bar{X}$, n = 40")
se = pop_mean / np.sqrt(n)                    # exponential: sigma = mean
x_clt = np.linspace(sample_means.min(), sample_means.max(), 400)
axes[1].plot(x_clt, np.exp(-0.5 * ((x_clt - pop_mean) / se) ** 2) / (se * np.sqrt(2 * np.pi)),
             color="black", linewidth=2, label=r"CLT: $N(\mu, \sigma^2/n)$")
axes[1].set_title("The sample means: normal anyway", fontsize=10)
axes[1].set_xlabel(r"$\bar{x}$")
axes[1].legend(fontsize=9)
for ax in axes:
    ax.set_ylabel("density")
    ax.grid(True, alpha=0.3)
plt.suptitle("The recipe never needed a normal population -- only a normal sample mean")
plt.tight_layout()
plt.show()

coverage = (np.abs(sample_means - pop_mean) <= margin_of_error(pop_mean, n)).mean()
print(f"95% recipe coverage on the exponential population (n = {n}): {coverage:.1%}")
```

```python
# Exercise 3: 90% interval for wait times
sigma_w, n_w, xbar_w, alpha_w = 4.0, 25, 12.3, 0.10
se_w = sigma_w / np.sqrt(n_w)
z_w = z_score(1 - alpha_w / 2)
m_w = z_w * se_w
print(f"standard error = 4/sqrt(25) = {se_w}")
print(f"critical value z_0.95 = {z_w:.3f}")
print(f"margin = {z_w:.3f} * {se_w} = {m_w:.3f}")
print(f"interval: 12.3 +/- {m_w:.2f} = [{xbar_w - m_w:.2f}, {xbar_w + m_w:.2f}]")
print("helper agrees:", tuple(round(v, 2)
      for v in confidence_interval(xbar_w, sigma_w, n_w, alpha_w)))
```

```python
# Exercise 4: the flip, symbolically
mu, xbar_s, m_s = sp.symbols("mu xbar m", real=True, positive=True)
flipped = sp.reduce_inequalities([mu - m_s <= xbar_s, xbar_s <= mu + m_s], mu)
print("Exercise 4:", flipped)
print("SymPy isolates mu between xbar - m and xbar + m -- the confidence interval.")
```

```python
# Exercise 5: Bernoulli coverage
p_true, n_b = 0.3, 100
sigma_b = np.sqrt(p_true * (1 - p_true))
m_b = margin_of_error(sigma_b, n_b)

samples_b = rng.random((20_000, n_b)) < p_true
p_hats = samples_b.mean(axis=1)
coverage_b = (np.abs(p_hats - p_true) <= m_b).mean()
print(f"Exercise 5: margin = 1.96 * sqrt(0.3 * 0.7 / 100) = {m_b:.3f}")
print(f"coverage of p = 0.3: {coverage_b:.1%}  (CLT working on a Bernoulli population)")
```

```python
# Exercise 6: the tail promises vs sample size for the skewed population
sizes = (2, 5, 10, 30, 100)
miss_low, miss_high, total = [], [], []
for n_e in sizes:
    means_e = rng.exponential(pop_mean, size=(100_000, n_e)).mean(axis=1)
    m_e = margin_of_error(pop_mean, n_e)
    miss_low.append((means_e < pop_mean - m_e).mean())
    miss_high.append((means_e > pop_mean + m_e).mean())
    total.append(1 - miss_low[-1] - miss_high[-1])

plt.figure(figsize=(9, 4.4))
plt.plot(sizes, miss_high, "o-", color="tab:red", linewidth=2,
         label="miss high (x-bar too large)")
plt.plot(sizes, miss_low, "o-", color="tab:blue", linewidth=2,
         label="miss low (x-bar too small)")
plt.axhline(0.025, color="black", linestyle=":", label=r"promised $\alpha/2$ = 2.5%")
plt.xscale("log")
plt.xticks(sizes, sizes)
plt.xlabel("sample size n (log scale)")
plt.ylabel("tail miss rate")
plt.title("Exercise 6: total coverage hides the broken 2.5%/2.5% tail promise")
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.show()

print(f"{'n':>4s} {'coverage':>9s} {'miss low':>9s} {'miss high':>10s}")
for n_e, cov, lo, hi in zip(sizes, total, miss_low, miss_high):
    print(f"{n_e:>4d} {cov:>9.1%} {lo:>9.2%} {hi:>10.2%}")
print("\nTotal coverage looks fine even at n = 2 -- but only by accident: the low")
print("tail is IMPOSSIBLE there (mu - margin < 0 and an exponential mean can't be")
print("negative), while the long right skew misses high nearly 5% of the time.")
print("The symmetric normal margins have the wrong shape; the two tails only")
print("approach the promised 2.5% each as the CLT kicks in around n ~ 100.")
```

## 8. Conclusion

- **z-scores** measure position in standard deviations; standardizing $Z = (X - \mu)/\sigma$ turns every normal into the z-distribution, where ~68% lies within 1 and ~95% within 2 (exactly 1.96).
- **Critical values** are cutoffs indexed by area-to-left: $z_{\alpha/2}$ and $z_{1-\alpha/2}$ bracket exactly $1 - \alpha$ of the curve ($\pm 1.96$ for 95%, $\pm 1.645$ for 90%) — looked up, never hand-computed.
- The **standard error** is $\sigma/\sqrt{n}$, and the **margin of error** is $z_{1-\alpha/2} \cdot \sigma/\sqrt{n}$.
- The **flip**: subtracting $\mu$, subtracting $\bar{X}$, and multiplying by $-1$ (directions reverse!) turns the bound on $\bar{X}$ into $P(\bar{X} - \text{margin} \le \mu \le \bar{X} + \text{margin}) = 1 - \alpha$ — the same event, so the same probability, now bounding what you care about.
- Non-normal population? The construction only ever used the distribution of $\bar{X}$, and the **central limit theorem** makes that approximately normal for large $n$ — same parameters, same margins, nothing changes.

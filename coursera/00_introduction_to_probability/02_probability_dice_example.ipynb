# What Is Probability? — Dice Example

This notebook reinforces the concept of probability using the experiment of rolling a fair six-sided die: first one die, then two dice, and finally a simulation showing how observed frequencies approach theoretical probabilities. It is based on the second video of Week 1 (Introduction to Probability).


## 1. Introduction

In the previous notebook you learned the basic probability formula

$$
P(E) = \frac{\text{number of favorable outcomes}}{\text{total number of possible outcomes}}
$$

Here we apply it to a new experiment — **rolling dice** — and use it to practice identifying the sample space and the event.

After completing this notebook you will be able to:

- Compute probabilities for a single die roll.
- Enumerate the full sample space of rolling two dice and compute joint probabilities.
- Show with a simulation that the frequency of an outcome approaches its theoretical probability as the number of rolls grows.


## 2. Theory

Rolling a fair six-sided die is an **experiment**: a process with an uncertain outcome.

- The **sample space** is $S = \{1, 2, 3, 4, 5, 6\}$, so $|S| = 6$.
- *Fair* means all six outcomes are **equally likely**.

For an event $E$, the probability is $P(E) = |E| / |S|$.

### Two dice

When rolling two dice, for each of the 6 outcomes of the first die there are 6 possibilities for the second die. The sample space is all ordered pairs

$$
S = \{(1,1),\ (1,2),\ (1,3),\ \dots,\ (6,6)\}, \qquad |S| = 6 \times 6 = 36
$$

This multiplication of possibilities is the same pattern we saw with coins: repeating an experiment multiplies the sizes of the sample spaces.


## 3. Worked Examples

### Example 1: Rolling a six with one die

What is $P(6)$?

- Six things can happen, all equally likely: $|S| = 6$.
- One outcome is favorable (rolling a 6): $|E| = 1$.

$$
P(6) = \frac{1}{6} \approx 0.167
$$

### Example 2: Rolling six-six with two dice

What is $P(6, 6)$?

- The sample space has $36$ outcomes: $(1,1), (1,2), (1,3), \dots, (6,6)$.
- Only **one** of them is favorable: $(6, 6)$.

$$
P(6,6) = \frac{1}{36} \approx 0.028
$$


## 4. Python Examples


```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import product

rng = np.random.default_rng(seed=42)
```

### Example 1 in code: one die


```python
sample_space = [1, 2, 3, 4, 5, 6]
event = [outcome for outcome in sample_space if outcome == 6]

p_six = len(event) / len(sample_space)
print(f"P(6) = {len(event)}/{len(sample_space)} = {p_six:.4f}")
```

### Example 2 in code: two dice

We enumerate the full sample space with `itertools.product`, just like listing the 36 pairs by hand.


```python
two_dice = list(product(range(1, 7), repeat=2))
event = [pair for pair in two_dice if pair == (6, 6)]

print(f"Size of the sample space: {len(two_dice)}")
print(f"Favorable outcomes: {event}")
print(f"P(6,6) = {len(event)}/{len(two_dice)} = {len(event) / len(two_dice):.4f}")
```

### Frequencies approach probabilities

The video points to an interactive tool where you roll a die repeatedly and watch the frequency of each outcome approach its theoretical probability. We can reproduce that experiment in code: roll a die many times and track how often each face appears.


```python
n_rolls = 10_000
rolls = rng.integers(1, 7, size=n_rolls)

faces, counts = np.unique(rolls, return_counts=True)
for face, count in zip(faces, counts):
    print(f"Face {face}: frequency {count / n_rolls:.4f}   (theoretical {1/6:.4f})")
```

## 5. Visualizations

### The sample space of two dice

Each cell in the 6×6 grid is one equally likely outcome. The event "six-six" is a single cell out of 36.


```python
fig, ax = plt.subplots(figsize=(6, 6))

for first in range(1, 7):
    for second in range(1, 7):
        is_event = (first, second) == (6, 6)
        color = "#e15759" if is_event else "#d8e6f2"
        rect = plt.Rectangle((second - 1, 6 - first), 1, 1, facecolor=color,
                             edgecolor="white", linewidth=2)
        ax.add_patch(rect)
        ax.text(second - 0.5, 6 - first + 0.5, f"{first},{second}",
                ha="center", va="center", fontsize=9)

ax.set_xlim(0, 6)
ax.set_ylim(0, 6)
ax.set_xticks([])
ax.set_yticks([])
ax.set_aspect("equal")
ax.set_title("Sample space of two dice: 36 outcomes, event (6,6) in red")
plt.show()
```

### Frequency of each face over many rolls

With 10,000 rolls, all six bars sit close to the theoretical probability $1/6$.


```python
plt.figure(figsize=(8, 5))
plt.bar(faces, counts / n_rolls, color="#4e79a7", label="Observed frequency")
plt.axhline(1 / 6, color="red", linestyle="--", label="Theoretical P = 1/6")
plt.title(f"Frequencies of each face after {n_rolls:,} rolls")
plt.xlabel("Die face")
plt.ylabel("Frequency")
plt.grid(True, alpha=0.3, axis="y")
plt.legend()
plt.show()
```

### Convergence of the frequency of rolling a six

Early on the frequency fluctuates a lot; as the number of rolls grows, it settles at $1/6$.


```python
is_six = rolls == 6
running_frequency = np.cumsum(is_six) / np.arange(1, n_rolls + 1)

plt.figure(figsize=(8, 5))
plt.plot(running_frequency, label="Frequency of rolling a 6")
plt.axhline(1 / 6, color="red", linestyle="--", label="Theoretical P(6) = 1/6")
plt.title("Frequency of a six approaches its probability")
plt.xlabel("Number of rolls")
plt.ylabel("Frequency")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 6. Exercises

### Basic

1. What is the probability of rolling an odd number with one fair die?
2. What is the probability of rolling a number greater than 4 with one fair die?

### Intermediate

3. With two dice, what is the probability that **both** show the same number (a "double")?
4. With two dice, what is the probability that the **sum** of the two faces is 7?

### Advanced

5. With three dice, what is the probability that all three show a six? Verify your answer with a simulation.


## 7. Solutions


```python
# Exercise 1: odd numbers {1, 3, 5}
print("1.  P(odd) = 3/6 =", 3 / 6)

# Exercise 2: greater than 4 -> {5, 6}
print("2.  P(>4) = 2/6 =", round(2 / 6, 4))

# Exercise 3: doubles -> (1,1), (2,2), ..., (6,6)
two_dice = list(product(range(1, 7), repeat=2))
doubles = [p for p in two_dice if p[0] == p[1]]
print("3.  P(double) =", len(doubles), "/ 36 =", round(len(doubles) / 36, 4))

# Exercise 4: sum equals 7 -> (1,6), (2,5), (3,4), (4,3), (5,2), (6,1)
sum_seven = [p for p in two_dice if sum(p) == 7]
print("4.  P(sum = 7) =", len(sum_seven), "/ 36 =", round(len(sum_seven) / 36, 4))

# Exercise 5: three sixes -> 1 / 6**3
rolls = rng.integers(1, 7, size=(200_000, 3))
simulated = np.all(rolls == 6, axis=1).mean()
print(f"5.  Theoretical 1/216 = {1 / 216:.6f}, simulated = {simulated:.6f}")
```

## 8. Conclusion

- A fair die has a sample space of 6 equally likely outcomes, so $P(6) = 1/6$.
- Rolling two dice gives $6 \times 6 = 36$ outcomes; the single favorable outcome $(6,6)$ has probability $1/36$.
- Repeating an experiment multiplies sample-space sizes — the same pattern as flipping several coins.
- Observed frequencies fluctuate for small numbers of trials but **converge to the theoretical probability** as the number of rolls grows.

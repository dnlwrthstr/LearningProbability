# What Is Probability?

This notebook introduces the concept of probability: what it measures, the vocabulary used to talk about it (experiment, outcome, event, sample space), and how to compute probabilities by counting outcomes. It is based on the first video of Week 1 (Introduction to Probability).


## 1. Introduction

Probability is the measure of **how likely an event is to occur**.

- If you throw a fair coin, the probability that it lands on heads is $\frac{1}{2}$ (50%).
- If you throw a fair die, the probability that it lands on the number 4 is $\frac{1}{6}$.

Understanding probability at this basic level is the foundation for more complex real-world applications in **machine learning and data science**, where uncertainty is everywhere.

After completing this notebook you will be able to:

- Define experiment, outcome, event, and sample space.
- Compute the probability of an event by counting favorable outcomes.
- Enumerate the outcomes of repeated experiments (e.g. flipping several coins).
- Verify theoretical probabilities with simple Python simulations.


## 2. Theory

### Key definitions

| Term | Meaning |
|---|---|
| **Experiment** | Any process that produces an uncertain outcome (e.g. flipping a coin). |
| **Outcome** | A single possible result of the experiment (e.g. heads). |
| **Sample space** | The set of *all* possible outcomes. |
| **Event** | The group of outcomes you are interested in (a subset of the sample space). |

### The probability formula

For an experiment where all outcomes are equally likely, the probability of an event $E$ is

$$
P(E) = \frac{\text{number of favorable outcomes}}{\text{total number of possible outcomes}} = \frac{|E|}{|S|}
$$

where $|E|$ is the size of the event and $|S|$ is the size of the sample space.

### Notation

We write $P(\text{soccer})$ for "the probability that a randomly picked kid plays soccer", $P(H)$ for "the probability that a coin lands on heads", and so on.

A probability is always a number between $0$ (impossible) and $1$ (certain). It can be expressed as a fraction ($\frac{3}{10}$), a decimal ($0.3$), or a percentage ($30\%$).


## 3. Worked Examples

### Example 1: Picking a kid who plays soccer

A school has **10 kids**. **3 play soccer**, 7 don't. You pick one kid at random. What is $P(\text{soccer})$?

- The **sample space** is all 10 kids: $|S| = 10$.
- The **event** is the kids who play soccer: $|E| = 3$.

$$
P(\text{soccer}) = \frac{\text{kids who play soccer}}{\text{total kids}} = \frac{3}{10} = 0.3 = 30\%
$$

### Example 2: Flipping one fair coin

Flipping a coin is an **experiment**: the outcome is uncertain. A *fair* coin means each outcome (heads or tails) is equally likely.

$$
P(H) = \frac{1}{2} = 0.5
$$

### Example 3: Flipping two coins

What is the probability that **both** coins land on heads?

Walk through the experiment: the first coin can land H or T, and *for each* of those, the second coin can land H or T. The sample space is

$$
S = \{HH,\ HT,\ TH,\ TT\}, \qquad |S| = 4
$$

Only one outcome is favorable ($HH$), so

$$
P(HH) = \frac{1}{4} = 0.25 = 25\%
$$

### Example 4: Flipping three coins

Now the sample space has $2 \times 2 \times 2 = 8$ outcomes:

$$
S = \{HHH,\ HHT,\ HTH,\ HTT,\ THH,\ THT,\ TTH,\ TTT\}
$$

Only $HHH$ is favorable, so

$$
P(HHH) = \frac{1}{8} = 0.125
$$

Notice the pattern: each extra coin doubles the number of outcomes and halves the probability that *all* coins land on heads.


## 4. Python Examples


```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import product

rng = np.random.default_rng(seed=42)
```

### Example 1 in code: the soccer problem


```python
kids_playing_soccer = 3   # size of the event
total_kids = 10           # size of the sample space

p_soccer = kids_playing_soccer / total_kids
print(f"P(soccer) = {kids_playing_soccer}/{total_kids} = {p_soccer} = {p_soccer:.0%}")
```

### Examples 2-4 in code: enumerating coin flips

`itertools.product` generates every combination of outcomes, exactly like walking the branches of the tree by hand.


```python
for n_coins in [1, 2, 3]:
    sample_space = list(product("HT", repeat=n_coins))
    all_heads = [outcome for outcome in sample_space if all(c == "H" for c in outcome)]

    outcomes_text = ", ".join("".join(o) for o in sample_space)
    p = len(all_heads) / len(sample_space)

    print(f"{n_coins} coin(s): sample space = {{{outcomes_text}}}")
    print(f"  P(all heads) = {len(all_heads)}/{len(sample_space)} = {p}\n")
```

### Checking the theory with a simulation

If the formula is right, flipping two coins many times should give "both heads" in about 25% of the trials. This is called a **Monte Carlo simulation**.


```python
n_trials = 100_000

# Each row is one trial of flipping two coins: 1 = heads, 0 = tails
flips = rng.integers(0, 2, size=(n_trials, 2))
both_heads = np.all(flips == 1, axis=1)

estimate = both_heads.mean()
print(f"Simulated P(HH) over {n_trials:,} trials: {estimate:.4f}")
print(f"Theoretical P(HH):                       {1/4:.4f}")
```

## 5. Visualizations

### The Venn diagram view of the soccer problem

The green rectangle is the **sample space** (100% of the population, all 10 kids). The circle is the **event**: the 30% of kids who play soccer. Kids who don't play soccer are outside the circle but still inside the rectangle, because they are still part of the population.


```python
fig, ax = plt.subplots(figsize=(8, 5))

# Sample space: all kids
rectangle = plt.Rectangle((0, 0), 10, 6, facecolor="#d8f0d8",
                          edgecolor="green", linewidth=2)
ax.add_patch(rectangle)

# Event: kids who play soccer
circle = plt.Circle((3, 3), 1.8, facecolor="#7fc97f",
                    edgecolor="darkgreen", linewidth=2)
ax.add_patch(circle)

ax.text(3, 3, "Play soccer\n3 kids (30%)", ha="center", va="center", fontsize=11)
ax.text(7.3, 3, "Don't play soccer\n7 kids (70%)", ha="center", va="center", fontsize=11)
ax.text(0.2, 5.6, "Sample space: all 10 kids (100%)", fontsize=11)

ax.set_xlim(-0.5, 10.5)
ax.set_ylim(-0.5, 6.5)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("Event and sample space as a Venn diagram")
plt.show()
```

### Convergence of the simulation

The more trials we run, the closer the simulated frequency of $HH$ gets to the theoretical value $0.25$.


```python
n_trials = 10_000
flips = rng.integers(0, 2, size=(n_trials, 2))
both_heads = np.all(flips == 1, axis=1)

running_estimate = np.cumsum(both_heads) / np.arange(1, n_trials + 1)

plt.figure(figsize=(8, 5))
plt.plot(running_estimate, label="Simulated frequency of HH")
plt.axhline(0.25, color="red", linestyle="--", label="Theoretical P(HH) = 0.25")
plt.title("Simulated probability converges to the theoretical value")
plt.xlabel("Number of trials")
plt.ylabel("Frequency of both coins landing heads")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 7. Solutions


```python
# Exercise 1: kid does not play soccer
print("1.  P(not soccer) = 7/10 =", 7 / 10)

# Exercise 2: even number on a die -> favorable outcomes {2, 4, 6}
print("2.  P(even) = 3/6 =", 3 / 6)

# Exercise 3: four coins
sample_space = list(product("HT", repeat=4))
print("3.  |S| =", len(sample_space), "and P(HHHH) = 1/16 =", 1 / 16)

# Exercise 4: exactly one head in two flips -> {HT, TH}
favorable = [o for o in product("HT", repeat=2) if o.count("H") == 1]
print("4.  P(exactly one head) =", len(favorable), "/ 4 =", len(favorable) / 4)

# Exercise 5: P(all heads with n coins) = 1 / 2**n
n = 5
flips = rng.integers(0, 2, size=(200_000, n))
simulated = np.all(flips == 1, axis=1).mean()
print(f"5.  Theoretical 1/2**{n} = {1 / 2**n}, simulated = {simulated:.4f}")
```

## 8. Conclusion

- **Probability** measures how likely an event is: $P(E) = |E| / |S|$ when all outcomes are equally likely.
- An **experiment** is any process with an uncertain outcome; the **sample space** contains all possible outcomes; the **event** is the subset you care about.
- For repeated experiments, enumerate outcomes systematically: $n$ coin flips give $2^n$ outcomes, so $P(\text{all heads}) = 1/2^n$.
- Simulations are a powerful way to check theoretical results: simulated frequencies converge to true probabilities as the number of trials grows.

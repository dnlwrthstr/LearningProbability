# Prior, Event, and Posterior

This short notebook formalizes the vocabulary behind everything in the Bayes lessons: the **prior** (what you know before), the **event** (the information you receive), and the **posterior** (your updated probability). It revisits all four running examples through this single lens. It is based on the thirteenth video of Week 1 (Introduction to Probability).


## 1. Introduction

Every Bayesian computation has the same three-part structure:

1. **Prior** — $P(A)$: the original probability, calculated knowing nothing else. The most basic unit of information in the problem.
2. **Event** — $E$: something happens that gives you information.
3. **Posterior** — $P(A \mid E)$: the probability recalculated using that information.

The posterior is always a **better estimate** than the prior, because the event narrowed down where you live in the sample space.

After completing this notebook you will be able to:

- Name the prior, event, and posterior in any conditional-probability problem.
- See the four examples from previous notebooks as one repeated pattern.


## 2. Theory

$$
\underbrace{P(A)}_{\text{prior}} \;\xrightarrow{\;\text{event } E\;}\; \underbrace{P(A \mid E)}_{\text{posterior}}
$$

Conditioning replaces the full sample space with the part where $E$ holds. The posterior can be larger or smaller than the prior — the event's information can point either way.

### The four examples, unified

| Example | Prior $P(A)$ | Event $E$ | Posterior $P(A \mid E)$ |
|---|---|---|---|
| Spam (nb 12) | $P(\text{spam}) = 0.2$ | contains "lottery" | $\frac{14}{24} \approx 0.583$ |
| Disease (nb 10-11) | $P(\text{sick}) = 0.0001$ | tested positive | $\frac{99}{10098} \approx 0.0098$ |
| Two dice (nb 08) | $P(\text{sum}=10) = \frac{3}{36}$ | first die is 6 | $\frac{1}{6}$ |
| Two coins (nb 08) | $P(HH) = \frac{1}{4}$ | first coin is heads | $\frac{1}{2}$ |

In the spam case the prior was "spam emails over all emails"; after the event it becomes "spam emails *with lottery* over all emails *with lottery*" — the same ratio, computed in the smaller world the event left behind.


## 3. Python Example

All four updates in one place.


```python
import matplotlib.pyplot as plt

examples = [
    ("Spam email\n('lottery')",  0.2,     14 / 24),
    ("Disease test\n(positive)", 0.0001,  99 / 10_098),
    ("Dice sum = 10\n(1st is 6)", 3 / 36, 1 / 6),
    ("Two heads\n(1st is H)",    1 / 4,   1 / 2),
]

for name, prior, posterior in examples:
    factor = posterior / prior
    print(f"{name.replace(chr(10), ' '):28s} prior {prior:>8.4f} -> "
          f"posterior {posterior:>7.4f}   (x{factor:.0f})")
```

## 4. Visualization

Each event multiplied the probability — by ~3 for spam, ~98 for the disease test, ~6 for the dice, and ~2 for the coins. (Log scale, since the sizes differ so much.)


```python
labels = [e[0] for e in examples]
priors = [e[1] for e in examples]
posteriors = [e[2] for e in examples]

x = range(len(examples))
w = 0.35

plt.figure(figsize=(9, 5))
plt.bar([i - w / 2 for i in x], priors, width=w, label="Prior", color="#9ecae1")
plt.bar([i + w / 2 for i in x], posteriors, width=w, label="Posterior", color="#e15759")
plt.yscale("log")
plt.xticks(list(x), labels)
plt.ylabel("Probability (log scale)")
plt.title("The same pattern everywhere: an event updates the prior into the posterior")
plt.grid(True, alpha=0.3, axis="y", which="both")
plt.legend()
plt.show()
```

## 6. Solutions

1. **Higher.** Sharing a birth month is implied by sharing a birthday, and the information removes many of the "all-spread-out" configurations — the remaining world is more collision-prone.
2. Any feature more common in ham than spam — e.g. the email is a reply to one you sent, or it contains a colleague's name. The posterior $P(\text{spam} \mid E)$ then drops below the 0.2 prior.
3. Given a first roll of 3, the sum is 10 only if the second die shows 7 — impossible. The posterior is $0$, *lower* than the prior $\frac{3}{36} \approx 0.083$: events can push probabilities down as well as up.


## 7. Conclusion

- **Prior → event → posterior** is the universal shape of Bayesian updating; every example in this course so far fits it.
- The posterior is the prior recomputed inside the smaller world that the event left behind — always a better-informed estimate, whether it goes up (spam, dice, coins) or down (a first roll of 3, a negative test).

# The Complement of an Event

This notebook introduces the **complement** of an event — the probability that an event does *not* occur — and the **complement rule** $P(A') = 1 - P(A)$. It is based on the third video of Week 1 (Introduction to Probability).


## 1. Introduction

Sometimes it is easier to compute the probability that an event does **not** happen than the probability that it does. For example, if an event occurs with probability 75%, the probability that it does not occur is 25%.

The complement rule turns "does not happen" questions into simple subtractions, which becomes especially powerful when the direct count of favorable outcomes is large or awkward.

After completing this notebook you will be able to:

- Define the complement $A'$ of an event $A$.
- State and apply the complement rule $P(A') = 1 - P(A)$.
- Use the rule to simplify probability calculations for coins and dice.


## 2. Theory

### The complement of an event

The **complement** of an event $A$, written $A'$ (read "A prime"), is the event that $A$ does **not** happen. Every outcome in the sample space belongs to exactly one of $A$ or $A'$.

### The complement rule

$$
P(A') = 1 - P(A)
$$

Equivalently, $P(A) + P(A') = 1$: an event and its complement together cover the whole sample space, whose total probability is 1 (100%).

### The Venn diagram view

In the Venn diagram from the first notebook, the rectangle is the sample space (100%) and the circle is the event $A$. The complement $A'$ is **everything outside the circle** but still inside the rectangle.


## 3. Worked Examples

### Example 1: The kid does not play soccer

In the school with 10 kids, 3 play soccer and 7 don't. What is $P(\text{not soccer})$?

**Direct count:** count the kids who don't play soccer and divide by the total:

$$
P(\text{not soccer}) = \frac{7}{10} = 0.7
$$

**Complement rule:** we already know $P(\text{soccer}) = 0.3$, and the two probabilities must add to 1:

$$
P(\text{not soccer}) = 1 - P(\text{soccer}) = 1 - 0.3 = 0.7
$$

Both approaches agree — but the second one needs no new counting.

### Example 2: Not obtaining three heads

Flip three coins. What is the probability of **not** obtaining three heads?

From the first notebook, $P(HHH) = \frac{1}{8}$. By the complement rule:

$$
P(\text{not } HHH) = 1 - \frac{1}{8} = \frac{7}{8}
$$

You can also see this directly: of the 8 possible outcomes, the 7 "good" cases are all outcomes except $HHH$, so the probability is $\frac{7}{8}$.

### Example 3: Rolling anything different than a six

Throw one die. What is the probability of obtaining anything different than 6?

There are 6 outcomes and $P(6) = \frac{1}{6}$, so

$$
P(\text{not } 6) = 1 - \frac{1}{6} = \frac{6}{6} - \frac{1}{6} = \frac{5}{6}
$$

Out of the six faces, we remove the one showing a 6 and five favorable outcomes remain.


## 4. Python Examples


```python
import numpy as np
import matplotlib.pyplot as plt
from itertools import product

rng = np.random.default_rng(seed=42)
```

### Example 1 in code: direct count vs. complement rule


```python
total_kids = 10
kids_playing_soccer = 3

p_soccer = kids_playing_soccer / total_kids

p_not_soccer_direct = (total_kids - kids_playing_soccer) / total_kids
p_not_soccer_rule = 1 - p_soccer

print(f"Direct count:     P(not soccer) = 7/10 = {p_not_soccer_direct}")
print(f"Complement rule:  P(not soccer) = 1 - {p_soccer} = {p_not_soccer_rule}")
```

### Example 2 in code: not three heads

We can verify the complement rule by enumerating the full sample space.


```python
sample_space = list(product("HT", repeat=3))
not_three_heads = [o for o in sample_space if o != ("H", "H", "H")]

p_direct = len(not_three_heads) / len(sample_space)
p_rule = 1 - 1 / 8

print(f"Outcomes that are not HHH: {len(not_three_heads)} of {len(sample_space)}")
print(f"Direct count:     P(not HHH) = {p_direct}")
print(f"Complement rule:  P(not HHH) = 1 - 1/8 = {p_rule}")
```

### Example 3 in code: not a six, checked by simulation


```python
p_not_six_rule = 1 - 1 / 6

n_rolls = 100_000
rolls = rng.integers(1, 7, size=n_rolls)
p_not_six_simulated = (rolls != 6).mean()

print(f"Complement rule:  P(not 6) = 5/6 = {p_not_six_rule:.4f}")
print(f"Simulated:        P(not 6) = {p_not_six_simulated:.4f}")
```

## 5. Visualizations

### The complement in a Venn diagram

The circle is the event (playing soccer, 30%); the complement is everything outside the circle but inside the sample space (70%).


```python
fig, ax = plt.subplots(figsize=(8, 5))

rectangle = plt.Rectangle((0, 0), 10, 6, facecolor="#f0c9c9",
                          edgecolor="green", linewidth=2)
ax.add_patch(rectangle)

circle = plt.Circle((3, 3), 1.8, facecolor="#7fc97f",
                    edgecolor="darkgreen", linewidth=2)
ax.add_patch(circle)

ax.text(3, 3, "A: soccer\nP(A) = 0.3", ha="center", va="center", fontsize=11)
ax.text(7.3, 3, "A': not soccer\nP(A') = 1 - 0.3 = 0.7",
        ha="center", va="center", fontsize=11)
ax.text(0.2, 5.6, "Sample space: P = 1 (100%)", fontsize=11)

ax.set_xlim(-0.5, 10.5)
ax.set_ylim(-0.5, 6.5)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("An event A and its complement A' fill the whole sample space")
plt.show()
```

### Event and complement always add to 1

For each experiment from the examples, the two bars stack to exactly 1.


```python
experiments = ["Soccer\n(10 kids)", "Three heads\n(3 coins)", "Rolling a 6\n(1 die)"]
p_event = np.array([0.3, 1 / 8, 1 / 6])
p_complement = 1 - p_event

plt.figure(figsize=(8, 5))
plt.bar(experiments, p_event, label="P(A)", color="#4e79a7")
plt.bar(experiments, p_complement, bottom=p_event, label="P(A')", color="#f28e2b")
plt.axhline(1.0, color="black", linewidth=0.8)
plt.title("P(A) + P(A') = 1 in every experiment")
plt.ylabel("Probability")
plt.grid(True, alpha=0.3, axis="y")
plt.legend()
plt.show()
```

## 6. Exercises

### Basic

1. An event has probability 0.75. What is the probability of its complement?
2. You roll one fair die. Use the complement rule to find the probability of **not** rolling a 1.

### Intermediate

3. You flip two coins. Use the complement rule to find the probability of getting **at least one tail**. (Hint: what is the complement of "at least one tail"?)
4. You roll two dice. What is the probability that the two dice do **not** both show a six?

### Advanced

5. You roll one die four times. What is the probability of rolling **at least one six**? Compute it with the complement rule and verify with a simulation. (Hint: the complement of "at least one six" is "no sixes at all".)


## 7. Solutions


```python
# Exercise 1
print("1.  P(A') = 1 - 0.75 =", 1 - 0.75)

# Exercise 2
print("2.  P(not 1) = 1 - 1/6 =", round(1 - 1 / 6, 4))

# Exercise 3: complement of 'at least one tail' is 'both heads'
print("3.  P(at least one tail) = 1 - P(HH) = 1 - 1/4 =", 1 - 1 / 4)

# Exercise 4: complement of 'not both six' is 'six-six'
print("4.  P(not both six) = 1 - 1/36 =", round(1 - 1 / 36, 4))

# Exercise 5: P(no six in one roll) = 5/6, four independent rolls -> (5/6)**4
p_at_least_one_six = 1 - (5 / 6) ** 4

rolls = rng.integers(1, 7, size=(200_000, 4))
simulated = np.any(rolls == 6, axis=1).mean()
print(f"5.  Complement rule: 1 - (5/6)^4 = {p_at_least_one_six:.4f}, "
      f"simulated = {simulated:.4f}")
```

## 8. Conclusion

- The **complement** $A'$ of an event $A$ is the event that $A$ does not occur.
- The **complement rule**: $P(A') = 1 - P(A)$, because $A$ and $A'$ together cover the entire sample space.
- The rule often replaces tedious counting — computing "at least one ..." probabilities via the complement ("none at all") is a classic shortcut you will use constantly in probability.

# A Complete Bayesian Worked Example: from Flat Prior to Firm Belief

Ten flips of a mystery coin — 8 heads, then 2 tails — turn a flat "anything goes" prior into a posterior $495\,\theta^8(1-\theta)^2$ peaking at $\theta = 0.8$. Ten more flips (6 heads, 4 tails) reshape that belief into $\theta^{14}(1-\theta)^6$ peaking at $0.7$. Along the way, every promise of the last three notebooks gets cashed in: why normalizing constants can be ignored, why MAP with a flat prior *is* MLE, why informative priors pull the answer away from the frequentist's, and why feeding data in chunks or all at once lands on the same beliefs. It is based on the eleventh video of Lesson 6 (Point Estimation), the finale of the Bayesian statistics arc.

## 2. Theory

### Framing the experiment

The coin has some unknown probability of heads. Treat it as a random variable $\Theta$ — the goal is to learn its probability density function, i.e. *how likely each heads-probability is*, given data.

The data are 10 flips, collected in a bold vector $\mathbf{X} = (X_1, \dots, X_{10})$, one random variable per flip. Each $X_i$ outputs 1 for heads and 0 for tails, so *if* you knew the coin's parameter, each flip would be a Bernoulli draw:

$$
X_i \mid \Theta = \theta \;\sim\; \text{Bernoulli}(\theta)
$$

$\Theta$ is continuous, $\mathbf{X}$ is discrete — the mixed variant of Bayes' theorem from the last notebook, with a PDF for $\Theta$ and a PMF for $\mathbf{X}$:

$$
f_{\Theta \mid \mathbf{X}}(\theta \mid \mathbf{x}) \;=\; \frac{p_{\mathbf{X} \mid \Theta}(\mathbf{x} \mid \theta)\; f_\Theta(\theta)}{p_{\mathbf{X}}(\mathbf{x})}
$$

### The likelihood of what you saw

The observed sequence is 8 heads followed by 2 tails. Given $\Theta = \theta$, each head has probability $\theta$, each tail $1 - \theta$, and the flips are independent — so the joint probability multiplies across:

$$
p_{\mathbf{X} \mid \Theta}(\mathbf{x} \mid \theta) \;=\; \underbrace{\theta \cdot \theta \cdots \theta}_{8 \text{ heads}} \cdot \underbrace{(1-\theta)(1-\theta)}_{2 \text{ tails}} \;=\; \theta^{8}(1-\theta)^{2}
$$

### Choosing the prior

Suppose you truly have no prior opinion: every heads-probability between 0 and 1 seems equally plausible. That is the **uniform prior**

$$
f_\Theta(\theta) = 1, \qquad 0 < \theta < 1
$$

(The constraint $0 < \theta < 1$ is part of the prior — and of the posterior — but since $\theta$ is always a probability, the notation below drops it.)

### Assembling the posterior

Multiply likelihood by prior, divide by the evidence:

$$
f_{\Theta \mid \mathbf{X}}(\theta \mid \mathbf{x}) \;=\; \frac{\theta^{8}(1-\theta)^{2} \cdot 1}{p_{\mathbf{X}}(\mathbf{x})}
$$

The denominator is just the constant that makes the total area equal 1. This particular shape is a **beta distribution** (not formally covered in this course — see the references), and its constant happens to be known exactly:

$$
p_{\mathbf{X}}(\mathbf{x}) = \frac{8!\,2!}{11!} = \frac{1}{495}
\qquad\Longrightarrow\qquad
f_{\Theta \mid \mathbf{X}}(\theta \mid \mathbf{x}) = 495\;\theta^{8}(1-\theta)^{2}
$$

These are your posterior beliefs: values of $\theta$ below $0.5$ are very unlikely, and the curve peaks somewhere around $0.8$.

### Why the constant can be ignored

Drop the constant entirely and write a proportionality:

$$
f_{\Theta \mid \mathbf{X}}(\theta \mid \mathbf{x}) \;\propto\; \underbrace{p_{\mathbf{X} \mid \Theta}(\mathbf{x} \mid \theta)}_{\text{how well model } \theta \text{ explains the data}} \times \underbrace{f_\Theta(\theta)}_{\text{how plausible model } \theta \text{ was}}
$$

To summarize the whole belief curve with one number, take the **MAP estimate** — the mode of the posterior, i.e. the $\theta$ where the curve peaks. In math terms, the MAP is an **argmax**: the *input* $\hat\theta$ that maximizes the posterior's *output*. And the constant only scales the curve up or down; it changes neither the shape nor *where* the maximum sits. So for MAP purposes the constant is dead weight:

$$
\hat\theta_{\text{MAP}} = \arg\max_\theta \; f_{\Theta \mid \mathbf{X}}(\theta \mid \mathbf{x}) = \arg\max_\theta \; \theta^{8}(1-\theta)^{2} = \frac{8}{10} = 0.8
$$

### Flat prior ⟹ MAP = MLE

$0.8$ is exactly the frequentist answer: 8 heads in 10 flips. No coincidence — an uninformative prior is just a constant, and constants (whether in the numerator or the denominator) never move an argmax. Strip them away and the only term left to maximize is the likelihood $p_{\mathbf{X}\mid\Theta}(\mathbf{x}\mid\theta)$ — which is *all MLE ever does*. **With a flat prior, MAP is MLE.**

### Round two: the posterior becomes the prior

Flip the coin 10 more times: 6 heads, 4 tails this time. To incorporate the new data, use your most recent belief — the posterior $\propto \theta^8(1-\theta)^2$ — as the **new prior**. This is the standard update cycle of Bayesian statistics. The new likelihood is $\theta^{6}(1-\theta)^{4}$, so

$$
f_{\Theta \mid \text{all data}}(\theta) \;\propto\; \underbrace{\theta^{6}(1-\theta)^{4}}_{\text{new likelihood}} \times \underbrace{\theta^{8}(1-\theta)^{2}}_{\text{new prior (old posterior)}} \;=\; \theta^{14}(1-\theta)^{6}
$$

another beta distribution, with MAP estimate

$$
\hat\theta_{\text{MAP}} = \frac{14}{20} = 0.7
$$

### Two lessons hiding in 0.7

1. **Informative priors matter.** A frequentist who only saw the second batch would answer $6/10 = 0.6$. Your prior — built from the first batch — pulled the estimate to $0.7$.
2. **Chunking doesn't.** Out of all 20 flips, 14 were heads, so a frequentist who saw *everything* would also say $14/20 = 0.7$. Feeding the data in two chunks of 10 or in one batch of 20 produces the same posterior — a general property of Bayesian updating.

### Choosing a philosophy

- Bayesian methods shine with **limited data** or **strong, trustworthy prior knowledge**; uninformative priors make MAP a complicated route to MLE, and (for reasonably behaved priors) MLE and MAP converge as data accumulates — new data slowly *waters down* the prior.
- The risk: **wrong priors cause wrong conclusions**, and with little data the prior dominates. If data is plentiful anyway, frequentist methods are usually fine.

Bayesian or frequentist? If you can't decide… flip a coin.

## 3. Worked Examples

### Example 1: the MAP of the first posterior, by calculus

Maximize $g(\theta) = \theta^8 (1-\theta)^2$. Maximizing $\ln g$ is equivalent (the logarithm is increasing) and easier:

$$
\ln g(\theta) = 8 \ln\theta + 2\ln(1-\theta), \qquad
\frac{d}{d\theta}\ln g(\theta) = \frac{8}{\theta} - \frac{2}{1-\theta} \stackrel{!}{=} 0
$$

$$
8(1-\theta) = 2\theta \;\;\Longrightarrow\;\; 8 = 10\,\theta \;\;\Longrightarrow\;\; \hat\theta = \frac{8}{10} = 0.8
$$

### Example 2: the second update, in one line of algebra

New likelihood times new prior:

$$
\theta^{6}(1-\theta)^{4} \times \theta^{8}(1-\theta)^{2} = \theta^{14}(1-\theta)^{6}
$$

The exponents simply **add** — 8+6 = 14 total heads, 2+4 = 6 total tails. The same calculus as Example 1 (or the general pattern $\hat\theta = a/(a+b)$ for $\theta^a(1-\theta)^b$, proven in Exercise 5) gives $\hat\theta = 14/20 = 0.7$. The exponent bookkeeping *is* the reason chunking doesn't matter: any way of splitting the 20 flips adds up to the same exponents.

### Example 3: the normalizing constant (and why we checked it once)

For the first posterior, the evidence works out to $\displaystyle\int_0^1 \theta^8(1-\theta)^2\,d\theta = \frac{8!\,2!}{11!} = \frac{80640}{39916800} = \frac{1}{495}$, so the honest, area-1 posterior is $495\,\theta^8(1-\theta)^2$ — a beta distribution. We verify the 495 numerically below, then never compute a constant again: proportionality is enough for everything MAP-related.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

theta = np.linspace(0.001, 0.999, 999)

def normalize(curve):
    """Scale a nonnegative curve so it integrates to 1 over theta."""
    return curve / np.trapezoid(curve, theta)
```

### The video's two posteriors, symbolically

SymPy re-derives both argmaxes and the exact normalizing constant.

```python
t = sp.symbols("theta", positive=True)

posterior_1 = t**8 * (1 - t)**2            # after 10 flips (8 H, 2 T)
posterior_2 = t**14 * (1 - t)**6           # after 20 flips (14 H, 6 T)

for name, post in [("first", posterior_1), ("second", posterior_2)]:
    critical = sp.solve(sp.diff(sp.log(post), t), t)
    print(f"{name} posterior ~ {post}:  MAP = {critical[0]} = {float(critical[0])}")

constant = 1 / sp.integrate(posterior_1, (t, 0, 1))
print(f"\nnormalizing constant of the first posterior: {constant}"
      f"  (= 11!/(8! 2!) = {sp.factorial(11) // (sp.factorial(8) * sp.factorial(2))})")
```

### The update cycle, numerically

One function does everything: multiply the current belief by the likelihood of the new flips, renormalize. Applying it twice replays the video.

```python
def bayes_update(prior_curve, heads, tails):
    """Posterior curve after observing `heads` heads and `tails` tails."""
    likelihood = theta**heads * (1 - theta)**tails
    return normalize(likelihood * prior_curve)

uniform_prior = normalize(np.ones_like(theta))
posterior_10 = bayes_update(uniform_prior, heads=8, tails=2)   # first batch
posterior_20 = bayes_update(posterior_10, heads=6, tails=4)    # second batch

print("uniform prior   MAP: undefined -- the curve is flat, every theta ties")
for label, curve in [("after 10 flips ", posterior_10),
                     ("after 20 flips ", posterior_20)]:
    print(f"{label} MAP = {theta[np.argmax(curve)]:.2f}")

# the numeric curve matches the exact 495 * theta^8 * (1-theta)^2
print(f"\npeak height of posterior_10: {posterior_10.max():.3f}"
      f"   (exact 495 * 0.8^8 * 0.2^2: {495 * 0.8**8 * 0.2**2:.3f})")
```

### Chunked vs all-at-once

The property the video highlights: incorporate all 20 flips in one update, in two chunks of 10, or one flip at a time — the final posterior is identical.

```python
all_at_once = bayes_update(uniform_prior, heads=14, tails=6)

one_at_a_time = uniform_prior.copy()
flips = [1]*8 + [0]*2 + [1]*6 + [0]*4          # the video's 20 flips in order
for flip in flips:
    one_at_a_time = bayes_update(one_at_a_time, heads=flip, tails=1 - flip)

print(f"two chunks of 10 : MAP = {theta[np.argmax(posterior_20)]:.2f}")
print(f"all 20 at once   : MAP = {theta[np.argmax(all_at_once)]:.2f}")
print(f"one flip at a time: MAP = {theta[np.argmax(one_at_a_time)]:.2f}")
print(f"max |difference| between the three curves: "
      f"{max(np.abs(posterior_20 - all_at_once).max(), np.abs(posterior_20 - one_at_a_time).max()):.2e}")
```

## 5. Visualizations

### The story in three curves

From a flat line, to a peak at 0.8, to a sharper peak at 0.7.

```python
fig, axes = plt.subplots(1, 3, figsize=(13, 4), sharey=True)

stages = [(uniform_prior, "uniform prior\n(no data)", "tab:gray"),
          (posterior_10, "posterior after 10 flips\n(8 H, 2 T)", "tab:blue"),
          (posterior_20, "posterior after 20 flips\n(+ 6 H, 4 T)", "tab:green")]

for ax, (curve, title, color) in zip(axes, stages):
    ax.plot(theta, curve, color=color, linewidth=2)
    ax.fill_between(theta, curve, color=color, alpha=0.15)
    if curve is uniform_prior:
        ax.plot([], [], " ", label="no peak: all $\\theta$ tie")
    else:
        map_theta = theta[np.argmax(curve)]
        ax.axvline(map_theta, color=color, linestyle=":",
                   label=f"MAP = {map_theta:.2f}")
    ax.set_title(title, fontsize=10)
    ax.set_xlabel(r"$\theta$ (heads probability)")
    ax.grid(True, alpha=0.3)
    ax.legend()
axes[0].set_ylabel("density")
plt.suptitle("Iterative Bayesian updating: each posterior becomes the next prior")
plt.tight_layout()
plt.show()
```

### Where the prior shows its influence

On the second batch alone (6 H, 4 T) a frequentist says 0.6 — but the Bayesian's prior, carrying the memory of the first batch, pulls the answer to 0.7. Seen over *all* the data, 0.7 is exactly what a frequentist would say too.

```python
likelihood_batch2 = normalize(theta**6 * (1 - theta)**4)

plt.figure(figsize=(9.5, 4.6))
plt.plot(theta, likelihood_batch2, color="tab:red", linewidth=2,
         label="batch 2 alone (frequentist view): peak 0.6")
plt.plot(theta, posterior_10, color="tab:blue", linewidth=2, linestyle="--",
         label="prior carried from batch 1: peak 0.8")
plt.plot(theta, posterior_20, color="tab:green", linewidth=2,
         label="Bayesian posterior (product): peak 0.7")
for x, color in [(0.6, "tab:red"), (0.8, "tab:blue"), (0.7, "tab:green")]:
    plt.axvline(x, color=color, linestyle=":", linewidth=1)
plt.xlabel(r"$\theta$ (heads probability)")
plt.ylabel("density (scaled)")
plt.title("An informative prior pulls 0.6 up to 0.7 — matching the all-data frequentist answer")
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.show()
```

### Data waters down the prior

Start Bayesians with three different priors — flat, mildly wrong, badly wrong — and feed them a long stream of flips from a coin with true $\theta = 0.8$. All three MAP trajectories converge to the truth (and to the MLE): with enough data, the prior stops mattering.

```python
rng = np.random.default_rng(42)
true_theta = 0.8
stream = (rng.random(400) < true_theta).astype(int)

priors = {"flat prior": normalize(np.ones_like(theta)),
          "mildly wrong (fair-ish)": normalize(theta**9 * (1 - theta)**9),
          "badly wrong (tails-biased)": normalize(theta**2 * (1 - theta)**18)}

plt.figure(figsize=(9.5, 4.6))
for (label, curve), color in zip(priors.items(),
                                 ["tab:gray", "tab:blue", "tab:red"]):
    trajectory = []                        # MAP after 1, 2, ... flips
    for flip in stream:
        curve = bayes_update(curve, heads=flip, tails=1 - flip)
        trajectory.append(theta[np.argmax(curve)])
    plt.plot(np.arange(1, len(stream) + 1), trajectory,
             color=color, linewidth=2, label=label)

mle = np.cumsum(stream) / np.arange(1, len(stream) + 1)
plt.plot(np.arange(1, len(stream) + 1), mle, color="black", linewidth=1,
         linestyle="--", alpha=0.6, label="MLE (running proportion)")
plt.axhline(true_theta, color="black", linestyle=":", linewidth=1)
plt.xlabel("number of flips observed")
plt.ylabel(r"MAP estimate of $\theta$")
plt.title(r"MLE and MAP converge: new data slowly waters down the prior (true $\theta = 0.8$)")
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.show()
```

Everything above in a single sheet: the flat prior turning into posteriors peaked at 0.8 and 0.7, the prior pulling batch 2's 0.6 up to 0.7, and three priors washed out by 400 flips.

```python
# Exercise 1: H, H, H, T with a uniform prior
# posterior ~ theta^3 * (1 - theta)^1, so MAP = 3/(3+1)
posterior_ex1 = normalize(theta**3 * (1 - theta))
print(f"Exercise 1: posterior ~ theta^3 (1-theta),  MAP = {theta[np.argmax(posterior_ex1)]:.2f}"
      f"  (exact: 3/4)")
```

```python
# Exercise 2: naming the factors
print("Exercise 2:")
print("  theta^8 (1-theta)^2  = likelihood  p_X|Theta(x | theta)")
print("  1                    = prior       f_Theta(theta)  (uniform)")
print("  p_X(x)               = evidence, the normalizing constant (= 1/495)")
print("  left side            = posterior   f_Theta|X(theta | x)")
print("Variant: continuous parameter Theta (PDF), discrete data X (PMF) --")
print("the mixed continuous/discrete form.")
```

```python
# Exercise 3: chunking invariance with uneven chunks
chunks = [(4, 1), (3, 2), (4, 0), (2, 2), (1, 1)]   # (heads, tails), total 14 H 6 T
assert sum(h for h, _ in chunks) == 14 and sum(t for _, t in chunks) == 6

curve = normalize(np.ones_like(theta))
for heads_k, tails_k in chunks:
    curve = bayes_update(curve, heads=heads_k, tails=tails_k)

print(f"Exercise 3: uneven chunks MAP = {theta[np.argmax(curve)]:.2f},  "
      f"max |diff| vs all-at-once = {np.abs(curve - all_at_once).max():.2e}")
print("Why: each chunk just adds its heads/tails to the exponents, and sums")
print("don't care how you group the terms -- every split ends at theta^14 (1-theta)^6.")
```

```python
# Exercise 4: the first batch under an informative fair-ish prior
fairish_prior = normalize(theta**9 * (1 - theta)**9)
posterior_ex4 = bayes_update(fairish_prior, heads=8, tails=2)

print(f"Exercise 4: posterior ~ theta^17 (1-theta)^11,  "
      f"MAP = {theta[np.argmax(posterior_ex4)]:.3f}  (exact: 17/28 = {17/28:.3f})")
print("The prior acts like 9 phantom heads and 9 phantom tails centered on 0.5,")
print("so it pulls the estimate down from the data-only 0.8 toward fairness.")
```

```python
# Exercise 5: the mode of theta^a (1-theta)^b is a/(a+b)
a, b = sp.symbols("a b", positive=True)
curve_sym = t**a * (1 - t)**b
mode = sp.solve(sp.diff(sp.log(curve_sym), t), t)[0]
print(f"Exercise 5: argmax of theta^a (1-theta)^b is theta = {mode}")
print("Hand proof: d/dtheta [a ln(theta) + b ln(1-theta)] = a/theta - b/(1-theta) = 0")
print("  =>  a (1-theta) = b theta  =>  theta = a/(a+b).")
print("Every posterior here has this form, so MAP = heads exponent / total exponent.")
```

```python
# Exercise 6: a stubborn wrong prior vs a theta = 0.8 coin
# MAP after n flips with h heads: (2 + h) / (40 + n).  With h ~ 0.8 n on average:
# (2 + 0.8 n)/(40 + n) >= 0.75  =>  0.05 n >= 28  =>  n >= 560 flips.
n_expected = (0.75 * 40 - 2) / (true_theta - 0.75)
print(f"Exercise 6: expected flips needed ~ {n_expected:.0f}")

rng_ex6 = np.random.default_rng(7)
first_hits = []
for _ in range(200):
    flips_sim = (rng_ex6.random(1200) < true_theta).astype(int)
    n = np.arange(1, len(flips_sim) + 1)
    map_traj = (2 + np.cumsum(flips_sim)) / (40 + n)     # exponent bookkeeping
    hits = np.nonzero(np.abs(map_traj - true_theta) <= 0.05)[0]
    first_hits.append(hits[0] + 1 if len(hits) else np.nan)

print(f"simulated median first time within 0.05 of the truth: "
      f"{np.nanmedian(first_hits):.0f} flips")
print("A prior worth ~40 phantom flips of misinformation costs hundreds of real")
print("flips to wash out -- with limited data, wrong priors mean wrong conclusions.")
```

## 8. Conclusion

- The full Bayesian workflow on one coin: model the unknown heads-probability as $\Theta$, the flips as Bernoulli variables $\mathbf{X}$, and apply the mixed (PDF/PMF) Bayes' theorem. Likelihood of 8 H + 2 T: $\theta^8(1-\theta)^2$; uniform prior; posterior $495\,\theta^8(1-\theta)^2$ — a beta distribution.
- The normalizing constant only scales the curve; it never moves the peak. So MAP works with proportionality alone: $\hat\theta = \arg\max \theta^8(1-\theta)^2 = 0.8$.
- A **flat prior makes MAP = MLE**: uninformative priors are constants, and constants don't move argmaxes — only the likelihood is left.
- **Updating is iterative**: the posterior becomes the new prior. After 6 H + 4 T more, the posterior is $\propto \theta^{14}(1-\theta)^6$ with MAP $= 0.7$ — the prior pulled the batch-2-only answer (0.6) toward the memory of batch 1, and matched the all-data frequentist answer exactly.
- **Chunking doesn't matter**: exponents add, so any way of splitting the data reaches the same posterior.
- Choose Bayesian methods when data is scarce or priors are genuinely informative; expect MLE and MAP to converge as data grows (data waters down the prior) — and beware that wrong priors plus little data means wrong conclusions.

This concludes the Bayesian statistics arc of the lesson. Bayesian or frequentist? If you're having trouble deciding — flip a coin.

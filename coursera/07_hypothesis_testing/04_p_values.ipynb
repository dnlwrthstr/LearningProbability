# The p-Value: How Unlikely Is "Too Far"?

The last notebook left a number dangling: about 3% of null-world sample means reach the observed 68.442. This notebook gives that number its name. Under $H_0$ (with $\sigma = 3$ known and $n = 10$), $\bar{X} \sim N(66.7,\, 3^2/10)$ — so the probability of a sample mean **as extreme as or more extreme than** the one observed, *assuming $H_0$ is true*, can be computed exactly: **the p-value**. Right-tailed: $p = 0.0332 < \alpha = 0.05$ → reject. Two-tailed: $p = 0.0663$ — same data, verdict flips. Left-tailed (with a different sample, $\bar{x} = 64.252$): $p = 0.0049$, rejecting even at $\alpha = 0.01$. And the whole procedure survives standardization intact: the **z-statistic** $z_{obs} = 1.837$ runs the same test on the standard normal. It is based on the fourth video of Lesson 8 (Hypothesis Testing).

## 1. Introduction

"Reject $H_0$ when the sample mean is too far from 66.7" was the intuition; the p-value is the ruler that measures "too far" in units of probability. This notebook computes it for all three tails, meets the decision rule $p \lessgtr \alpha$, and standardizes the test into its z-statistic form.

After completing this notebook you will be able to:

- State the distribution of $\bar{X}$ under $H_0$ and use it to price any observed sample mean.
- Define the p-value — the probability, assuming $H_0$, of a statistic as extreme or more extreme than observed, in $H_1$'s direction — and apply the rule $p < \alpha \Rightarrow$ reject.
- Compute p-values for right-, left-, and two-tailed tests, reproducing the video's 0.0332, 0.0663, and 0.0049.
- Explain how one dataset can reject under a right-tailed test yet fail to reject under a two-tailed one.
- Recast the test with the standardized z-statistic and show nothing changes.

## 2. Theory

### The statistic gets a distribution

Suppose the height standard deviation is $\sigma = 3$ (known — the unknown-$\sigma$ case comes later) and nothing changed over the years. Then with $n = 10$,

$$
\bar{X} \;\sim\; N\!\left(66.7,\; \frac{3^2}{10}\right) \qquad \text{if } H_0 \text{ is true}
$$

Now "how likely was my sample if $H_0$ is true?" is a computable question — and *very unlikely* is grounds to reject.

### The right-tailed argument

Observed: $\bar{x} = 68.442$, target Type I probability $\alpha = 0.05$. If 68.442 is convincing enough to reject, then any *larger* sample mean would be too. So the rejection rule "reject whenever $\bar{X} \ge 68.442$" has Type I probability

$$
P(\bar{X} \ge 68.442 \mid \mu = 66.7) = 0.0332
$$

— the shaded right tail. Since $0.0332 < 0.05$, this rejection criterion keeps the Type I risk below your tolerance: **reject $H_0$**. That probability has a name:

### The p-value

> **The p-value is the probability, assuming $H_0$ is true, that the test statistic takes a value as extreme as or more extreme than the value observed** — where "more extreme" means *further in $H_1$'s direction*.

It measures how well your sample is accommodated under the null: a small p-value says the sample landed in $H_0$'s tail — pretty unlikely if $H_0$ were true. The decision rule pairs it with the significance level:

$$
p < \alpha \;\Rightarrow\; \text{reject } H_0 \text{ (accept } H_1\text{)} \qquad\qquad
p \ge \alpha \;\Rightarrow\; \text{evidence too weak; don't reject}
$$

### The three formulas

Writing $t_{obs}$ for the observed statistic and $\mu_0$ for the null value:

| test | p-value |
|---|---|
| right-tailed | $P(\bar{X} > t_{obs} \mid \mu = \mu_0)$ |
| left-tailed | $P(\bar{X} < t_{obs} \mid \mu = \mu_0)$ |
| two-tailed | $P\big(\,\lvert\bar{X} - \mu_0\rvert > \lvert t_{obs} - \mu_0 \rvert \,\big\vert\, \mu = \mu_0\big)$ |

### Two-tailed: same data, doubled tail, flipped verdict

For $H_1: \mu \ne 66.7$, "as extreme" counts both directions: $p = P(|\bar{X} - 66.7| > 1.742) = 0.0663$ — exactly **twice** the right-tailed value, by symmetry. And $0.0663 > 0.05$: **don't reject**. The same 68.442 that confirmed "heights increased" fails to confirm "heights changed" — the two-tailed test spreads its $\alpha$ over two tails, demanding more of each.

### Left-tailed: a different sample

68.442 can't support "heights *decreased*," so imagine instead a sample with $\bar{x} = 64.252$. Then $p = P(\bar{X} < 64.252 \mid \mu = 66.7) = 0.0049$ — below 0.05, **and** below the stricter $\alpha = 0.01$: reject, and know you'd have rejected even at the conservative level.

### The z-statistic: same test, standard units

Instead of working with $\bar{X}$, standardize it:

$$
Z = \frac{\bar{X} - \mu_0}{\sigma/\sqrt{n}} \;\sim\; N(0, 1) \text{ under } H_0, \qquad
z_{obs} = \frac{68.442 - 66.7}{3/\sqrt{10}} = 1.837
$$

Transform the rejection event step by step — subtract $\mu_0$ from both sides, scale by $\sigma/\sqrt{n}$ — and $\{\bar{X} > 68.442\}$ becomes $\{Z > 1.837\}$, now priced on the *standard* normal (66.7 becomes 0; 68.442 becomes 1.837). The p-value must come out identical — same test, same data, merely standardized — but the z-form needs only one reference distribution, which is why tables and software speak z.

## 3. Worked Examples

### Example 1: the right-tailed p-value by hand

Standard error: $3/\sqrt{10} = 0.949$. Distance: $68.442 - 66.7 = 1.742$. In standard units: $z = 1.742/0.949 = 1.837$. Tail: $P(Z > 1.837) = 0.0332$. Verdict at $\alpha = 0.05$: reject.

### Example 2: the two-tailed flip

$p_{two} = P(\bar{X} > 68.442) + P(\bar{X} < 64.958) = 0.0332 + 0.0332 = 0.0663 > 0.05$: don't reject. Both computations used identical data — the *question* changed, so the price of "extreme" changed. (Moral from last notebook, now quantified: the sharper question, honestly chosen in advance, is easier to answer.)

### Example 3: reading the left-tailed result

$\bar{x} = 64.252 \Rightarrow z = (64.252 - 66.7)/0.949 = -2.580 \Rightarrow p = P(Z < -2.580) = 0.0049$. Because $0.0049 < 0.01 < 0.05$, the rejection is robust across both conventional significance levels — a fact worth reporting, since it tells the reader the conclusion doesn't hinge on the lenient $\alpha$.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project), then a p-value calculator for all three tails.

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(44)

MU0, SIGMA, N = 66.7, 3.0, 10
SE = SIGMA / np.sqrt(N)

Phi = lambda z: 0.5 * (1 + float(sp.erf(sp.Float(z) / sp.sqrt(2))))

def p_value(x_bar_obs, tail):
    """p-value for the mean test with known sigma, under H0: mu = MU0."""
    z_obs = (x_bar_obs - MU0) / SE
    if tail == "right":
        return 1 - Phi(z_obs)
    if tail == "left":
        return Phi(z_obs)
    if tail == "two":
        return 2 * (1 - Phi(abs(z_obs)))
```

### The video's three tests

```python
alpha = 0.05
cases = [("right-tailed, x-bar = 68.442", 68.442, "right"),
         ("two-tailed,   x-bar = 68.442", 68.442, "two"),
         ("left-tailed,  x-bar = 64.252", 64.252, "left")]

for name, xb, tail in cases:
    p = p_value(xb, tail)
    verdict = "REJECT H0" if p < alpha else "don't reject H0"
    print(f"{name}: p = {p:.4f}  ->  {verdict} (alpha = {alpha})")

print(f"\nleft-tailed p also below alpha = 0.01? {p_value(64.252, 'left') < 0.01}")
print("Same data, right vs two-tailed: the verdict flips at the factor of 2.")
```

### The p-value is a tail area — checked by brute force

Simulate a million null-world sample means and just count. (Notebook 03's mysterious 0.0333 was this number all along.)

```python
null_xbars = rng.normal(MU0, SE, size=1_000_000)
print(f"fraction >= 68.442:              {(null_xbars >= 68.442).mean():.4f}   "
      f"(exact p: {p_value(68.442, 'right'):.4f})")
print(f"fraction with |x - 66.7| > 1.742: {(np.abs(null_xbars - MU0) > 1.742).mean():.4f}   "
      f"(exact p: {p_value(68.442, 'two'):.4f})")
print(f"fraction <= 64.252:              {(null_xbars <= 64.252).mean():.4f}   "
      f"(exact p: {p_value(64.252, 'left'):.4f})")
```

### The z-statistic: the same test in standard units

```python
z_obs = (68.442 - MU0) / SE
print(f"observed z-statistic: (68.442 - 66.7) / (3/sqrt(10)) = {z_obs:.3f}")
print(f"p from the x-bar scale: P(X-bar > 68.442 | H0) = {p_value(68.442, 'right'):.4f}")
print(f"p from the z scale:     P(Z > {z_obs:.3f})          = {1 - Phi(z_obs):.4f}")
print("\nIdentical, necessarily: same test, same data -- only the units changed.")
print("The z form is what tables and software use: one standard reference curve")
print("for every mean test, whatever its mu0, sigma, and n.")
```

## 5. Visualizations

### The video's three shaded areas

```python
x_axis = np.linspace(MU0 - 4 * SE, MU0 + 4 * SE, 700)
density = np.exp(-0.5 * ((x_axis - MU0) / SE) ** 2) / (SE * np.sqrt(2 * np.pi))

fig, axes = plt.subplots(1, 3, figsize=(13.5, 4.2), sharey=True)
panels = [
    ("right-tailed\n$\\bar{x}$ = 68.442", x_axis >= 68.442, 68.442,
     p_value(68.442, "right"), "REJECT"),
    ("two-tailed\n$\\bar{x}$ = 68.442", np.abs(x_axis - MU0) >= 1.742, 68.442,
     p_value(68.442, "two"), "don't reject"),
    ("left-tailed\n$\\bar{x}$ = 64.252", x_axis <= 64.252, 64.252,
     p_value(64.252, "left"), "REJECT"),
]
for ax, (title, region, xb, p, verdict) in zip(axes, panels):
    ax.plot(x_axis, density, color="black", linewidth=1.6)
    ax.fill_between(x_axis, density, where=region, color="tab:red", alpha=0.55)
    ax.axvline(xb, color="tab:red", linewidth=1.8)
    ax.axvline(MU0, color="black", linestyle=":", linewidth=1)
    ax.set_title(f"{title}\np = {p:.4f} -> {verdict}", fontsize=10)
    ax.set_xlabel(r"$\bar{x}$ (inches)")
axes[0].set_ylabel("density under $H_0$")
plt.suptitle("The p-value is the shaded area: as-extreme-or-more, in $H_1$'s direction ($\\alpha$ = 0.05)")
plt.tight_layout()
plt.show()
```

### Every possible observation, priced

The right-tailed p-value as a function of the observed sample mean. Where the curve crosses $\alpha$ is the rejection threshold — the p-value view and the rejection-region view are the same test read off two axes.

```python
xb_axis = np.linspace(65.5, 70, 300)
p_curve = [p_value(xb, "right") for xb in xb_axis]
threshold = MU0 + 1.645 * SE

plt.figure(figsize=(9.5, 4.6))
plt.plot(xb_axis, p_curve, color="tab:blue", linewidth=2, label="right-tailed p-value")
plt.axhline(0.05, color="black", linestyle=":", label=r"$\alpha$ = 0.05")
plt.axvline(threshold, color="gray", linestyle="--", linewidth=1.2,
            label=f"rejection threshold {threshold:.2f}")
plt.plot(68.442, p_value(68.442, "right"), "ro", markersize=6,
         label=f"observed: p = {p_value(68.442, 'right'):.4f}")
plt.yscale("log")
plt.xlabel(r"observed sample mean $\bar{x}$ (inches)")
plt.ylabel("p-value (log scale)")
plt.title("p < alpha exactly where the observation passes the threshold")
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.show()
```

### Two rulers, one measurement

The identical test on the $\bar{x}$ scale and the z scale — the shaded areas are the same number, 0.0332.

```python
fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.2))

axes[0].plot(x_axis, density, color="black", linewidth=1.6)
axes[0].fill_between(x_axis, density, where=x_axis >= 68.442,
                     color="tab:red", alpha=0.55)
axes[0].set_title(r"$\bar{X} \sim N(66.7,\, 3^2/10)$: area beyond 68.442", fontsize=10)
axes[0].set_xlabel(r"$\bar{x}$ (inches)")
axes[0].set_ylabel("density")
for x, lab in [(MU0, "66.7"), (68.442, "68.442")]:
    axes[0].axvline(x, color="gray", linestyle=":", linewidth=1)

z_axis = np.linspace(-4, 4, 700)
z_density = np.exp(-0.5 * z_axis**2) / np.sqrt(2 * np.pi)
axes[1].plot(z_axis, z_density, color="black", linewidth=1.6)
axes[1].fill_between(z_axis, z_density, where=z_axis >= z_obs,
                     color="tab:red", alpha=0.55)
axes[1].set_title(r"$Z \sim N(0, 1)$: area beyond 1.837", fontsize=10)
axes[1].set_xlabel("z")
for x in (0, z_obs):
    axes[1].axvline(x, color="gray", linestyle=":", linewidth=1)

for ax in axes:
    ax.grid(True, alpha=0.3)
plt.suptitle("Standardizing relabels the axis; the p-value (shaded) is untouched")
plt.tight_layout()
plt.show()
```

## 7. Solutions

```python
# Exercise 1: x-bar = 67.5
p1 = p_value(67.5, "right")
z1 = (67.5 - MU0) / SE
print(f"p = P(X-bar > 67.5 | H0) = {p1:.4f} -> "
      f"{'reject' if p1 < 0.05 else 'do NOT reject'} at alpha = 0.05")
print(f"z-statistic: {z1:.3f}; P(Z > {z1:.3f}) = {1 - Phi(z1):.4f} -- identical.")
```

```python
# Exercise 2: true/false
print("(a) FALSE. H0 is fixed -- it has no probability (lesson 7's confidence-vs-")
print("    probability lesson again). The p-value conditions ON H0 being true.")
print("(b) TRUE. That is precisely the definition: P(as extreme or more | H0).")
print("(c) FALSE. Large p = weak evidence against H0, never proof of it --")
print("    failing to reject is not accepting.")
```

```python
# Exercise 3: the borderline observation
z95 = float(sp.sqrt(2) * sp.erfinv(sp.Float(0.90)))
x_borderline = MU0 + z95 * SE
print(f"borderline x-bar = 66.7 + {z95:.3f} * {SE:.4f} = {x_borderline:.3f}")
print(f"check: p at that observation = {p_value(x_borderline, 'right'):.4f} (= alpha)")
print("The p-value rule and the rejection-region rule agree everywhere: p < alpha")
print("exactly when the observation crosses the region's edge.")
```

```python
# Exercise 4: right vs two-tailed price list
print(f"{'x-bar':>8s} {'p right':>9s} {'p two':>9s}   verdicts at 0.05")
for xb in (67.5, 68.0, 68.442, 69.0):
    pr, pt = p_value(xb, "right"), p_value(xb, "two")
    vr = "reject" if pr < 0.05 else "keep"
    vt = "reject" if pt < 0.05 else "keep"
    flag = "  <-- flip" if vr != vt else ""
    print(f"{xb:>8.3f} {pr:>9.4f} {pt:>9.4f}   {vr:>6s} / {vt}{flag}")
print("\nRelation: p_two = 2 * p_right (for observations above mu0). An x-bar flips")
print("between the tests exactly when p_right lands in (0.025, 0.05) -- here only")
print("68.442 does; 68.0 fails both tests and 69.0 passes both.")
```

```python
# Exercise 5: the p-value's own distribution
xb_null = rng.normal(MU0, SE, size=100_000)
p_null = 1 - np.vectorize(Phi)((xb_null - MU0) / SE)
xb_alt = rng.normal(68.0, SE, size=100_000)
p_alt = 1 - np.vectorize(Phi)((xb_alt - MU0) / SE)

fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=False)
axes[0].hist(p_null, bins=40, density=True, color="tab:green", alpha=0.7)
axes[0].axhline(1, color="black", linestyle=":")
axes[0].set_title("H0 true: p-values are UNIFORM on [0, 1]", fontsize=10)
axes[1].hist(p_alt, bins=40, density=True, color="tab:red", alpha=0.7)
axes[1].set_title("mu really 68: p-values pile up near 0", fontsize=10)
for ax in axes:
    ax.set_xlabel("right-tailed p-value")
    ax.grid(True, alpha=0.3)
axes[0].set_ylabel("density")
plt.tight_layout()
plt.show()

print(f"P(p < 0.05 | H0 true) = {(p_null < 0.05).mean():.3f}  -- the uniform makes")
print("'reject when p < alpha' commit Type I errors at rate exactly alpha: the")
print("rule is calibrated by construction.")
print(f"P(p < 0.05 | mu = 68)  = {(p_alt < 0.05).mean():.3f}  -- the pile-up near 0")
print("is the test's power to detect a real change.")
```

```python
# Exercise 6: significant but unimportant
n6 = 1_000_000
se6 = SIGMA / np.sqrt(n6)
z6 = (66.72 - MU0) / se6
p6 = 1 - Phi(z6)
print(f"n = {n6:,}, expected x-bar = 66.72: z = {z6:.2f}, p = {p6:.2e}")
print("\nThe test is CORRECT: mu truly is greater than 66.7, and with a million")
print("measurements the test can see even a 0.02-inch difference. But 'the mean")
print("increased' and 'the increase matters' are different claims -- the p-value")
print("answers only the first. With huge n, report the effect size (here: half a")
print("millimeter) alongside the p-value, or significance becomes a trivial fact")
print("about your sample size rather than an interesting fact about the world.")
```

## 8. Conclusion

- Under $H_0$ (known $\sigma$), $\bar{X} \sim N(\mu_0, \sigma^2/n)$ — so "how unlikely is my sample if nothing changed?" has an exact answer.
- **The p-value**: the probability, assuming $H_0$ is true, of a test statistic as extreme as or more extreme than observed, in $H_1$'s direction. Decision rule: $p < \alpha$ → reject; otherwise the evidence is too weak.
- The three tails price "extreme" differently: right $P(\bar{X} > t_{obs})$, left $P(\bar{X} < t_{obs})$, two-tailed $P(|\bar{X} - \mu_0| > |t_{obs} - \mu_0|)$ — the video's numbers: **0.0332** (reject), **0.0663** (don't reject — same data, doubled tail), **0.0049** (reject even at $\alpha = 0.01$).
- The **z-statistic** $(\bar{x} - \mu_0)/(\sigma/\sqrt{n}) = 1.837$ recasts the same test onto the standard normal — the p-value is unchanged because only the units moved.
- From the exercises: under a true null, p-values are *uniform* (which is exactly what calibrates the $\alpha$ rule), and with enormous samples a microscopic effect earns a microscopic p-value — significance is not importance.

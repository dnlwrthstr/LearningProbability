# Testing a Population Mean: Hypotheses, Test Statistics, and the Three Tails

In the 1970s, 18-year-olds in the US averaged 66.7 inches. Your sample of ten averages 68.442 — bigger, sure, but *is that enough to confirm heights have increased?* Setting up that question properly takes three ideas: hypotheses are **always about population parameters** ($H_0: \mu = 66.7$, never about $\bar{x}$); the decision runs through a **test statistic** (the random variable $\bar{X}$) whose realized value 68.442 is the **observed statistic**; and the alternative hypothesis comes in three flavors — **right-tailed** ($\mu > 66.7$), **left-tailed** ($\mu < 66.7$), and **two-tailed** ($\mu \ne 66.7$) — each with its own rejection style and its own reading of the two error types. ## 2. Theory

### First, the data has to deserve a decision

Decisions based on data are only as good as the data. Reliable means:

1. **Representative & randomized.** The sample must reflect the population, with randomization guarding against bias — measure only school basketball teams and your "average 18-year-old" is systematically tall before any statistics happen.
2. **Enough of it.** A good rule of thumb: **30 samples or more**. (Our running example has $n = 10$ — the video's examples *assume* the data is good enough; the rule is what you check in practice.)

### The question, and the rule for writing it

Historical data: the mean height of 18-year-olds in the US in the 1970s was **66.7 inches**. Your sample of ten (the same dataset used for maximum likelihood back in the estimation lesson) averages **68.442 inches**. Can you confirm the mean height has *increased*?

The hypotheses:

$$
H_0: \mu = 66.7 \qquad\qquad H_1: \mu > 66.7
$$

One iron rule: **hypotheses are formulated in terms of the population** — the parameter $\mu$ — and must not involve samples at all. "$H_0: \bar{x} = 66.7$" is not a hypothesis; $\bar{x}$ is evidence, not a claim about the world.

### Test statistic vs observed statistic

The hypotheses are about the population; the *decision* is based on observations. The quantity that carries the decision is the **test statistic** — here the sample mean $\bar{X}$, a **random variable**, defined before any data is seen. The number you actually computed, $68.442$, is the **observed statistic**: the test statistic evaluated on your measurements.

In general a test statistic is *any function of the random samples that is informative about the parameter under study*:

- population mean → $\bar{X}$; same idea for probabilities and occurrence rates (sample proportion);
- population variance → $S^2$ — though statistics are **not unique**: the plain sum of squared differences $\sum_i (X_i - \bar{X})^2$ would also do.

### The three kinds of alternative

One baseline, three possible research questions — the null never changes; the alternative encodes what you want to prove:

| question | $H_0$ | $H_1$ | name |
|---|---|---|---|
| did heights *increase*? | $\mu = 66.7$ | $\mu > 66.7$ | **right-tailed** |
| did heights *decrease*? | $\mu = 66.7$ | $\mu < 66.7$ | **left-tailed** |
| did heights *change at all*? | $\mu = 66.7$ | $\mu \ne 66.7$ | **two-tailed** |

"Right-tailed" because the alternative's values sit to the right of the null value on the number line; likewise left. (You may also see the right-tailed null written $H_0: \mu \le 66.7$ and the left-tailed one $\mu \ge 66.7$ — equivalent tests; we keep the simple $\mu = 66.7$ form.)

### What triggers rejection — and what the errors mean

- **Right-tailed:** reject when $\bar{x}$ is *much greater* than 66.7. Type I: declaring $\mu > 66.7$ when it's still 66.7. Type II: declaring "no change" when the mean truly grew.
- **Left-tailed:** reject when $\bar{x}$ is *much less* than 66.7. Type I: declaring a decrease that didn't happen. Type II: missing a real decrease.
- **Two-tailed:** the deviation can go either way, so reject when $|\bar{x} - 66.7|$ is large. Type I: declaring a change when there was none. Type II: missing a real change (in either direction).

How much is "much greater"? That threshold is exactly what $\alpha$ — and the next video's p-value — will make precise.

## 3. Worked Examples

### Example 1: three questions, three formulations

- *"Has our new battery's lifetime improved beyond the old 40 hours?"* → $H_0: \mu = 40$, $H_1: \mu > 40$ — right-tailed.
- *"Has the factory's defect rate changed from 2%?"* → $H_0: p = 0.02$, $H_1: p \ne 0.02$ — two-tailed (either direction is news).
- *"Did the new routing cut mean delivery time below 3 days?"* → $H_0: \mu = 3$, $H_1: \mu < 3$ — left-tailed.

### Example 2: legal and illegal hypothesis statements

| statement | verdict |
|---|---|
| $H_0: \mu = 66.7$ | ✅ about the population parameter |
| $H_1: \bar{x} > 66.7$ | ❌ $\bar{x}$ is a sample quantity — evidence, not hypothesis |
| $H_1: \mu > 68.442$ | ❌ the observed value has snuck into the *claim* — hypotheses are written before data speaks |
| $H_0: S^2 = 9$ | ❌ sample statistic again; the population version $\sigma^2 = 9$ would be ✅ |

### Example 3: statistics for other parameters

Testing an occurrence rate (say, the fraction of spam): statistic $\hat{P}$, observed value $\hat{p} = 0.05$-ish. Testing a variance: statistic $S^2$ — or equivalently $\sum (X_i - \bar{X})^2$, which differs only by the constant factor $n - 1$; both carry identical information, illustrating that the *choice* of statistic is a modeling decision, not a law.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project), then the sample of ten heights (in inches, averaging the video's 68.442).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(70)

heights = np.array([64.12, 71.15, 66.41, 70.17, 73.13,
                    65.10, 69.16, 67.39, 71.97, 65.82])   # ten 18-year-olds, inches
MU_1970s = 66.7

observed_statistic = heights.mean()
print(f"n = {len(heights)} heights, observed statistic (sample mean): "
      f"{observed_statistic:.3f} inches")
print(f"the 1970s baseline (H0 value): {MU_1970s} inches")
print(f"difference: {observed_statistic - MU_1970s:+.3f} inches -- but is that enough?")
```

### The test statistic is a random variable

Before the data arrived, $\bar{X}$ had a whole distribution of possible values *under $H_0$* (take $\sigma = 3$ inches as known for now, so $\bar{X} \sim N(66.7, 3^2/10)$ if nothing changed). The observed 68.442 is one point on that landscape — the entire question is whether it's an unremarkable point or a remote one.

```python
SIGMA = 3.0
se = SIGMA / np.sqrt(len(heights))

simulated_xbars = rng.normal(MU_1970s, se, size=100_000)   # X-bar under H0

plt.figure(figsize=(9.5, 4.6))
plt.hist(simulated_xbars, bins=80, density=True, color="tab:green", alpha=0.55,
         label=r"test statistic $\bar{X}$ under $H_0$ ($\mu = 66.7$)")
plt.axvline(observed_statistic, color="tab:red", linewidth=2.5,
            label=f"observed statistic: {observed_statistic:.3f}")
plt.axvline(MU_1970s, color="black", linestyle=":", linewidth=1.2)
plt.xlabel(r"sample mean $\bar{x}$ (inches)")
plt.ylabel("density")
plt.title("One number meets one distribution: how unusual is 68.442 if nothing changed?")
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.show()

frac_beyond = (simulated_xbars >= observed_statistic).mean()
print(f"fraction of H0-world sample means at or beyond 68.442: {frac_beyond:.4f}")
print("(A number this small is begging for a name -- it gets one in the next video.)")
```

## 5. Visualizations

### The three tails, side by side

Same null distribution of $\bar{X}$, three different alternatives — and three different regions whose occupation would discredit $H_0$ (drawn here at $\alpha = 0.05$ just to make the shapes concrete).

```python
import sympy as sp
z95 = float(sp.sqrt(2) * sp.erfinv(sp.Float(0.90)))    # z for one 5% tail
z975 = float(sp.sqrt(2) * sp.erfinv(sp.Float(0.95)))   # z for two 2.5% tails

x_axis = np.linspace(MU_1970s - 4 * se, MU_1970s + 4 * se, 600)
density = np.exp(-0.5 * ((x_axis - MU_1970s) / se) ** 2) / (se * np.sqrt(2 * np.pi))

fig, axes = plt.subplots(1, 3, figsize=(13.5, 4.2), sharey=True)
setups = [
    (r"right-tailed: $H_1: \mu > 66.7$", x_axis >= MU_1970s + z95 * se,
     "reject for large $\\bar{x}$"),
    (r"left-tailed: $H_1: \mu < 66.7$", x_axis <= MU_1970s - z95 * se,
     "reject for small $\\bar{x}$"),
    (r"two-tailed: $H_1: \mu \neq 66.7$", np.abs(x_axis - MU_1970s) >= z975 * se,
     "reject for large $|\\bar{x} - 66.7|$"),
]
for ax, (title, region, caption) in zip(axes, setups):
    ax.plot(x_axis, density, color="black", linewidth=1.6)
    ax.fill_between(x_axis, density, where=region, color="tab:red", alpha=0.5)
    ax.fill_between(x_axis, density, where=~region, color="tab:green", alpha=0.25)
    ax.axvline(MU_1970s, color="black", linestyle=":", linewidth=1)
    ax.set_title(f"{title}\n{caption}", fontsize=10)
    ax.set_xlabel(r"$\bar{x}$ (inches)")
axes[0].set_ylabel("density under $H_0$")
plt.suptitle(r"One null, three alternatives: where the rejection region lives ($\alpha$ = 0.05 shown)")
plt.tight_layout()
plt.show()
```

### Same evidence, different verdicts

The observed 68.442 lands deep in the right tail — impressive for a right-tailed or two-tailed test, and *utterly uninteresting* to a left-tailed one. The tail must match the question, and it is chosen **before** looking at the data.

```python
fig, ax = plt.subplots(figsize=(10, 4.4))
ax.plot(x_axis, density, color="black", linewidth=1.6)
ax.fill_between(x_axis, density, where=x_axis >= MU_1970s + z95 * se,
                color="tab:red", alpha=0.5, label="right-tailed rejection region")
ax.fill_between(x_axis, density, where=x_axis <= MU_1970s - z95 * se,
                color="tab:blue", alpha=0.4, label="left-tailed rejection region")
ax.axvline(observed_statistic, color="tab:red", linewidth=2.5,
           label=f"observed 68.442")
ax.set_xlabel(r"$\bar{x}$ (inches)")
ax.set_ylabel("density under $H_0$")
ax.set_title("68.442 screams 'increase', says nothing about 'decrease'")
ax.grid(True, alpha=0.3)
ax.legend(fontsize=9)
plt.show()
```

```python
# Exercise 3: statistics and observed values
xbar = heights.mean()
s2 = heights.var(ddof=1)
ss = ((heights - xbar) ** 2).sum()
prop_tall = (heights > 70).mean()
print(f"(a) statistic X-bar, observed {xbar:.3f}")
print(f"(b) statistic S^2, observed {s2:.3f}; equally valid: sum of squared")
print(f"    deviations, observed {ss:.3f} (= (n-1) * S^2 -- same information)")
print(f"(c) statistic P-hat = fraction of sample > 70, observed {prop_tall:.1f}")
```

```python
# Exercise 4: the left-tailed delivery test, in words
print("Trigger: reject H0 when the sample mean delivery time is MUCH LESS than 3 days.")
print("Type I: concluding the routing sped things up when nothing changed --")
print("        celebrating (and maybe advertising) an improvement that isn't real.")
print("Type II: failing to detect a genuine speed-up.")
print("The company likely fears Type I more (false marketing claims, wasted")
print("rollout), so it sets a small alpha and demands strong evidence.")
```

```python
# Exercise 5: bias beats sample size
trials = 20_000
n5 = 30
trigger = MU_1970s + 1.645 * SIGMA / np.sqrt(n5)

honest = rng.normal(MU_1970s, SIGMA, size=(trials, n5))
false_alarm_honest = (honest.mean(axis=1) > trigger).mean()

groups = rng.normal(MU_1970s, SIGMA, size=(trials, n5, 4))
basketball = groups.max(axis=2)                     # tallest of each group of 4
false_alarm_biased = (basketball.mean(axis=1) > trigger).mean()

print(f"P(false 'heights increased!' | honest sampling)     = {false_alarm_honest:.1%}")
print(f"P(false 'heights increased!' | basketball sampling) = {false_alarm_biased:.1%}")
print("\nThe biased scheme rejects a TRUE null essentially always -- no sample")
print("size fixes a broken sampling procedure. Randomize first, then count.")
```

```python
# Exercise 6: right-tailed vs two-tailed power at mu = 68
trials = 50_000
n6 = 10
se6 = SIGMA / np.sqrt(n6)
xbars6 = rng.normal(68.0, se6, size=trials)

right_reject = (xbars6 > MU_1970s + 1.645 * se6).mean()
two_reject = (np.abs(xbars6 - MU_1970s) > 1.96 * se6).mean()
print(f"true mu = 68: right-tailed rejects {right_reject:.1%} of the time,")
print(f"              two-tailed rejects  {two_reject:.1%}")
print("\nThe right-tailed test concentrates its whole alpha budget on the side")
print("where the truth lies, so it detects the increase more often. But the tail")
print("was defensible only because the QUESTION was 'did heights increase' --")
print("choosing it after seeing x-bar = 68.4 would quietly double the real")
print("Type I rate (you'd have gone left had the data leaned left).")
```

## 8. Conclusion

- Before testing anything: the data must be **representative, randomized** (no basketball-team samples), and **big enough** ($n \ge 30$ as the rule of thumb).
- Hypotheses are statements about **population parameters** — $H_0: \mu = 66.7$, $H_1: \mu > 66.7$ — never about sample quantities.
- The **test statistic** is a random variable computed from the sample ($\bar{X}$ for means; $\hat{P}$ for rates; $S^2$ — among others — for variances); its realized value (68.442) is the **observed statistic**.
- Three alternatives, one baseline: **right-tailed** ($\mu >$), **left-tailed** ($\mu <$), **two-tailed** ($\mu \ne$) — rejecting for large $\bar{x}$, small $\bar{x}$, or large $|\bar{x} - \mu_0|$ respectively, with the two error types re-read accordingly. The tail comes from the *question*, chosen before the data speaks.
- The cliffhanger: 68.442 sits far into $H_0$'s right tail — only ~3% of null-world sample means reach it. Quantifying exactly that tail probability is the next video's job: the **p-value**.

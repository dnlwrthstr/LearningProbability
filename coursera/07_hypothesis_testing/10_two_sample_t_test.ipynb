# The Two-Sample t-Test: Comparing Two Populations

Every test so far interrogated one population; the questions people actually ask usually compare two. *Are 18-year-olds in the US taller than in Argentina?* Two samples — ten Americans ($\bar{x} = 68.442$, $s_x = 3.113$) and nine Argentinians ($\bar{y} = 65.949$, $s_y = 3.106$) — and a null hypothesis about the **difference**: $H_0: \mu_{US} - \mu_{Arg} = 0$. The machinery generalizes cleanly: $\bar{X} - \bar{Y}$ is Gaussian with standard deviation $\sqrt{\sigma_x^2/n_x + \sigma_y^2/n_y}$, plugging in the sample deviations gives a t-statistic, and the only ugliness is the **degrees of freedom** — a cumbersome formula best left to software, which here returns $\nu = 16.8$. The right-tailed verdict: $t_{obs} = 1.745$, $p = 0.0495$ — rejecting by three ten-thousandths. It is based on the tenth video of Lesson 8 (Hypothesis Testing).

## 1. Introduction

One-sample tests ask "did this population move from a known baseline?"; two-sample tests drop the baseline and ask the more natural question, "do these two populations differ *from each other*?" This notebook builds the two-sample t-test (the unequal-variances, or Welch, form the course uses), reruns the three tails, and lands on the thinnest rejection of the entire lesson.

After completing this notebook you will be able to:

- Formulate two-population hypotheses in terms of the difference $\mu_X - \mu_Y$ and classify the three tails.
- State the three assumptions: distinct individuals, independent measurements, normal populations.
- Derive the distribution of $\bar{X} - \bar{Y}$ and assemble the two-sample t-statistic.
- Explain what the Welch degrees of freedom are for — and let the software compute $\nu = 16.8$.
- Reproduce the video's verdicts: right-tailed $p = 0.0495$ (reject), two-tailed $p = 0.0991$ (don't) — and appreciate just how fragile the first one is.

## 2. Theory

### Two countries, two samples

You believe the people of one country are taller than another's — but you can only take samples. The data:

| group | $n$ | sample mean | sample sd |
|---|---|---|---|
| US, $X_1 \dots X_{10}$ | 10 | $\bar{x} = 68.442$ | $s_x = 3.113$ |
| Argentina, $Y_1 \dots Y_9$ | 9 | $\bar{y} = 65.949$ | $s_y = 3.106$ |

### Hypotheses about a difference

All three tests share the null that both population means are the same; writing everything in terms of $\mu_{US} - \mu_{Arg}$:

$$
H_0: \mu_{US} - \mu_{Arg} = 0 \qquad\text{vs}\qquad
H_1: \mu_{US} - \mu_{Arg} \;>\; 0 \;\;(\text{right}),\;\; <\; 0 \;\;(\text{left}),\;\; \ne\; 0 \;\;(\text{two-tailed})
$$

### Three assumptions

1. **Distinct individuals** — no one person appears in both groups.
2. **Independence** — each height measurement is independent of the others.
3. **Normal populations** — $X \sim N(\mu_{US}, \sigma_{US}^2)$ and $Y \sim N(\mu_{Arg}, \sigma_{Arg}^2)$.

### The distribution of $\bar{X} - \bar{Y}$

A linear combination of Gaussians is Gaussian (the Lesson 3 result), so

$$
\bar{X} - \bar{Y} \;\sim\; N\!\left(\mu_{US} - \mu_{Arg},\;\; \frac{\sigma_x^2}{n_x} + \frac{\sigma_y^2}{n_y}\right)
$$

— the mean is the difference of means, and the variance *adds* the two sample means' variances. Standardizing gives a standard normal — but neither $\sigma$ is known, so replace each with its sample estimate:

$$
T = \frac{(\bar{X} - \bar{Y}) - (\mu_{US} - \mu_{Arg})}{\sqrt{\dfrac{s_x^2}{10} + \dfrac{s_y^2}{9}}}
$$

This is again a t-statistic — a mean-type statistic for Gaussian data with unknown $\sigma$s.

### The degrees of freedom (let software do it)

The formula for this statistic's degrees of freedom is **cumbersome** — don't worry about memorizing it; software packages compute it for you. (For the curious, it's the Welch–Satterthwaite expression, written out in the worked examples.) Plugging in $n_x, n_y, s_x, s_y$ here gives

$$
\nu = 16.8
$$

— yes, degrees of freedom need not be whole numbers.

### The right-tailed test

$H_0: \mu_{US} - \mu_{Arg} = 0$, $H_1: > 0$, $\alpha = 0.05$. Under $H_0$ the statistic reduces to $(\bar{X} - \bar{Y})$ over the standard error, following $t_{16.8}$:

$$
t_{obs} = \frac{68.442 - 65.949}{\sqrt{\frac{3.113^2}{10} + \frac{3.106^2}{9}}} = 1.745,
\qquad p = P(T > t_{obs}) = 0.0495
$$

$0.0495 < 0.05$: **reject** — accept that the US population mean is greater than Argentina's. (By a margin of half a thousandth in p — hold that thought for the exercises.)

### The two-tailed test

Same data, same statistic; only the p-value changes: $p = P(|T| > t_{obs}) = 0.0991 > 0.05$ — **not enough evidence** that the means differ. The one-sample pattern repeats: the directional question squeaks through, the non-directional one doesn't.

## 3. Worked Examples

### Example 1: the standard error and the statistic

$$
SE = \sqrt{\frac{3.113^2}{10} + \frac{3.106^2}{9}} = \sqrt{0.969 + 1.072} = 1.429,
\qquad t_{obs} = \frac{2.493}{1.429} = 1.745
$$

### Example 2: the cumbersome formula, once

The Welch–Satterthwaite degrees of freedom:

$$
\nu \;=\; \frac{\left(\frac{s_x^2}{n_x} + \frac{s_y^2}{n_y}\right)^2}
{\frac{(s_x^2/n_x)^2}{n_x - 1} + \frac{(s_y^2/n_y)^2}{n_y - 1}}
\;=\; \frac{(2.041)^2}{\frac{0.969^2}{9} + \frac{1.072^2}{8}} = 16.8
$$

Notice $\nu$ lands *between* $\min(n_x, n_y) - 1 = 8$ and $n_x + n_y - 2 = 17$ — it always does; unequal spreads and sizes drag it below the naive 17.

### Example 3: the three verdicts

Right-tailed: $p = 0.0495$ → reject. Two-tailed: $0.0991$ → don't. Left-tailed ($H_1: \mu_{US} < \mu_{Arg}$): $p = P(T < 1.745) = 0.9505$ — nowhere close, as expected when the observed difference points the other way. The tail must be chosen by the question, *before* the data — by now a familiar refrain, and never more consequential than at $p = 0.0495$.

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project). The t-machinery accepts fractional degrees of freedom for free — the density formula only involves gamma functions.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import gamma, sqrt, pi

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(10)

def t_pdf(t, df):
    """Density of the Student t-distribution (df may be fractional)."""
    coeff = gamma((df + 1) / 2) / (sqrt(df * pi) * gamma(df / 2))
    return coeff * (1 + np.asarray(t, dtype=float) ** 2 / df) ** (-(df + 1) / 2)

def t_tail(t_value, df):
    """P(T > t_value) for T ~ t_df, by numeric integration."""
    grid = np.linspace(0, abs(t_value), 20_000)
    cdf_half = np.trapezoid(t_pdf(grid, df), grid)
    tail = 0.5 - cdf_half
    return tail if t_value >= 0 else 1 - tail

def two_sample_t_test(x, y, tail="two", alpha=0.05, label=""):
    """Welch two-sample t-test on raw data arrays."""
    nx, ny = len(x), len(y)
    vx, vy = np.var(x, ddof=1) / nx, np.var(y, ddof=1) / ny
    se = np.sqrt(vx + vy)
    t_obs = (np.mean(x) - np.mean(y)) / se
    df = (vx + vy) ** 2 / (vx**2 / (nx - 1) + vy**2 / (ny - 1))
    p = {"right": t_tail(t_obs, df),
         "left": 1 - t_tail(t_obs, df),
         "two": 2 * t_tail(abs(t_obs), df)}[tail]
    verdict = "REJECT H0" if p < alpha else "fail to reject H0"
    print(f"{label}: t_obs = {t_obs:.3f}, df = {df:.1f}, p = {p:.4f} -> {verdict}")
    return t_obs, df, p
```

### The two samples, and the video's tests

```python
us_heights = np.array([64.12, 71.15, 66.41, 70.17, 73.13,
                       65.10, 69.16, 67.39, 71.97, 65.82])
arg_heights = np.array([64.75, 66.17, 71.61, 68.64, 61.82,
                        66.67, 64.84, 67.12, 61.92])

print(f"US:        n = {len(us_heights)}, x-bar = {us_heights.mean():.3f}, "
      f"s = {us_heights.std(ddof=1):.3f}")
print(f"Argentina: n = {len(arg_heights)}, y-bar = {arg_heights.mean():.3f}, "
      f"s = {arg_heights.std(ddof=1):.3f}")
print(f"observed difference: {us_heights.mean() - arg_heights.mean():.3f} inches\n")

t_r, df_w, p_r = two_sample_t_test(us_heights, arg_heights, "right",
                                   label="right-tailed (US taller)")
t_2, _, p_2 = two_sample_t_test(us_heights, arg_heights, "two",
                                label="two-tailed   (US different)")
```

### The Welch degrees of freedom, unpacked

The cumbersome formula in the open — and a check that $\nu$ always lands between $\min(n)-1$ and $n_x + n_y - 2$.

```python
vx = us_heights.var(ddof=1) / len(us_heights)
vy = arg_heights.var(ddof=1) / len(arg_heights)
df_manual = (vx + vy) ** 2 / (vx**2 / 9 + vy**2 / 8)
print(f"variance of X-bar: sx^2/10 = {vx:.4f}")
print(f"variance of Y-bar: sy^2/9  = {vy:.4f}")
print(f"Welch-Satterthwaite: nu = ({vx:.3f} + {vy:.3f})^2 / "
      f"({vx:.3f}^2/9 + {vy:.3f}^2/8) = {df_manual:.2f}")
print(f"bounds: min(n)-1 = 8  <=  {df_manual:.1f}  <=  nx+ny-2 = 17  (always true)")
```

### Is $t_{16.8}$ really the right curve? A simulation check

Draw both groups from the *same* normal population (so $H_0$ is true), compute the Welch statistic thousands of times, and compare its histogram — and its realized Type I rate — with the $t_{16.8}$ approximation.

```python
trials = 60_000
sim_x = rng.normal(67, 3.11, size=(trials, 10))
sim_y = rng.normal(67, 3.11, size=(trials, 9))
svx = sim_x.var(axis=1, ddof=1) / 10
svy = sim_y.var(axis=1, ddof=1) / 9
sim_t = (sim_x.mean(axis=1) - sim_y.mean(axis=1)) / np.sqrt(svx + svy)

t_axis = np.linspace(-5, 5, 500)
plt.figure(figsize=(10, 4.6))
plt.hist(sim_t, bins=150, range=(-5, 5), density=True, color="tab:green",
         alpha=0.5, label="simulated Welch statistics under $H_0$")
plt.plot(t_axis, t_pdf(t_axis, 16.8), color="tab:red", linewidth=2,
         label=r"$t_{16.8}$ density")
plt.xlabel("t")
plt.ylabel("density")
plt.title("The Welch approximation earns its keep")
plt.grid(True, alpha=0.3)
plt.legend(fontsize=9)
plt.show()

t_crit = 1.7396          # ~ t quantile 0.95 at df = 16.8
print(f"realized Type I rate of 'reject when t > {t_crit}': "
      f"{(sim_t > t_crit).mean():.3f} (target 0.05)")
```

## 5. Visualizations

### The data themselves

Nineteen people, two countries — the entire evidence base of the test.

```python
fig, ax = plt.subplots(figsize=(9.5, 4))
for i, (data, label, color) in enumerate([
        (us_heights, "US (n = 10)", "tab:blue"),
        (arg_heights, "Argentina (n = 9)", "tab:orange")]):
    jitter = rng.normal(0, 0.04, size=len(data))
    ax.scatter(data, np.full(len(data), i) + jitter, s=60, alpha=0.7,
               color=color, edgecolor="white")
    ax.plot([data.mean()] * 2, [i - 0.2, i + 0.2], color=color, linewidth=3)
    ax.annotate(f"mean {data.mean():.2f}", xy=(data.mean(), i + 0.27),
                ha="center", fontsize=10, color=color)
ax.set_yticks([0, 1])
ax.set_yticklabels(["US", "Argentina"])
ax.set_xlabel("height (inches)")
ax.set_ylim(-0.6, 1.6)
ax.set_title("Two samples, overlapping spreads, a 2.49-inch gap in means")
ax.grid(True, alpha=0.3, axis="x")
plt.tight_layout()
plt.show()
```

### The two p-values on the $t_{16.8}$ curve

```python
density = t_pdf(t_axis, df_w)
fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.2), sharey=True)
panels = [
    (f"right-tailed: p = {p_r:.4f} -> REJECT (barely)", t_axis >= t_r, [t_r]),
    (f"two-tailed: p = {p_2:.4f} -> don't reject", np.abs(t_axis) >= t_r,
     [-t_r, t_r]),
]
for ax, (title, region, marks) in zip(axes, panels):
    ax.plot(t_axis, density, color="black", linewidth=1.6)
    ax.fill_between(t_axis, density, where=region, color="tab:red", alpha=0.55)
    for m in marks:
        ax.axvline(m, color="tab:red", linewidth=1.6)
    ax.axvline(0, color="black", linestyle=":", linewidth=1)
    ax.set_title(title, fontsize=10)
    ax.set_xlabel("t")
axes[0].set_ylabel(r"density under $H_0$: $t_{16.8}$")
plt.suptitle(f"Same statistic t = {t_r:.3f}, two questions, two verdicts")
plt.tight_layout()
plt.show()
```

## 7. Solutions

```python
# Exercise 1: the left-tailed test
_ = two_sample_t_test(us_heights, arg_heights, "left",
                      label="left-tailed (US shorter)")
print("\nWith x-bar ABOVE y-bar, the observed statistic is positive, and nearly")
print("the whole t-curve sits below it: p ~ 0.95. A left-tailed alternative can")
print("only be supported by a negative observed difference -- this one points the")
print("other way, so the test was doomed from the summary statistics alone.")
```

```python
# Exercise 2: assumption breaks
print("(a) breaks assumption 1 (distinct individuals): the groups share people,")
print("    so the two sample means are no longer independent quantities.")
print("(b) breaks assumption 2 (independence): family heights are correlated --")
print("    nine relatives carry far less information than nine random citizens.")
print("(c) breaks assumption 3 (normal populations), and n = 6 is far too small")
print("    for the CLT to patch it.")
```

```python
# Exercise 3: Argentina at n = 30 with the same summary stats
sx, sy = 3.113, 3.106
xb, yb = 68.442, 65.949
vx3, vy3 = sx**2 / 10, sy**2 / 30
se3 = np.sqrt(vx3 + vy3)
t3 = (xb - yb) / se3
df3 = (vx3 + vy3) ** 2 / (vx3**2 / 9 + vy3**2 / 29)
p3_right = t_tail(t3, df3)
print(f"SE: {se3:.4f} (was 1.4286), t_obs = {t3:.3f} (was 1.745)")
print(f"df = {df3:.1f} (was 16.8)")
print(f"right-tailed p = {p3_right:.4f}, two-tailed p = {2 * p3_right:.4f}")
print("\nA subtlety: the standard error shrank a lot (more Argentine data), but")
print("the Welch df actually DIPPED to 15.4 -- the two groups' variance")
print("contributions (0.97 vs 0.32) are now more unbalanced. The SE gain")
print("dominates by far: even the two-tailed test now rejects comfortably.")
```

```python
# Exercise 4: the equal-groups collapse
n_eq, s_eq = 10, 3.0
v = s_eq**2 / n_eq
df_eq = (v + v) ** 2 / (v**2 / (n_eq - 1) + v**2 / (n_eq - 1))
print(f"equal n and s: nu = {df_eq:.1f} = 2(n-1) = {2 * (n_eq - 1)}")
print("\nAlgebra: (2v)^2 / (2 v^2/(n-1)) = 4v^2 (n-1) / (2v^2) = 2(n-1).")
print("Our groups had nearly equal s but unequal n (10 vs 9), so nu = 16.8 --")
print("just below the balanced maximum nx + ny - 2 = 17. Imbalance in n or s")
print("always drags nu down toward min(n) - 1.")
```

```python
# Exercise 5: how thin is the ice?
t_crit = 1.7396
se_obs = np.sqrt(us_heights.var(ddof=1) / 10 + arg_heights.var(ddof=1) / 9)
diff_star = t_crit * se_obs
diff_obs = us_heights.mean() - arg_heights.mean()
print(f"smallest rejecting difference: {t_crit} * {se_obs:.4f} = {diff_star:.4f} in")
print(f"observed difference:            {diff_obs:.4f} in")
print(f"margin: {diff_obs - diff_star:.4f} inches -- about {2.54 * (diff_obs - diff_star) * 10:.1f} mm")
print("\nThe rejection hangs on roughly a fifth of a millimeter of average height.")
print("'Significant at 0.05' and 'p = 0.0496' are technically the same claim, but")
print("only the second lets a reader see how binary luck shaped the verdict.")
```

```python
# Exercise 6: naive df = nx + ny - 2 -- harmless here, harmful elsewhere
p_welch = t_tail(t_r, df_w)
p_naive = t_tail(t_r, 17)
print(f"this dataset: p with Welch df 16.8 = {p_welch:.4f}, "
      f"with naive df 17 = {p_naive:.4f} -- nearly identical.")

# a hostile scenario: tiny noisy group vs large quiet group
trials = 40_000
gx = rng.normal(0, 10, size=(trials, 4))       # n = 4, sigma = 10
gy = rng.normal(0, 1, size=(trials, 40))       # n = 40, sigma = 1
wvx = gx.var(axis=1, ddof=1) / 4
wvy = gy.var(axis=1, ddof=1) / 40
tw = (gx.mean(axis=1) - gy.mean(axis=1)) / np.sqrt(wvx + wvy)
df_true = (wvx + wvy) ** 2 / (wvx**2 / 3 + wvy**2 / 39)

naive_crit = 1.6820                             # t quantile 0.95, df = 42
print(f"\nhostile design (n = 4, s = 10 vs n = 40, s = 1):")
print(f"mean Welch df: {df_true.mean():.1f}  (naive df would claim 42)")
print(f"Type I rate with naive df-42 threshold: {(tw > naive_crit).mean():.3f} "
      f"(target 0.05)")
print("The naive df pretends the big quiet group contributes information about")
print("the small noisy one's mean; Welch's formula (df ~ 3) charges honestly.")
```

## 8. Conclusion

- Two-sample testing compares two populations through the **difference of sample means**: $H_0: \mu_X - \mu_Y = 0$ with the usual three tails, under three assumptions — distinct individuals, independence, normal populations.
- $\bar{X} - \bar{Y}$ is Gaussian with variance $\sigma_x^2/n_x + \sigma_y^2/n_y$; plugging in sample deviations yields a **t-statistic** whose degrees of freedom come from the cumbersome Welch–Satterthwaite formula — software's job; here $\nu = 16.8$ (fractional is fine, and always between $\min(n)-1$ and $n_x + n_y - 2$).
- The video's verdicts: right-tailed $t_{obs} = 1.745$, $p = 0.0495$ → reject that the US and Argentine means are equal, in favor of *US taller* — by half a thousandth; two-tailed $p = 0.0991$ → not enough evidence of *any* difference.
- The exercises' sharpest finding: the rejection hangs on ~0.2 mm of average height — one more argument for reporting p-values, not just verdicts — and more Argentine data ($n_y = 30$) would let even the two-tailed test reject comfortably.
- Welch's df matters little for balanced designs but protects $\alpha$ fiercely in hostile ones (a tiny noisy group vs a large quiet one).

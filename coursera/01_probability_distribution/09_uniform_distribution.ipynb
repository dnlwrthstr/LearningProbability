# The Uniform Distribution

This notebook introduces the simplest continuous distribution: the **uniform distribution**, where every value in an interval is equally likely. It derives the PDF and CDF from first principles and simulates the classic waiting-for-the-bus example. It is based on the ninth video of Lesson 2 (Probability Distributions).


## 1. Introduction

A bus passes every 10 minutes but you don't know the schedule, so you walk out at a random moment. Sometimes you wait 1 minute, sometimes 5, or 9, or 3.52... the wait times spread **homogeneously** over $[0, 10]$ — no value or window is favored. That is the uniform distribution.

After completing this notebook you will be able to:

- Write the uniform PDF $f(x) = \frac{1}{b-a}$ and explain where the constant comes from.
- Describe the effect of the two parameters $a$ and $b$.
- Derive and use the flat-diagonal-flat CDF $F(x) = \frac{x-a}{b-a}$.


## 2. Theory

### The PDF must be constant — and area fixes its height

Suppose a call center answers at any time between 0 and 15 minutes **with equal probability** (after 15 the call disconnects). No value is favored, so the PDF must be **constant** on $[0, 15]$. Which constant? The area under the PDF must be 1:

$$
\text{height} \times 15 = 1 \quad\Rightarrow\quad \text{height} = \frac{1}{15} \approx 0.067
$$

### The general form

$T \sim \text{Uniform}(a, b)$ has **two parameters** — the endpoints of the interval:

$$
f(x) = \begin{cases} \dfrac{1}{b-a} & a \leq x \leq b \\[2mm] 0 & \text{otherwise} \end{cases}
$$

The wider the interval, the **lower** the height (the same total area of 1 is spread thinner); shrink the interval and the height shoots up.

### The CDF: flat, diagonal, flat

For $\text{Uniform}(0, 1)$:

- $x < 0$: no area gathered yet → $F(x) = 0$;
- $0 \leq x \leq 1$: the area is a rectangle of base $x$ and height 1 → $F(x) = x$ (a diagonal of slope 1);
- $x > 1$: all the area is gathered → $F(x) = 1$.

For general $(a, b)$, the diagonal becomes the fraction of the interval covered:

$$
F(x) = \begin{cases} 0 & x < a \\ \dfrac{x-a}{b-a} & a \leq x \leq b \\ 1 & x > b \end{cases}
$$


## 3. Worked Examples

### Example 1: The call center, Uniform(0, 15)

- $f(x) = \frac{1}{15}$ on $[0, 15]$.
- $P(T \leq 5) = \frac{5 - 0}{15} = \frac{1}{3}$.
- $P(5 \leq T \leq 10) = \frac{10-5}{15} = \frac{1}{3}$ — every 5-minute window carries the same probability.

### Example 2: The bus, Uniform(0, 10)

- $P(\text{wait} \leq 2) = \frac{2}{10} = 0.2$.
- $P(\text{wait} > 8) = 1 - F(8) = 1 - 0.8 = 0.2$ — by symmetry, same as waiting under 2.


## 4. Python Examples


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=42)

def uniform_pdf(x, a, b):
    return np.where((x >= a) & (x <= b), 1 / (b - a), 0.0)

def uniform_cdf(x, a, b):
    return np.clip((x - a) / (b - a), 0, 1)

print(f"Call center Uniform(0,15): height = {1/15:.3f}")
print(f"P(T <= 5)       = {uniform_cdf(5, 0, 15):.3f}")
print(f"P(5 <= T <= 10) = {uniform_cdf(10, 0, 15) - uniform_cdf(5, 0, 15):.3f}")
print(f"Bus Uniform(0,10): P(wait > 8) = {1 - uniform_cdf(8, 0, 10):.3f}")
```

### The video's 200 call measurements


```python
waits = rng.uniform(0, 15, size=200)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].scatter(range(200), waits, s=12, color="#4e79a7", alpha=0.8)
axes[0].set_title("200 recorded wait times: homogeneous over [0, 15]")
axes[0].set_xlabel("Call number")
axes[0].set_ylabel("Wait (min)")

axes[1].hist(waits, bins=15, density=True, color="#4e79a7", edgecolor="white")
axes[1].axhline(1 / 15, color="red", linestyle="--", label="PDF height 1/15")
axes[1].set_title("Histogram ≈ flat (deviations expected: only 200 calls)")
axes[1].set_xlabel("Wait (min)")
axes[1].set_ylabel("Density")
axes[1].legend()

for ax in axes:
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Visualizations

### The parameters: wider interval, lower height

The video's animation as a static frame: all three rectangles have area 1.


```python
xs = np.linspace(-1, 16, 800)

plt.figure(figsize=(9, 5))
for a, b, color in [(0, 15, "#4e79a7"), (2, 8, "#f28e2b"), (5, 7, "#e15759")]:
    plt.plot(xs, uniform_pdf(xs, a, b), color=color, linewidth=2,
             label=f"Uniform({a}, {b}): height 1/{b-a} = {1/(b-a):.3f}")
plt.title("Same area (1), different intervals: shorter interval → taller PDF")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

### PDF and CDF together: flat → diagonal → flat


```python
a, b = 0, 15
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot(xs, uniform_pdf(xs, a, b), color="#f28e2b", linewidth=2)
x0 = 5
mask = (xs >= a) & (xs <= x0)
axes[0].fill_between(xs[mask], uniform_pdf(xs[mask], a, b),
                     color="#4e79a7", alpha=0.5)
axes[0].set_title(f"PDF: area up to {x0} = {x0}/15 = {x0/15:.3f}")
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")

axes[1].plot(xs, uniform_cdf(xs, a, b), color="#e15759", linewidth=2)
axes[1].plot([x0, x0], [0, x0 / 15], "k--", linewidth=1)
axes[1].plot([-1, x0], [x0 / 15, x0 / 15], "k--", linewidth=1)
axes[1].plot(x0, x0 / 15, "ko")
axes[1].set_title("CDF: flat at 0, diagonal (x-a)/(b-a), flat at 1")
axes[1].set_xlabel("x")
axes[1].set_ylabel("F(x)")

for ax in axes:
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 7. Solutions


```python
# Exercise 1
print(f"1.  height = 1/6 = {1/6:.3f};  P(X<=5) = (5-2)/6 = {3/6:.3f}")

# Exercise 2
print(f"2.  P(4<=X<=6) = 2/10 = {2/10:.1f}")

# Exercise 3: conditional on X > 5, the remaining wait is Uniform(5, 10)
print(f"3.  P(X<=7 | X>5) = (7-5)/(10-5) = {2/5:.1f} — the remaining wait "
      "is Uniform(5,10)")

# Exercise 4
print("4.  (m-a)/(b-a) = 0.5  ->  m = (a+b)/2, the midpoint of the interval.")

# Exercise 5
u = rng.random(100_000)
x = 2 + (8 - 2) * u

plt.figure(figsize=(8, 4))
plt.hist(x, bins=60, density=True, color="#4e79a7", edgecolor="white",
         label="transformed samples")
grid = np.linspace(1, 9, 400)
plt.plot(grid, uniform_pdf(grid, 2, 8), "r--", linewidth=2,
         label="Uniform(2,8) PDF")
plt.title("a + (b-a)·U reproduces Uniform(a, b)")
plt.xlabel("x")
plt.ylabel("Density")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
print("5.  Histogram sits on the flat PDF at height 1/6 ✓")
```

## 8. Conclusion

- $\text{Uniform}(a, b)$: every value in $[a, b]$ equally likely; PDF is the constant $\frac{1}{b-a}$, forced by total area 1.
- Two parameters — the interval endpoints; widening the interval flattens the PDF.
- CDF: $0$, then the diagonal $\frac{x-a}{b-a}$ (the fraction of the interval covered), then $1$.
- Interval probabilities are proportional to interval lengths — the defining property, and the reason uniform random numbers are the raw material for simulating everything else.

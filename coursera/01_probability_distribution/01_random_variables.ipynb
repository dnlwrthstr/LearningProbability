# Random Variables

## 1. Introduction

In algebra, a variable has one fixed value: $x = 3$, always. **Random variables** are different — they can take many values:

- the temperature tomorrow;
- the number of heads in 10 coin tosses;
- the wait time until the next bus.

After completing this notebook you will be able to:

- Define a random variable and contrast it with a deterministic one.
- Distinguish a *specific outcome sequence* from the *grouped event* $X = k$.
- Estimate a random variable's probabilities with a histogram of repeated experiments.
- Tell **discrete** random variables from **continuous** ones — by countability, not finiteness.


## 2. Theory

### From experiment to number

A random variable attaches a **number** to each outcome of an experiment. Flip one coin and let $X$ = the number of heads:

- outcome heads → $X = 1$, with $P(X = 1) = 0.5$;
- outcome tails → $X = 0$, with $P(X = 0) = 0.5$.

Half of the time $X$ is 1, half of the time it is 0 (roughly).

### Grouping outcomes: $X$ = heads in 10 tosses

Each *specific sequence* of 10 tosses has probability $\left(\frac{1}{2}\right)^{10}$ — all-heads is $\left(\frac{1}{2}\right)^{10}$, and a sequence with nine heads and one tail is $\left(\frac{1}{2}\right)^9 \cdot \frac{1}{2}$, the same number.

But $P(X = 9)$ is a **harder question**: it groups *all* the different sequences with exactly nine heads. Only $P(X = 10)$ (and $P(X=0)$) equals a single-sequence probability, because there is just one way to get all heads.

### Any assignment works

You can define any random variable you like — e.g. $X = 1$ with probability 0.5, $X = -7$ with probability 0.2, and $X = 3.14159$ with probability 0.3 — as long as the probabilities **add to 1**.

### Discrete vs. continuous

| Discrete | Continuous |
|---|---|
| number of heads in 10 flips | wait time until the next bus |
| number of 1s when rolling dice | height of a gymnast's jump |
| number of sick patients | millimeters of rain in November |

The real distinction is **countability**, not finiteness. Flipping a coin *until* the first heads can take 1, 2, 3, ... , a million tries — infinitely many values, yet still discrete, because they can be put in a list. A continuous random variable takes values on an **entire interval** (1 minute, 1.01, 1.001, ...), which cannot be listed.

### Deterministic vs. random

Variables in algebra and calculus ($x = 2$, the input of $f(x) = x^2$) are **deterministic** — once defined, fixed forever, associated with a certain outcome. Random variables are associated with an **uncertain** outcome.


## 3. Worked Example: the 10-toss experiment

Run the 10-coin-toss experiment many times: you might get $X = 5$, then $X = 6$, then $X = 5$ again... The possible values are $0$ through $10$.

Repeating it 500 times and drawing a **histogram** gives a good *estimate* of the probabilities (not the exact values — it's an experiment):

- $X = 0$ and $X = 10$ are rarest — 10 heads or 10 tails is very unlikely (each is $\frac{1}{1024}$).
- $X = 5$ is the most frequent — roughly half heads is the typical result.


## 4. Python Examples


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=42)
```

### One coin: the simplest random variable


```python
flips = rng.integers(0, 2, size=10)   # X for ten separate one-flip experiments
print("Ten realizations of X (number of heads in ONE flip):", flips)
print("A deterministic variable, for contrast: x = 3, always 3.")
```

### The video's experiment: 500 runs of 10 tosses


```python
n_runs, n_tosses = 500, 10
tosses = rng.integers(0, 2, size=(n_runs, n_tosses))
x = tosses.sum(axis=1)                 # X = number of heads, once per run

print("First 12 realizations of X:", x[:12])
print(f"Most common value: {np.bincount(x, minlength=11).argmax()}")
print(f"P(X=10) estimated: {(x == 10).mean():.4f}   exact: {0.5**10:.4f}")
```

### A custom random variable

Any value list works, as long as the probabilities sum to 1.


```python
values = [1, -7, 3.14159]
probs = [0.5, 0.2, 0.3]
print("Probabilities sum to:", sum(probs))

samples = rng.choice(values, size=10, p=probs)
print("Ten realizations:", samples)
```

## 5. Visualizations

### The histogram of X (10 tosses, 500 runs)


```python
counts = np.bincount(x, minlength=11)

plt.figure(figsize=(9, 5))
plt.bar(range(11), counts / n_runs, color="#4e79a7", edgecolor="white")
plt.title(f"X = number of heads in 10 tosses — histogram of {n_runs} runs")
plt.xlabel("Value of X")
plt.ylabel("Relative frequency")
plt.xticks(range(11))
plt.grid(True, alpha=0.3, axis="y")
plt.show()
```

### Discrete vs. continuous, side by side

Left: a discrete random variable's values sit on isolated points (they can be listed). Right: a continuous one — the wait time for a bus — fills an entire interval; any value like 3.7208... minutes can occur.


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# Discrete: number of heads in 10 tosses
axes[0].bar(range(11), counts / n_runs, color="#4e79a7", edgecolor="white")
axes[0].set_title("Discrete: values you can list (0, 1, ..., 10)")
axes[0].set_xlabel("Number of heads")
axes[0].set_xticks(range(11))

# Continuous: bus wait times (uniform-ish between 0 and 10 minutes)
waits = rng.uniform(0, 10, size=5_000)
axes[1].hist(waits, bins=60, density=True, color="#f28e2b", edgecolor="white")
axes[1].set_title("Continuous: values fill an interval (wait time)")
axes[1].set_xlabel("Minutes until the next bus")

for ax in axes:
    ax.set_ylabel("Relative frequency")
    ax.grid(True, alpha=0.3, axis="y")
plt.tight_layout()
plt.show()
```

## 8. Conclusion

- A **random variable** attaches numbers to uncertain outcomes; unlike an algebra variable, it takes different values on different runs of the experiment.
- $P(X = k)$ groups *all* outcome sequences with the same count — usually many more than one.
- Histograms of repeated experiments estimate a random variable's probabilities; 500 runs of 10 tosses already show the peak at 5 and the rare tails at 0 and 10.
- **Discrete** = values can be listed (even if the list is infinite); **continuous** = values fill an interval.
- Next: describing *all* of a random variable's probabilities at once — its **probability distribution**.

# The Bernoulli Distribution

This short notebook introduces the **Bernoulli distribution** — the distribution of a single success/failure trial, and the building block from which the binomial distribution is made. It is based on the fifth video of Lesson 2 (Probability Distributions).


## 1. Introduction

Flip one coin and let $X$ = the number of heads: $X = 1$ with probability 0.5, $X = 0$ with probability 0.5. Call heads a **success** and tails a **failure**. That's the whole experiment — and its distribution has a name.

The same pattern appears everywhere:

| Experiment | Success ($X=1$) | Failure ($X=0$) | $p$ |
|---|---|---|---|
| Coin flip | heads | tails | 0.5 |
| Die roll, counting 1s | rolls a 1 | anything else | 1/6 |
| Patient status | sick* | healthy | $p$ |

*"Success = sick" sounds contradictory, but since $X$ counts sick patients, $X = 1$ is the success of the count.

After completing this notebook you will be able to:

- Write the Bernoulli PMF and its single parameter $p$.
- Recognize Bernoulli trials inside larger experiments.
- Connect Bernoulli to the binomial distribution.


## 2. Theory

$X \sim \text{Bernoulli}(p)$ takes exactly two values:

$$
p(1) = p, \qquad p(0) = 1 - p
$$

- **One parameter**: $p$, the probability of success.
- Both PMF requirements hold: values are non-negative, and $p + (1-p) = 1$.
- A compact way to write the PMF in one line: $p(x) = p^x (1-p)^{1-x}$ for $x \in \{0, 1\}$.

### Relation to the binomial

A Bernoulli variable is a binomial with a single trial: $\text{Bernoulli}(p) = \text{Binomial}(1, p)$. Conversely, summing $n$ independent Bernoulli trials gives $\text{Binomial}(n, p)$ — the binomial *is* stacked Bernoullis.


## 3. Python Examples


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(seed=42)

def bernoulli_pmf(x, p):
    return p ** x * (1 - p) ** (1 - x)

for name, p in [("coin", 0.5), ("die counting 1s", 1/6), ("sick patient", 0.1)]:
    print(f"{name:18s} p(1) = {bernoulli_pmf(1, p):.4f}, "
          f"p(0) = {bernoulli_pmf(0, p):.4f}, sum = 1.0")
```

### Bernoullis stack into a binomial

Ten independent Bernoulli(0.5) trials, summed, give exactly the 10-toss binomial from earlier notebooks.


```python
n_runs = 100_000
bernoullis = rng.random((n_runs, 10)) < 0.5   # 10 Bernoulli trials per run
x = bernoullis.sum(axis=1)                     # their sum

from math import comb
print("k   sum-of-Bernoullis   Binomial(10, 0.5)")
for k in [0, 3, 5, 7, 10]:
    exact = comb(10, k) * 0.5 ** 10
    print(f"{k:2d}       {(x == k).mean():.4f}          {exact:.4f}")
```

## 4. Visualization

The three Bernoulli examples from the video, side by side — same two bars, different parameter.


```python
fig, axes = plt.subplots(1, 3, figsize=(11, 3.6), sharey=True)

for ax, (name, p) in zip(axes, [("Coin: p = 0.5", 0.5),
                                ("Die (ones): p = 1/6", 1/6),
                                ("Sick patient: p = 0.1", 0.1)]):
    ax.bar([0, 1], [1 - p, p], color=["#9ecae1", "#e15759"],
           edgecolor="white", width=0.5)
    ax.set_title(name)
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["0 (failure)", "1 (success)"])
    ax.grid(True, alpha=0.3, axis="y")

axes[0].set_ylabel("p(x)")
plt.suptitle("Bernoulli(p): one distribution, one parameter")
plt.tight_layout()
plt.show()
```

## 5. Exercises

1. A basketball player makes 78% of free throws. Model one free throw as a Bernoulli variable: what are $p(1)$ and $p(0)$?
2. Write $\text{Bernoulli}(p)$ as a binomial. What are its parameters?
3. An email is spam with probability 0.2 (the prior from Lesson 1). If you receive 40 independent emails and count the spam ones, what distribution does the count follow?


## 6. Solutions

1. $p(1) = 0.78$, $p(0) = 0.22$.
2. $\text{Bernoulli}(p) = \text{Binomial}(n = 1, p)$ — one trial.
3. A sum of 40 Bernoulli(0.2) trials: $\text{Binomial}(40, 0.2)$.


## 7. Conclusion

- $\text{Bernoulli}(p)$ is the distribution of one success/failure trial: $p(1) = p$, $p(0) = 1 - p$ — a single parameter.
- "Success" just means *the thing being counted*, even when it's a sick patient.
- Bernoulli is the atom of discrete probability: $n$ independent Bernoullis sum to a $\text{Binomial}(n, p)$.

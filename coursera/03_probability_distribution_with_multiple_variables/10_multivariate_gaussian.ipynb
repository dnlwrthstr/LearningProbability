# The Multivariate Gaussian Distribution

The Gaussian doesn't stop at one variable. In two variables it is a literal **bell over the plane**; in $n$ it is the workhorse of machine learning. Everything from this module assembles into its formula: the **mean vector** $\boldsymbol\mu$ replaces $\mu$, the **covariance matrix** $\Sigma$ replaces $\sigma^2$, and the density reads $f(\mathbf{x}) = \frac{1}{(2\pi)^{n/2}|\Sigma|^{1/2}} e^{-\frac12 (\mathbf{x}-\boldsymbol\mu)^T \Sigma^{-1} (\mathbf{x}-\boldsymbol\mu)}$. Independent variables give circular bells; covariance stretches them into tilted ellipses. It is based on the tenth video of Lesson 4 (Probability Distributions with Multiple Variables).

## 1. Introduction

Recall the one-variable Gaussian PDF with its parameters $\mu$ (center of the bell) and $\sigma$ (spread):

$$
f_X(x) = \frac{1}{\sqrt{2\pi}\,\sigma}\, e^{-\frac12 \left(\frac{x-\mu}{\sigma}\right)^2}
$$

What if you have more than one variable — $x_1, x_2, \dots, x_n$? Take **height** $H$ (inches) and **weight** $W$ (pounds) of adults: in a dataset of thousands of people, *each marginal* is Gaussian. What does the **joint** look like? That is the multivariate Gaussian — and its shape depends on exactly the object from the last notebooks: the covariance matrix.

After completing this notebook you will be able to:

- Write the joint PDF of independent Gaussians as a product and recognize its circular bell.
- Follow the vectorization: sum of squared exponents → $(\mathbf{x}-\boldsymbol\mu)^T \Sigma^{-1} (\mathbf{x}-\boldsymbol\mu)$ with a diagonal $\Sigma$.
- State the general multivariate Gaussian formula and map each piece to its univariate counterpart.
- Connect covariance to geometry: level-curve circles vs. tilted ellipses, and $|\Sigma|$ as the volume of the spread.

## 2. Theory

### Independent case: a product of bells

If $H$ and $W$ were independent, the joint PDF would be the product of the marginal PDFs:

$$
f_{HW}(h, w) = \frac{1}{2\pi\,\sigma_H \sigma_W}\,
\exp\!\left(-\frac12\left[\frac{(h-\mu_H)^2}{\sigma_H^2} + \frac{(w-\mu_W)^2}{\sigma_W^2}\right]\right)
$$

Plotted, this is a perfectly **symmetric** bell: seen from the top, its level curves are (axis-aligned) circles/ellipses.

But real height–weight data doesn't look like that: the density is **elongated along a line with positive slope** — taller people tend to weigh more. The level curves (the mountain seen from the top: green curves high up, purple low) are *tilted ellipses*. The deformation is caused by the **covariance** between the variables.

### Vectorizing the independent case

Rewrite the exponent's sum of squares as a vector norm:

$$
\frac{(h-\mu_H)^2}{\sigma_H^2} + \frac{(w-\mu_W)^2}{\sigma_W^2}
= (\mathbf{x} - \boldsymbol\mu)^T
\begin{bmatrix} \sigma_H^2 & 0 \\ 0 & \sigma_W^2 \end{bmatrix}^{-1}
(\mathbf{x} - \boldsymbol\mu)
$$

where $\mathbf{x} = \begin{bmatrix} h \\ w \end{bmatrix}$ and $\boldsymbol\mu = \begin{bmatrix} \mu_H \\ \mu_W \end{bmatrix}$ is the **mean vector**. The diagonal matrix in the middle is exactly the **covariance matrix** of independent variables (covariances 0 off the diagonal). And the normalizing product $\sigma_H \sigma_W$ is $\sqrt{|\Sigma|}$ — the square root of the determinant. So the independent joint is

$$
f(\mathbf{x}) = \frac{1}{2\pi\,|\Sigma|^{1/2}}\, e^{-\frac12 (\mathbf{x}-\boldsymbol\mu)^T \Sigma^{-1} (\mathbf{x}-\boldsymbol\mu)}
$$

### The general formula

Here is the payoff: **this expression stays valid in the dependent case** — the only change is that $\Sigma$ is no longer diagonal; its off-diagonal entries are the covariances. For $n$ variables:

$$
f(\mathbf{x}) = \frac{1}{(2\pi)^{n/2}\,|\Sigma|^{1/2}}\,
\exp\!\left(-\frac12 (\mathbf{x}-\boldsymbol\mu)^T \Sigma^{-1} (\mathbf{x}-\boldsymbol\mu)\right)
$$

### Univariate ↔ multivariate dictionary

| univariate | multivariate | role |
|---|---|---|
| $x$ | $\mathbf{x}$ (random vector) | the variable(s) |
| $\mu$ | $\boldsymbol\mu$ (mean vector) | center of the bell |
| $\sigma^2$ | $\Sigma$ (covariance matrix) | the variation |
| $\frac{1}{\sqrt{2\pi}}$ | $\frac{1}{(2\pi)^{n/2}}$ | accounts for $n$ variables |
| $\frac{1}{\sigma}$ | $\frac{1}{|\Sigma|^{1/2}}$ | $|\Sigma|$ = volume of the spread (bigger determinant, bigger spread) |
| $\left(\frac{x-\mu}{\sigma}\right)^2$ | $(\mathbf{x}-\boldsymbol\mu)^T\Sigma^{-1}(\mathbf{x}-\boldsymbol\mu)$ | standardized squared distance |

Every scalar became a vector, and the scalar variance became the covariance matrix — $\Sigma^{-1}$ standardizes and rescales the variances *and* covariances at once.

## 3. Worked Examples

### Example 1: the standard bivariate Gaussian at its peak

$\boldsymbol\mu = \mathbf{0}$, $\Sigma = I$ (independent, unit variances): $|\Sigma| = 1$ and the exponent at $\mathbf{x} = \mathbf{0}$ is 0, so

$$
f(\mathbf{0}) = \frac{1}{2\pi} \approx 0.159
$$

### Example 2: a correlated 2×2 covariance matrix, by hand

$\Sigma = \begin{bmatrix} 1 & 0.8 \\ 0.8 & 1 \end{bmatrix}$: determinant $|\Sigma| = 1 - 0.64 = 0.36$, inverse $\Sigma^{-1} = \frac{1}{0.36}\begin{bmatrix} 1 & -0.8 \\ -0.8 & 1 \end{bmatrix}$. The peak height is $\frac{1}{2\pi\sqrt{0.36}} = \frac{1}{2\pi \cdot 0.6} \approx 0.265$ — *taller* than the independent bell, because the determinant (spread volume) shrank: the mass squeezed onto the diagonal ridge.

### Example 3: height and weight

$\boldsymbol\mu = (67, 150)$, $\sigma_H = 3$ in, $\sigma_W = 20$ lb, correlation $0.75$ → $\mathrm{Cov} = 0.75 \cdot 3 \cdot 20 = 45$:

$$
\Sigma = \begin{bmatrix} 9 & 45 \\ 45 & 400 \end{bmatrix}, \qquad |\Sigma| = 3600 - 2025 = 1575
$$

The level curves are ellipses tilted along the "taller ⇒ heavier" direction, and each marginal is still exactly Gaussian ($N(67, 9)$ and $N(150, 400)$).

## 4. Python Examples

First the imports (with the numpy 2.x compatibility shim used throughout this project).

```python
import numpy as np
import matplotlib.pyplot as plt

# numpy < 2 fallback: np.trapezoid was called np.trapz
if not hasattr(np, "trapezoid"):
    np.trapezoid = np.trapz

rng = np.random.default_rng(42)

def multivariate_gaussian_pdf(x, mu, sigma):
    n = len(mu)
    diff = x - mu
    exponent = -0.5 * diff @ np.linalg.inv(sigma) @ diff
    norm = (2 * np.pi) ** (n / 2) * np.sqrt(np.linalg.det(sigma))
    return np.exp(exponent) / norm
```

### The formula from scratch, checked piece by piece

```python
# Example 1: standard bivariate at the peak
mu0 = np.zeros(2)
identity = np.eye(2)
print(f"standard bivariate f(0,0) = {multivariate_gaussian_pdf(mu0, mu0, identity):.4f}"
      f"   (1/(2 pi) = {1 / (2 * np.pi):.4f})")

# Example 2: the correlated matrix by hand
sigma_c = np.array([[1.0, 0.8], [0.8, 1.0]])
print(f"\ncorrelated: det = {np.linalg.det(sigma_c):.2f} (by hand: 0.36)")
print(f"peak height = {multivariate_gaussian_pdf(mu0, mu0, sigma_c):.4f}"
      f"   (1/(2 pi * 0.6) = {1 / (2 * np.pi * 0.6):.4f})")
print("Smaller determinant -> smaller spread volume -> taller, thinner bell.")
```

### For a diagonal $\Sigma$, the formula IS the product of marginals

```python
sigma_h, sigma_w = 3.0, 20.0
mu = np.array([67.0, 150.0])
sigma_diag = np.array([[sigma_h**2, 0], [0, sigma_w**2]])

def univariate(x, m, s):
    return np.exp(-0.5 * ((x - m) / s)**2) / (s * np.sqrt(2 * np.pi))

point = np.array([70.0, 160.0])
joint = multivariate_gaussian_pdf(point, mu, sigma_diag)
product = univariate(point[0], 67, sigma_h) * univariate(point[1], 150, sigma_w)
print(f"matrix formula:       {joint:.8f}")
print(f"product of marginals: {product:.8f}")
print("Independent variables: the multivariate formula factors exactly.")
```

### Sampling height and weight, dependent vs independent

`rng.multivariate_normal` takes exactly $(\boldsymbol\mu, \Sigma)$. The sample statistics recover the inputs — and both clouds have the *same marginals*.

```python
rng = np.random.default_rng(42)

corr = 0.75
cov_hw = corr * sigma_h * sigma_w                       # 45
sigma_dep = np.array([[sigma_h**2, cov_hw], [cov_hw, sigma_w**2]])

dep = rng.multivariate_normal(mu, sigma_dep, size=8000)
indep = rng.multivariate_normal(mu, sigma_diag, size=8000)

print(f"input Sigma (dependent):\n{sigma_dep}")
print(f"\nsample mean:  {dep.mean(axis=0).round(2)}   (input {mu})")
print(f"sample Sigma:\n{np.cov(dep.T, ddof=0).round(1)}")
print(f"\ndeterminants: dependent {np.linalg.det(sigma_dep):.0f} < "
      f"independent {np.linalg.det(sigma_diag):.0f}")
```

### The marginals stay Gaussian either way

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
grid_h = np.linspace(55, 79, 300)
grid_w = np.linspace(80, 220, 300)

axes[0].hist(dep[:, 0], bins=50, density=True, alpha=0.5, color="tab:blue",
             label="heights (dependent cloud)")
axes[0].plot(grid_h, univariate(grid_h, 67, 3), color="black", linewidth=2,
             label="N(67, 3²)")
axes[0].set_title("marginal of height")
axes[0].set_xlabel("height (in)")

axes[1].hist(dep[:, 1], bins=50, density=True, alpha=0.5, color="tab:orange",
             label="weights (dependent cloud)")
axes[1].plot(grid_w, univariate(grid_w, 150, 20), color="black", linewidth=2,
             label="N(150, 20²)")
axes[1].set_title("marginal of weight")
axes[1].set_xlabel("weight (lb)")

for ax in axes:
    ax.set_ylabel("density")
    ax.grid(True, alpha=0.3)
    ax.legend()
plt.suptitle("Marginals alone cannot reveal the covariance")
plt.tight_layout()
plt.show()
```

## 5. Visualizations

### The two bells in 3D: symmetric vs. elongated

```python
h = np.linspace(58, 76, 90)
w = np.linspace(90, 210, 90)
H, W = np.meshgrid(h, w)

def density_grid(sigma):
    inv = np.linalg.inv(sigma)
    det = np.linalg.det(sigma)
    dh, dw = H - 67, W - 150
    expo = -0.5 * (inv[0, 0] * dh**2 + 2 * inv[0, 1] * dh * dw + inv[1, 1] * dw**2)
    return np.exp(expo) / (2 * np.pi * np.sqrt(det))

fig = plt.figure(figsize=(13, 5))
for k, (sig, title) in enumerate([(sigma_diag, "independent: symmetric bell"),
                                  (sigma_dep, "correlated: bell stretched along a ridge")]):
    ax = fig.add_subplot(1, 2, k + 1, projection="3d")
    ax.plot_surface(H, W, density_grid(sig), cmap="viridis")
    ax.set_title(title)
    ax.set_xlabel("height (in)")
    ax.set_ylabel("weight (lb)")
plt.tight_layout()
plt.show()
```

### The mountain from the top: level curves

The video's key picture — circles (axis-aligned ellipses) for independence, tilted ellipses when the covariance kicks in, with the sample cloud underneath.

```python
fig, axes = plt.subplots(1, 2, figsize=(12.5, 5.2), sharey=True)
for ax, sig, data, title in [(axes[0], sigma_diag, indep, "independent"),
                             (axes[1], sigma_dep, dep, "correlated (Cov = 45)")]:
    ax.scatter(data[:, 0], data[:, 1], s=3, alpha=0.15, color="gray")
    cs = ax.contour(H, W, density_grid(sig), levels=7, cmap="viridis_r")
    ax.set_title(f"{title}: level curves of the density")
    ax.set_xlabel("height (in)")
    ax.grid(True, alpha=0.3)
axes[0].set_ylabel("weight (lb)")
plt.suptitle("Covariance tilts and stretches the level curves")
plt.tight_layout()
plt.show()
```

### Sweeping the correlation: from circle to line

The same marginals throughout — only the off-diagonal of $\Sigma$ changes. As $\rho \to 1$ the ellipse collapses toward a line and $|\Sigma| \to 0$.

```python
fig, axes = plt.subplots(1, 4, figsize=(15, 4), sharey=True)
zg = np.linspace(-3.2, 3.2, 120)
Z1, Z2 = np.meshgrid(zg, zg)

for ax, r in zip(axes, [0.0, 0.5, 0.9, 0.99]):
    sig = np.array([[1, r], [r, 1]])
    inv = np.linalg.inv(sig)
    det = np.linalg.det(sig)
    expo = -0.5 * (inv[0, 0] * Z1**2 + 2 * inv[0, 1] * Z1 * Z2 + inv[1, 1] * Z2**2)
    ax.contourf(Z1, Z2, np.exp(expo) / (2 * np.pi * np.sqrt(det)),
                levels=12, cmap="viridis")
    ax.set_title(f"$\\rho$ = {r}   $|\\Sigma|$ = {det:.2f}")
    ax.set_xlabel("$x_1$")
axes[0].set_ylabel("$x_2$")
plt.suptitle("Growing correlation squeezes the bell onto its ridge (determinant -> 0)")
plt.tight_layout()
plt.show()
```

```python
# Exercise 1: standard bivariate evaluations
for pt in [np.array([1.0, 0.0]), np.array([1.0, 1.0])]:
    val = multivariate_gaussian_pdf(pt, np.zeros(2), np.eye(2))
    print(f"f({pt[0]:.0f}, {pt[1]:.0f}) = {val:.4f}")
print(f"check: e^(-1/2)/(2 pi) = {np.exp(-0.5) / (2 * np.pi):.4f}")
```

```python
# Exercise 2: 2x2 determinant and inverse by hand
S = np.array([[4.0, 2.0], [2.0, 3.0]])
# det = 4*3 - 2*2 = 8; inverse = (1/8) [[3, -2], [-2, 4]]
print(f"Exercise 2: det = {np.linalg.det(S):.0f} (by hand 8)")
print(f"inverse:\n{np.linalg.inv(S)}")
print(f"(1/8)[[3,-2],[-2,4]] = \n{np.array([[3, -2], [-2, 4]]) / 8}")
print(f"peak height = 1/(2 pi sqrt(8)) = {1 / (2 * np.pi * np.sqrt(8)):.4f}")
```

```python
# Exercise 3: n = 1 reduction
sigma1 = 2.5
grid = np.linspace(-8, 8, 200)
multi = np.array([multivariate_gaussian_pdf(np.array([x]), np.array([0.0]),
                                            np.array([[sigma1**2]])) for x in grid])
uni = univariate(grid, 0, sigma1)
print(f"Exercise 3: max |multivariate(n=1) - univariate| = {np.abs(multi - uni).max():.2e}")
print("With n = 1: (2 pi)^(1/2) = sqrt(2 pi), |Sigma|^(1/2) = sigma, and the")
print("exponent is ((x - mu)/sigma)^2 -- the familiar formula, letter for letter.")
```

```python
# Exercise 4: standardizing yields the correlation matrix
rng = np.random.default_rng(0)
cloud = rng.multivariate_normal(mu, sigma_dep, size=50_000)
z = (cloud - cloud.mean(axis=0)) / cloud.std(axis=0)
print(f"Exercise 4: covariance of standardized data:\n{np.cov(z.T, ddof=0).round(3)}")
print("Diagonal 1's, off-diagonal ~0.75 = rho: the correlation matrix.")
print("Geometrically, standardizing removed the units (height in inches vs weight")
print("in pounds), making both spreads 1 -- the ellipse keeps its tilt but its")
print("axes are now comparable.")
```

```python
# Exercise 5: a conditional slice is Gaussian (and narrower)
rng = np.random.default_rng(0)
cloud = rng.multivariate_normal(mu, sigma_dep, size=2_000_000)
in_slice = np.abs(cloud[:, 0] - 70) < 0.1
weights = cloud[in_slice, 1]

cond_mu = 150 + 0.75 * (20 / 3) * (70 - 67)
cond_sd = 20 * np.sqrt(1 - 0.75**2)
grid_w = np.linspace(80, 220, 300)

plt.figure(figsize=(9, 4.5))
plt.hist(weights, bins=60, density=True, alpha=0.5, color="tab:blue",
         label=f"weights of ~70-inch people (n = {in_slice.sum()})")
plt.plot(grid_w, univariate(grid_w, cond_mu, cond_sd), color="black", linewidth=2,
         label=f"theory N({cond_mu:.0f}, {cond_sd:.1f}²)")
plt.plot(grid_w, univariate(grid_w, 150, 20), color="tab:red", linestyle="--",
         label="marginal N(150, 20²)")
plt.title("Conditioning on height: Gaussian again, shifted and narrower")
plt.xlabel("weight (lb)")
plt.ylabel("density")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print(f"Exercise 5: conditional mean {cond_mu:.0f} (marginal 150), "
      f"conditional sd {cond_sd:.1f} (marginal 20)")
print("Knowing someone is tall raises the expected weight AND cuts the")
print("uncertainty by the factor sqrt(1 - rho^2) = 0.66.")
```

## 8. Conclusion

- The **multivariate Gaussian** extends the bell to $n$ variables: $f(\mathbf{x}) = \frac{1}{(2\pi)^{n/2}|\Sigma|^{1/2}} \exp\big({-\tfrac12}(\mathbf{x}-\boldsymbol\mu)^T\Sigma^{-1}(\mathbf{x}-\boldsymbol\mu)\big)$ — scalars became vectors, the variance became the **covariance matrix**.
- For **independent** variables $\Sigma$ is diagonal and the formula factors into the product of marginal bells (a symmetric mountain, circular level curves); **covariance** stretches the bell along a ridge and tilts the level-curve ellipses — exactly the height–weight picture.
- The dictionary: $\boldsymbol\mu$ = center; $\Sigma^{-1}$ standardizes and rescales variances and covariances; $|\Sigma|$ = the *volume* of the spread (0.36 < 1 for $\rho = 0.8$: taller, thinner bell; $|\Sigma| \to 0$ as $\rho \to 1$: collapse onto a line).
- Structure carries over beautifully: marginals of a multivariate Gaussian are Gaussian, and conditional slices are Gaussian too — shifted by the correlation and *narrower* by $\sqrt{1-\rho^2}$.
- This distribution — parameterized entirely by $(\boldsymbol\mu, \Sigma)$ — appears everywhere in machine learning: Gaussian classifiers, mixtures, processes, Kalman filters.
